# 数据表格

学习目标：能把行列数据组织成表格，正确设置跨格与表头关联，并区分结构语义和视觉样式。

前置知识：HTML 嵌套、id/class 属性、相对路径和浏览器开发者工具。

适用范围：WHATWG HTML Living Standard；使用现代浏览器查看配套页面。本章是社区工具站的教学数据，不是实际业务记录；页面不含 JavaScript。

环境准备：[环境配置与运行](README.md)。

配套脚本：位于 scripts/09-data-tables/。

1. [basic.html](scripts/09-data-tables/basic.html)、[index.html](scripts/09-data-tables/index.html)：最小表格、完整库存表及导航。
2. [columns.html](scripts/09-data-tables/columns.html)：周末借还、列组及两层表头。
3. [groups.html](scripts/09-data-tables/groups.html)：分站库存、跨行与行组边界。
4. [complex.html](scripts/09-data-tables/complex.html)：月份、工具、站点的复杂表头关联。
5. [implicit.html](scripts/09-data-tables/implicit.html)：省略行组标签，比较源码和元素树。
6. [styles.css](scripts/09-data-tables/styles.css)：边框、间距、对齐与列背景。

## 1 标签总览

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;table&gt; | 数据表格 | 组织二维数据 |
| &lt;caption&gt; | 表格标题 | 说明整张表的主题 |
| &lt;tr&gt; | 表格行 | 容纳一行单元格 |
| &lt;th&gt; | 表头单元格 | 标识行、列或分组的含义 |
| &lt;td&gt; | 数据单元格 | 保存具体数据 |
| &lt;thead&gt; | 头部行组 | 组织字段标题所在的行 |
| &lt;tbody&gt; | 主体行组 | 组织数据记录所在的行 |
| &lt;tfoot&gt; | 尾部行组 | 组织汇总等内容 |
| &lt;colgroup&gt; | 列组 | 定义一组列 |
| &lt;col&gt; | 列 | 描述一列或连续多列 |

## 2 &lt;table&gt;：表达行列关系

工具名称与可借数量是行列数据，适合表格。导航与正文的左右排布应使用 CSS 布局，不用数据表格摆放页面。

&lt;caption&gt; 是 &lt;table&gt; 的直接子元素，放在其他表格结构之前，一张表至多一个。它说明整表；&lt;th&gt; 标识表头，&lt;td&gt; 保存数据。单元格放在 &lt;tr&gt; 内，不能直接放在 &lt;table&gt; 下。

```html
<table>
  <caption>可借工具摘录（件）</caption>
  <tbody>
    <tr><th scope="col">工具</th><th scope="col">可借数量</th></tr>
    <tr><th scope="row">卷尺</th><td>4</td></tr>
  </tbody>
</table>
<!-- 4 表示卷尺可借 4 件；表格标题说明主题与单位。 -->
```

配套文件：[scripts/09-data-tables/basic.html](scripts/09-data-tables/basic.html) · [浏览器预览](http://127.0.0.1:8009/scripts/09-data-tables/basic.html)

scope 用在 &lt;th&gt; 上，明确表头的作用范围：

- col：同列数据的表头。
- row：同行数据的表头。
- colgroup：对应列组的表头，第 4 节介绍。
- rowgroup：对应行组的表头，第 5 节介绍。

把数据加粗不能代替 &lt;th&gt;；给 &lt;td&gt; 加 scope 也不会使它成为表头。省略 scope 时浏览器按上下文确定关联，本章显式书写，方便检查。

## 3 &lt;thead&gt;、&lt;tbody&gt;、&lt;tfoot&gt;：组织行

先读懂每个数字属于哪一行、哪一列，再把完整库存表分成头部、主体与尾部。行组组织结构，表头关联解释数值，两者各有用途：

- &lt;thead&gt;：头部行组，每表至多一个。
- &lt;tbody&gt;：主体行组，可以有多个。
- &lt;tfoot&gt;：尾部行组，每表至多一个；没有汇总等内容时无需添加。

行组内放 &lt;tr&gt;，其中再放单元格。&lt;tbody&gt; 可以包含行表头，&lt;thead&gt; 不会把内部 &lt;td&gt; 自动变成 &lt;th&gt;。表格的头尾与页面 &lt;header&gt;、&lt;footer&gt; 不同。

下面截取完整库存表的主体与尾部，前面已有表格标题和列标题。

![卷尺行和借出列共同指向数值 2，表达卷尺借出两件。](image/illustration/09-01-table-header-relations.svg)

图 1：库存表中的双向表头关系。图只截取主体相关记录；行组边界不决定某格是不是表头。

下面只截取 tbody 与 tfoot；向前对照已有的列标题，再检查卷尺一行的 scope="row"，不要把 tbody 误当成“只能放 td”。

```html
<tbody>
  <tr><th scope="row">卷尺</th><td>4</td><td>2</td></tr>
  <tr><th scope="row">钳子</th><td>3</td><td>1</td></tr>
  <tr><th scope="row">手锯</th><td>2</td><td>0</td></tr>
</tbody>
<tfoot>
  <tr><th scope="row">合计</th><td>9</td><td>3</td></tr>
</tfoot>
<!-- 检查：卷尺的 2 同时对应“卷尺”和“借出”，不能只读出数字。 -->
<!-- 合计为 4+3+2=9、2+1+0=3；修改记录时必须同步重新计算。 -->
```

配套文件：[scripts/09-data-tables/index.html](scripts/09-data-tables/index.html) · [浏览器预览](http://127.0.0.1:8009/scripts/09-data-tables/index.html)

本例从前到后的结构是 &lt;caption&gt;、可选 &lt;colgroup&gt;、可选 &lt;thead&gt;、&lt;tbody&gt;、可选 &lt;tfoot&gt;。HTML 不会自动求和；新增记录后需要自己更新合计。

## 4 列组与跨列：组织两层表头

周末借还表有五列：工具、周六借出、周六归还、周日借出、周日归还。这里数量单位是件次，记录一段时间内的借还次数，不是库存快照。

在 &lt;caption&gt; 之后、行组之前声明列组：

```html
<colgroup><col class="tool-column"></colgroup>
<colgroup span="2" class="saturday-columns"></colgroup>
<colgroup span="2" class="sunday-columns"></colgroup>
```

配套文件：[scripts/09-data-tables/columns.html](scripts/09-data-tables/columns.html) · [浏览器预览](http://127.0.0.1:8009/scripts/09-data-tables/columns.html)

- span：&lt;col&gt; 或 &lt;colgroup&gt; 覆盖的列数，默认一列。本例后两组各为两列。
- &lt;colgroup&gt; 可以直接写 span；或不写 span，在内部放 &lt;col&gt;。
- &lt;col&gt; 是空元素，也可用 span 描述多列；它不容纳单元格。

class 只供样式选取列组。这里的 span 属性与通用容器 &lt;span&gt; 是不同概念。

### 4.1 colspan 与 rowspan：单元格占位

列组定义列的分组；单元格的跨格属性决定占据多少格位。格位是行列交叉的位置，跨格仍只有一个单元格元素。

- colspan="2"：当前单元格横跨两列。
- rowspan="2"：当前单元格占当前行及下一行。

```html
<thead>
  <tr>
    <th rowspan="2" scope="col">工具</th>
    <th colspan="2" scope="colgroup">周六</th>
    <th colspan="2" scope="colgroup">周日</th>
  </tr>
  <tr>
    <th scope="col">借出</th><th scope="col">归还</th>
    <th scope="col">借出</th><th scope="col">归还</th>
  </tr>
</thead>
```

配套文件：[scripts/09-data-tables/columns.html](scripts/09-data-tables/columns.html) · [浏览器预览](http://127.0.0.1:8009/scripts/09-data-tables/columns.html)

“周六”和“周日”的 scope="colgroup" 分别对应前面定义的两个列组，表头必须位于对应列组内。第二层“借出”“归还”各用 scope="col"。

第一层占位为 1+2+2=5；第二层第一列已被“工具”占用，只写剩下四个表头，不再补空单元格。

## 5 跨行与行组边界

跨格后的单元格必须覆盖整个网格且不重叠。检查一行时，既数本行单元格，也算入上方延伸下来的格位。

- colspan：整数 1～1000；colspan="0" 不表示跨到末列。
- rowspan：整数 0～65534；两种属性省略时都按 1 处理。
- rowspan="0"：从当前行延伸到当前行组末尾。
- 单元格不能跨越两个行组，例如不能从一个 &lt;tbody&gt; 跨到另一个。

下面是南站的完整行组，scope="rowgroup" 说明站点表头关联该组记录，scope="row" 说明工具表头关联当前记录。

```html
<tbody>
  <tr>
    <th rowspan="0" scope="rowgroup">南站</th>
    <th scope="row">卷尺</th><td>2</td><td>1</td>
  </tr>
  <tr><th scope="row">钳子</th><td>5</td><td>0</td></tr>
</tbody>
```

配套文件：[scripts/09-data-tables/groups.html](scripts/09-data-tables/groups.html) · [浏览器预览](http://127.0.0.1:8009/scripts/09-data-tables/groups.html)

同页北站使用 rowspan="2"。两组现在各有两条记录；增加第三条时，北站需调整固定跨行数，南站的 0 仍延伸到本组末尾。已被跨格覆盖的位置不要再补空 &lt;td&gt;。

## 6 headers 与 id：明确复杂表头关联

当一个数值同时属于月份、工具和站点时，可用 headers 明确列出表头。先给同一张表中的 &lt;th&gt; 设置文档内唯一的 id。

下面三个列标题位于月度表的 &lt;thead&gt; 内：

```html
<thead>
  <tr>
    <th id="tool-name">工具</th>
    <th id="station-north">北站</th>
    <th id="station-south">南站</th>
  </tr>
</thead>
```

配套文件：[scripts/09-data-tables/complex.html](scripts/09-data-tables/complex.html) · [浏览器预览](http://127.0.0.1:8009/scripts/09-data-tables/complex.html)

随后六月组的开头如下；完整文件还包含六月钳子和七月记录。

```html
<tr><th id="month-june" colspan="3">六月</th></tr>
<tr>
  <th id="june-tape" headers="month-june tool-name">卷尺</th>
  <td headers="month-june june-tape station-north">12</td>
  <td headers="month-june june-tape station-south">9</td>
</tr>
```

配套文件：[scripts/09-data-tables/complex.html](scripts/09-data-tables/complex.html) · [浏览器预览](http://127.0.0.1:8009/scripts/09-data-tables/complex.html)

- id：这里的 month-june、june-tape、station-north 是作者为“六月”“卷尺”“北站”选择的标识符。
- headers：用空格分隔互不重复的 id，每个目标都必须是同一张表的 &lt;th&gt;。不加 #，不用逗号，也不填写 class。
- headers 也能用于 &lt;th&gt;，但不能直接或间接引用自身。

12 的 headers 明确列出六月、卷尺和北站；这些值的书写次序不保证所有屏幕阅读器按相同次序朗读。月份行的颜色或 colspan 也不自动建立这些关联。

复杂表旁应简述组织方式；若不需要合表，可拆成每月一张简单表。不是所有两层表头都必须改用 headers。

## 7 CSS 辅助阅读，语义由 HTML 表达

边框、间距、对齐让格位更清楚，但不建立表头关联。下面的 CSS 合并相邻边框：

```css
table {
  border-collapse: collapse;
  margin: 16px 0;
}
```

配套文件：[scripts/09-data-tables/styles.css](scripts/09-data-tables/styles.css) · [浏览器预览](http://127.0.0.1:8009/scripts/09-data-tables/columns.html)

&lt;col&gt;、&lt;colgroup&gt; 只支持部分样式，例如列背景；它们不是单元格的父元素，不能靠在列上设置 text-align 让单元格继承。本例直接设置 &lt;td&gt; 的对齐，上层行或单元格的背景会遮住列背景。

在开发者工具 Elements 中临时删除样式 &lt;link&gt;，比较外观和 &lt;th&gt;、scope、headers；刷新即可恢复。查看 Accessibility 中的角色与名称，再逐个数值核对表头。原生表格语义优先，开发者工具不能代替屏幕阅读器的实际体验检查。

## 8 源码与隐式 &lt;tbody&gt;（补充）

下面特意省略 &lt;tbody&gt; 的开始和结束标签。浏览器解析这段 HTML 时会生成 &lt;tbody&gt;，把两行放在它里面；这是本例允许的标签省略情形。

```html
<table id="implicit-table">
  <caption>可借工具摘录（件）</caption>
  <tr><th scope="col">工具</th><th scope="col">可借数量</th></tr>
  <tr><th scope="row">卷尺</th><td>4</td></tr>
</table>
<!-- 预期：浏览器 Elements 中，两个 tr 共享自动生成的 tbody 父元素。 -->
<!-- 源码没有行组标签，DOM 却有；本页没有脚本，也不会凭空出现 thead。 -->
```

配套文件：[scripts/09-data-tables/implicit.html](scripts/09-data-tables/implicit.html) · [浏览器预览](http://127.0.0.1:8009/scripts/09-data-tables/implicit.html)

用“查看网页源代码”与 Elements 对照：源码文本与解析后的 DOM 元素树不总是一一对应。日常显式写出行组更容易看清结构；省略行组也不会自动把 &lt;td&gt; 改为 &lt;th&gt;。

## 本章小结

- &lt;table&gt; 表达数据，&lt;caption&gt; 说明整表，&lt;th&gt; 标识表头，行组组织记录。
- scope 明确行、列和分组的关系；复杂关联可用 headers 引用同表 &lt;th&gt; 的 id。
- 列组描述分组，colspan、rowspan 描述单元格占位；跨行不能越过行组。
- 配色和边框不代替语义。任取一个数值，应能解释它的记录、字段和单位，并找到 HTML 中的关联。
