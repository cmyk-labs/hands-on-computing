# SVG 与 MathML 嵌入

学习目标：能在 HTML 中嵌入 SVG 图形和基础 MathML 公式，区分展示尺寸、坐标范围与可访问信息。

前置知识：HTML 元素与属性、相对路径、简单 CSS 选择器和浏览器开发者工具。

适用范围：WHATWG HTML Living Standard、SVG 2 与 MathML Core 的基础页面集成；使用支持原生 MathML 的现代浏览器，不展开完整绘图和公式排版课程。

环境准备：[环境配置与运行](README.md)。

配套脚本：位于 scripts/15-svg-and-mathml/。

1. [basics.html](scripts/15-svg-and-mathml/basics.html)：图形入口、数学标记、分数与上下标的最小示例。
2. [shapes.svg](scripts/15-svg-and-mathml/shapes.svg)：矩形、圆、线段和三角形组成的独立图像。
3. [viewbox.html](scripts/15-svg-and-mathml/viewbox.html)：同一坐标下的展示尺寸与宽高比对照。
4. [index.html](scripts/15-svg-and-mathml/index.html)：组合外部图片、内联图形与公式。
5. [styles.css](scripts/15-svg-and-mathml/styles.css)：页面排版、视口边框和内联圆形的配色。

## 1 在 HTML 中加入图形与公式

SVG（Scalable Vector Graphics，可缩放矢量图形）用形状和坐标描述图像；MathML（Mathematical Markup Language，数学标记语言）组织公式的排版结构。HTML 负责把它们放进页面内容。

内联（inline）指直接在当前 HTML 中写入标记。HTML 解析器通过 &lt;svg&gt;、&lt;math&gt; 入口识别对应命名空间；命名空间用于区分元素所属的标记语言，内部元素仍分别属于 SVG、MathML。

下面 &lt;circle&gt; 画圆，cx、cy 指圆心坐标，r 指半径；&lt;mn&gt; 表示数字。

```html
<p>一个圆点：
  <svg width="60" height="40" role="img">
    <title>圆点</title>
    <circle cx="30" cy="20" r="12" fill="#b45309"></circle>
  </svg>
</p>
<p>一个数字：<math><mn>3</mn></math></p>
<!-- 圆由 SVG 绘制，数字由 MathML 排版，外层段落仍属于 HTML。 -->
```

配套文件：[scripts/15-svg-and-mathml/basics.html](scripts/15-svg-and-mathml/basics.html) · [浏览器预览](http://127.0.0.1:8015/scripts/15-svg-and-mathml/basics.html#embedding)

## 2 用 &lt;img&gt; 引用外部 SVG

已有 SVG 文件时，在 &lt;img&gt; 的 src 中写入地址。以下 shapes.svg 与页面同目录，width、height 设置图片的展示宽高。

alt 按图片在当前页面的用途提供替代文本。外部文件即使已有 &lt;title&gt; 和 &lt;desc&gt;，也不能代替 &lt;img&gt; 的 alt；纯装饰图片可使用空 alt。本例图形承担信息，因此描述了形状的位置。

```html
<img id="external-card" src="shapes.svg" width="240" height="120"
  role="img" alt="浅色矩形中，圆在左、竖线在中、朝右三角形在右。">
```

配套文件：[scripts/15-svg-and-mathml/index.html](scripts/15-svg-and-mathml/index.html) · [浏览器预览](http://127.0.0.1:8015/scripts/15-svg-and-mathml/index.html)

通过 &lt;img&gt; 加载时，SVG 作为图片使用，其内部脚本和链接交互不作为活动文档运行，也不能依赖它加载外部样式等资源。这些限制针对图片场景，不能推广到直接打开的 SVG 文档。

## 3 用基本图形组成 SVG

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;svg&gt; | SVG 根元素 | 建立图形视口 |
| &lt;rect&gt; | 矩形 | 用位置与宽高定义矩形 |
| &lt;circle&gt; | 圆 | 用圆心与半径定义圆 |
| &lt;line&gt; | 线段 | 连接两个端点 |
| &lt;polygon&gt; | 多边形 | 连接顶点并闭合 |
| &lt;title&gt; | SVG 简短名称 | 命名图形 |
| &lt;desc&gt; | SVG 描述 | 补充图形内容 |

本例没有旋转坐标系：x 向右增大，y 向下增大。位置与长度使用用户单位，经过缩放后不一定与 CSS 像素相等。

- &lt;rect&gt; 的 x、y 指左上角，width、height 指宽高。
- &lt;circle&gt; 的 cx、cy 指圆心，r 指半径。
- &lt;line&gt; 的 x1、y1 与 x2、y2 指两个端点。
- &lt;polygon&gt; 的 points 逐对写横、纵坐标，末尾顶点自动连回起点。
- fill 指填充颜色；stroke 指描边颜色，stroke-width 指描边宽度。线段需要描边才能看见，填充不能代替描边。

以下片段位于 shapes.svg 的 &lt;svg&gt; 内；先绘制背景，再绘制上面的形状。

```html
<rect x="4" y="4" width="232" height="112" fill="#edf2f4"></rect>
<circle class="sample-dot" cx="60" cy="60" r="24"
  fill="#b45309"></circle>
<line x1="120" y1="28" x2="120" y2="92"
  stroke="#334155" stroke-width="3"></line>
<polygon points="166,36 198,60 166,84" fill="#176b4a"></polygon>
```

配套文件：[scripts/15-svg-and-mathml/shapes.svg](scripts/15-svg-and-mathml/shapes.svg) · [浏览器预览](http://127.0.0.1:8015/scripts/15-svg-and-mathml/shapes.svg)

独立 SVG 按 XML 书写，根元素用 xmlns="http://www.w3.org/2000/svg" 声明命名空间。该地址只是标识，不是要下载的库；text/html 文档里的内联 &lt;svg&gt; 不必重复声明。

## 4 viewBox：坐标如何映射到显示区域

根 &lt;svg&gt; 的 width、height 指定本例的显示区域，即 SVG 视口（viewport）；这里没有 CSS 改写尺寸，数值按 CSS 像素使用。

viewBox="min-x min-y width height" 指定映射到视口的坐标矩形。四个名称是占位符，分别表示：

- min-x：矩形左边界的横坐标。
- min-y：矩形上边界的纵坐标。
- width：坐标矩形的宽度。
- height：坐标矩形的高度。

viewBox 中的宽高与外层同名属性是不同数据。下面把 100 × 100 的坐标范围放进 200 × 200 的视口；圆的坐标与半径仍按原来的用户单位书写。

```html
<svg id="size-b" class="viewport-demo" width="200" height="200"
  viewBox="0 0 100 100" role="img" aria-labelledby="size-b-title">
  <title id="size-b-title">等比例放大两倍的圆形</title>
  <rect width="100" height="100" fill="#dcecf2"></rect>
  <circle cx="50" cy="50" r="30" fill="#176b4a"></circle>
</svg>
<!-- 预期：B 的显示直径为 A 的两倍；源码 r 仍然是 30。 -->
```

配套文件：[scripts/15-svg-and-mathml/viewbox.html](scripts/15-svg-and-mathml/viewbox.html) · [浏览器预览](http://127.0.0.1:8015/scripts/15-svg-and-mathml/viewbox.html)

对照页面的 A 使用 100 × 100 视口，B 只将展示宽高加倍。viewBox 的宽、高必须为非负数；负数使属性无效，零会停止绘制，本章采用正数。

### 4.1 宽高比不同时：保持比例或拉伸

preserveAspectRatio 控制 viewBox 的宽高比与视口不同时如何缩放。

- xMidYMid meet：默认值。保持比例、横纵居中，让整个坐标范围放进视口，可能留白。
- none：允许横纵方向分别缩放，圆可能变成椭圆。

下面 C、D 的显示尺寸与图形坐标相同，只有缩放策略不同。id 和图形名称用于区分两幅图。

```html
<svg id="size-c" class="viewport-demo" width="200" height="100"
  viewBox="0 0 100 100" preserveAspectRatio="xMidYMid meet"
  role="img" aria-labelledby="size-c-title">
  <title id="size-c-title">保持圆形并左右留白</title>
  <rect width="100" height="100" fill="#dcecf2"></rect>
  <circle cx="50" cy="50" r="30" fill="#176b4a"></circle>
</svg>
<!-- 预期：圆直径仍为 60；浅色方块居中，左右各留 50 CSS 像素。 -->
```

```html
<svg id="size-d" class="viewport-demo" width="200" height="100"
  viewBox="0 0 100 100" preserveAspectRatio="none"
  role="img" aria-labelledby="size-d-title">
  <title id="size-d-title">被横向拉伸的圆形</title>
  <rect width="100" height="100" fill="#dcecf2"></rect>
  <circle cx="50" cy="50" r="30" fill="#176b4a"></circle>
</svg>
<!-- 预期：圆显示为宽 120、高 60 CSS 像素的椭圆，背景填满视口。 -->
```

配套文件：[scripts/15-svg-and-mathml/viewbox.html](scripts/15-svg-and-mathml/viewbox.html) · [浏览器预览](http://127.0.0.1:8015/scripts/15-svg-and-mathml/viewbox.html)

## 5 内联图形的名称、描述与样式

内联 SVG 的节点属于当前页面 DOM，可由页面 CSS 或脚本访问。&lt;img&gt; 所加载 SVG 的内部节点不会成为 &lt;img&gt; 的 HTML 子节点。

可访问名称（accessible name）说明“这是什么”，可访问描述（accessible description）补充内容。下面是 index.html 中内联 &lt;svg&gt; 的起始部分，后面接第 3 节的图形：

```html
<svg id="inline-card" width="240" height="120" viewBox="0 0 240 120"
  role="img" aria-labelledby="inline-card-title"
  aria-describedby="inline-card-desc">
  <title id="inline-card-title">圆、竖线与三角形（内联）</title>
  <desc id="inline-card-desc">浅色矩形中，圆在左、竖线在中、朝右三角形在右。</desc>
```

配套文件：[scripts/15-svg-and-mathml/index.html](scripts/15-svg-and-mathml/index.html) · [浏览器预览](http://127.0.0.1:8015/scripts/15-svg-and-mathml/index.html)

- role="img"：把本例静态图形作为整体图像理解，本身不提供名称。
- aria-labelledby：以 &lt;title&gt; 的 id 关联名称。
- aria-describedby：以 &lt;desc&gt; 的 id 关联描述。

引用值不加 #，目标 id 必须存在且唯一；也可关联合适的可见说明。显式 ARIA 名称、描述引用优先于默认的 &lt;title&gt;、&lt;desc&gt; 来源，避免提供互相矛盾的文本。这两个 SVG 元素不直接绘制文字，也不同于 HTML 中的文档标题或 title 属性。

两个版本的圆都带 sample-dot 类。styles.css 中这条规则只匹配当前页面的内联圆；CSS fill 会覆盖其同名展示属性。

```css
.sample-dot {
  fill: #156b84;
}
```

配套文件：[scripts/15-svg-and-mathml/styles.css](scripts/15-svg-and-mathml/styles.css) · [浏览器预览](http://127.0.0.1:8015/scripts/15-svg-and-mathml/index.html)

## 6 MathML：变量、数字和运算符

本章使用 MathML 表现标记（presentation markup）组织公式；标记负责排版，不执行计算。

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;math&gt; | 公式根元素 | 包住一条公式 |
| &lt;mi&gt; | 数学标识符 | 表示变量、函数名等 |
| &lt;mn&gt; | 数字字面量 | 表示数值常量 |
| &lt;mo&gt; | 运算符或分隔符 | 表示加号、等号和括号等 |
| &lt;mtext&gt; | 普通文字 | 在公式内放说明文字 |
| &lt;mrow&gt; | 子表达式分组 | 将多个项目组成一个整体 |
| &lt;mfrac&gt; | 分数 | 组织分子与分母 |
| &lt;msup&gt; | 上标结构 | 把上标附着于基础表达式 |
| &lt;msub&gt; | 下标结构 | 把下标附着于基础表达式 |

令 x 表示实数，1 是常量，+ 表示加法；按含义选择元素，不把整串 x + 1 放进 &lt;mi&gt;。

- &lt;math&gt; 的 display="inline" 是默认值，公式参与文字行。
- display="block" 使公式单独成块，并采用展示公式的排版。这里是 MathML 属性，不是 CSS 声明。

```html
<p>行内算式：<math><mi>x</mi><mo>+</mo><mn>1</mn></math>，仍在这句话中。</p>
<math display="block">
  <mtext>算式：</mtext><mi>x</mi><mo>+</mo><mn>1</mn>
</math>
<!-- 第一条参与文字行；第二条单独成块。标记没有给 x 赋值或求值。 -->
```

配套文件：[scripts/15-svg-and-mathml/basics.html](scripts/15-svg-and-mathml/basics.html) · [浏览器预览](http://127.0.0.1:8015/scripts/15-svg-and-mathml/basics.html#tokens)

## 7 分组与分数

&lt;mfrac&gt; 的第一个子元素作分子，第二个作分母。分子 x + 1 包含三项，需要用 &lt;mrow&gt; 合成一个子元素。以下表示先相加再除以 2：

```html
<math display="block">
  <mfrac>
    <mrow><mi>x</mi><mo>+</mo><mn>1</mn></mrow>
    <mn>2</mn>
  </mfrac>
</math>
<!-- x + 1 全在分数线上方；mfrac 的两个直接子元素分别是分子、分母。 -->
```

配套文件：[scripts/15-svg-and-mathml/basics.html](scripts/15-svg-and-mathml/basics.html) · [浏览器预览](http://127.0.0.1:8015/scripts/15-svg-and-mathml/basics.html#fractions)

改变分组就改变了表达式。下面表示 x 加上二分之一：

```html
<math display="block">
  <mi>x</mi><mo>+</mo><mfrac><mn>1</mn><mn>2</mn></mfrac>
</math>
<!-- 只有 1 在分数线上方，x 不属于分子。 -->
```

配套文件：[scripts/15-svg-and-mathml/basics.html](scripts/15-svg-and-mathml/basics.html) · [浏览器预览](http://127.0.0.1:8015/scripts/15-svg-and-mathml/basics.html#fractions)

本例不改变 MathML 子元素的 CSS 布局；每个 &lt;mfrac&gt; 写两个直接子元素。浏览器显示出文字不代表结构正确，不要让它猜测分子的范围。

## 8 上下标附着于哪个表达式

&lt;msup&gt; 的两项依次为基础表达式、上标；&lt;msub&gt; 的两项依次为基础表达式、下标。它们属于 MathML；HTML 的 &lt;sup&gt;、&lt;sub&gt; 用于文本上下标。

令 x 仍为实数。要表示 x 与 1 的和的平方，把带括号的整组作为第一项；&lt;mrow&gt; 只分组，可见括号用 &lt;mo&gt; 写出。

```html
<math display="block">
  <msup>
    <mrow><mo>(</mo><mi>x</mi><mo>+</mo><mn>1</mn><mo>)</mo></mrow>
    <mn>2</mn>
  </msup>
</math>
<!-- 上标 2 附着于带括号的整组；mrow 本身不会绘制括号。 -->
```

配套文件：[scripts/15-svg-and-mathml/basics.html](scripts/15-svg-and-mathml/basics.html) · [浏览器预览](http://127.0.0.1:8015/scripts/15-svg-and-mathml/basics.html#scripts)

令 a 为实数序列，n 为从 1 开始的正整数索引。把表示第 n 项的 &lt;msub&gt; 整体放进 &lt;msup&gt;，可写这一项的平方：

```html
<math display="block">
  <msup><msub><mi>a</mi><mi>n</mi></msub><mn>2</mn></msup>
</math>
<!-- 下标 n 是索引变量；每个 msub、msup 都有两个直接子元素。 -->
```

## 本章小结

- 外部 SVG 通过 &lt;img&gt; 引用并提供 alt；内联 SVG 的节点属于当前页面。

- W3C：[SVG 2 基本图形](https://www.w3.org/TR/SVG2/shapes.html) §10.2、10.3、10.5、10.7；[坐标系统](https://www.w3.org/TR/SVG2/coords.html#ViewBoxAttribute) §8.3–8.7，支持视口、viewBox 与宽高比；[SVG 名称与描述映射](https://www.w3.org/TR/svg-aam-1.0/#mapping_additional_nd) §7.1；[MathML Core](https://www.w3.org/TR/mathml-core/) §2.1.1、3.2、3.3.1–2、3.4.1，支持公式入口、符号、分组、分数与上下标；WAI [信息性图片](https://www.w3.org/WAI/tutorials/images/informative/)与[装饰性图片](https://www.w3.org/WAI/tutorials/images/decorative/)，支持替代文本按用途选择。
- MDN：[SVG 命名空间](https://developer.mozilla.org/en-US/docs/Web/SVG/Guides/Namespaces_crash_course#declaring_namespaces)、[SVG as an image](https://developer.mozilla.org/en-US/docs/Web/SVG/Guides/SVG_as_an_image#restrictions)、[将 SVG 加入页面](https://developer.mozilla.org/en-US/docs/Learn_web_development/Core/Structuring_content/Including_vector_graphics_in_HTML#adding_svg_to_your_pages)与 [fill](https://developer.mozilla.org/en-US/docs/Web/SVG/Reference/Attribute/fill)，支持外部图片、内联节点和 CSS 边界；[&lt;math&gt;](https://developer.mozilla.org/en-US/docs/Web/MathML/Reference/Element/math#attributes)、[Authoring MathML](https://developer.mozilla.org/en-US/docs/Web/MathML/Guides/Authoring#using_mathml)和 [Baseline 范围](https://developer.mozilla.org/en-US/docs/Glossary/Baseline/Compatibility)，支持展示方式、数学字体与支持范围。
- Chrome for Developers：[Accessibility features reference](https://developer.chrome.com/docs/devtools/accessibility/reference#tab)，支持 Elements 中的可访问信息检查。
- Python 3.12：[http.server 命令行](https://docs.python.org/3.12/library/http.server.html#command-line-interface)，支持本地端口、绑定地址和服务根目录。
