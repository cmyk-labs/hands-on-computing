适用范围：WHATWG HTML Living Standard；使用现代浏览器查看配套页面。

## 1 标题表达内容层级

标题帮助读者理解结构，也便于屏幕阅读器按标题定位。本例用一个 &lt;h1&gt; 表达中心主题，再按内容从属关系安排后续级别。

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;h1&gt; | 一级标题 | 最高级别标题 |
| &lt;h2&gt; | 二级标题 | 主标题下的主题 |
| &lt;h3&gt; | 三级标题 | 二级主题的子主题 |
| &lt;h4&gt; | 四级标题 | 更深一层的子主题 |
| &lt;h5&gt; | 五级标题 | 更深一层的子主题 |
| &lt;h6&gt; | 六级标题 | 最低级别标题 |

“周末阅读记录”是页面主标题。下面摘自其中的“阅读安排”：星期六与星期日都是它的子主题。

```html
<h2 id="reading-plan-title">阅读安排</h2>
<p>这次先读完整篇，再回头整理问题。</p>
<h3>星期六</h3>
<p>上午读两篇短文，下午记下想继续追问的问题。</p>
<h3>星期日</h3>
<p>重读一篇短文，把两次阅读的想法放在一起比较。</p>
```

配套文件：[scripts/05-semantic-page-structure/index.html](scripts/05-semantic-page-structure/index.html) · [浏览器预览](http://127.0.0.1:8005/scripts/05-semantic-page-structure/index.html#reading-plan)

- 进入子主题时尽量逐级下降，避免从 &lt;h2&gt; 跳到 &lt;h4&gt;；结束子主题后可以回到 &lt;h2&gt; 开始新主题。
- 层级来自内容关系，不是把一个标题元素嵌进另一个标题元素，也不靠 &lt;section&gt; 嵌套自动改写标题级别。
- 字号由 CSS 控制，不为缩小文字而降低标题级别。

通常每页保留一个清楚的主标题，这是一种组织建议，不应改写成“HTML 语法在所有情形下只允许一个 &lt;h1&gt;”。

## 2 独立内容与主题分节

一篇能独立阅读、分发或复用的内容可用 &lt;article&gt;；同一主题下的一组内容可用 &lt;section&gt;，通常带标题。

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;article&gt; | 独立内容 | 文章、帖子或读者评论等 |
| &lt;section&gt; | 主题分节 | 围绕同一主题组织内容 |

本页的 &lt;article&gt; 包住整篇阅读记录，内部的“阅读安排”“读后整理”分别用 &lt;section&gt;。以下是第二个分节的完整内容：

```html
<section id="review">
  <h2>读后整理</h2>
  <p>挑出一个问题，写成下周的阅读清单。</p>
  <ol>
    <li>写下最想保留的一句话。</li>
    <li>记录一个尚未想清楚的问题。</li>
    <li>选择下次继续阅读的方向。</li>
  </ol>
</section>
```

配套文件：[scripts/05-semantic-page-structure/index.html](scripts/05-semantic-page-structure/index.html) · [浏览器预览](http://127.0.0.1:8005/scripts/05-semantic-page-structure/index.html#review)

文章可以包含分节，一组文章也可以放在某个分节里；相关评论可用嵌套 &lt;article&gt;。选择依据是内容关系，不是字数或固定模板。

&lt;article&gt;、&lt;section&gt;、&lt;nav&gt;、&lt;aside&gt; 都属于分节内容（sectioning content），但各有用途。不是每段话都需要 &lt;section&gt;；仅为样式或脚本分组时，可用第 5 节的通用容器。

## 3 &lt;header&gt; 与 &lt;footer&gt; 的归属

&lt;header&gt; 组织介绍或导航辅助内容，可以包含标题与简介；&lt;footer&gt; 放所属内容的作者、补充说明或相关链接。

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;header&gt; | 页头或内容头部 | 组织介绍性内容 |
| &lt;footer&gt; | 页脚或内容尾部 | 组织所属内容的补充信息 |

它们不是标题元素，也不会自行建立新分节。同一页面可有页面级的头尾，也可在文章内有自己的头尾。下面摘自本页 &lt;article&gt; 内：

```html
<!-- 文章级 header；本页唯一 h1 表达中心主题。 -->
<header>
  <h1>周末阅读记录</h1>
  <p>给周末留一段安静阅读的时间。</p>
  <div class="record-meta">
    <p>记录人：小林</p>
    <p>分类：<span class="category">阅读</span></p>
  </div>
</header>
```

配套文件：[scripts/05-semantic-page-structure/index.html](scripts/05-semantic-page-structure/index.html) · [浏览器预览](http://127.0.0.1:8005/scripts/05-semantic-page-structure/index.html)

```html
<!-- 文章 footer 的作用域是 article，不是整站。 -->
<footer>
  <p>整理说明：这份记录只保留本周末的阅读安排与想法。</p>
</footer>
```

配套文件：[scripts/05-semantic-page-structure/index.html](scripts/05-semantic-page-structure/index.html) · [浏览器预览](http://127.0.0.1:8005/scripts/05-semantic-page-structure/index.html)

&lt;footer&gt; 对应最近的分节内容祖先，没有这类祖先时对应 &lt;body&gt;；&lt;div&gt; 不会改变这种归属。它通常放在末尾，但不要求固定在窗口底部。

&lt;header&gt; 与 &lt;footer&gt; 内都不能再嵌套 &lt;header&gt; 或 &lt;footer&gt;。“文章有自己的头尾”不等于把文章页头放进站点页头。

## 4 主要内容、导航与旁支内容

先分清哪些内容属于本页主线，哪些用于导航，哪些是可单独理解的补充。

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;main&gt; | 主要内容区域 | 容纳本页主导内容 |
| &lt;nav&gt; | 导航区域 | 提供主要的站点或页内导航 |
| &lt;aside&gt; | 旁支或补充内容 | 容纳与正文相关但可分离的内容 |

本页 &lt;main&gt; 容纳阅读记录，&lt;article&gt; 表达记录的独立性。站点页头、导航与页脚放在 &lt;main&gt; 外，书单用同级的 &lt;aside&gt;。

同一文档至多一个不带 hidden 属性的 &lt;main&gt;；单靠 display:none 隐藏第二个不满足这条限制。本例将 &lt;main&gt; 放在 &lt;body&gt; 内的 &lt;div&gt; 中，不放进 &lt;article&gt;、&lt;section&gt;、&lt;aside&gt;、&lt;nav&gt;、&lt;header&gt; 或 &lt;footer&gt;。

页内目录通过 &lt;a&gt; 链接到目标 id。aria-label 为导航提供名称，第 6 节说明其作用。

```html
<nav aria-label="页内目录">
  <ul>
    <li><a href="#reading-plan">阅读安排</a></li>
    <li><a href="#review">读后整理</a></li>
  </ul>
</nav>
```

配套文件：[scripts/05-semantic-page-structure/index.html](scripts/05-semantic-page-structure/index.html) · [浏览器预览](http://127.0.0.1:8005/scripts/05-semantic-page-structure/index.html)

- \#reading-plan：指向本页“阅读安排”分节。
- \#review：指向本页“读后整理”分节。

&lt;nav&gt; 用于主要导航块，页脚的一条“回到页头”链接不必再包进去。链接的详细规则见[链接与 URL](06-链接与 URL.ipynb)。

### 4.1 &lt;aside&gt; 不由左右位置决定

书单与阅读主题相关，但能从本次安排中分离，所以用 &lt;aside&gt;。正文必需的解释不能只因摆在侧边就改成旁支内容。下面区域通过标题 id 获得名称：

```html
<aside aria-labelledby="booklist-title">
  <h2 id="booklist-title">随手书单</h2>
  <p>下次想试试这些阅读方向：</p>
  <ul>
    <li>一篇旅行随笔</li>
    <li>一则短篇故事</li>
  </ul>
</aside>
```

配套文件：[scripts/05-semantic-page-structure/index.html](scripts/05-semantic-page-structure/index.html) · [浏览器预览](http://127.0.0.1:8005/scripts/05-semantic-page-structure/index.html)

把书单从右侧移到正文下方，不改变它与主线内容的关系。布局在第 7 节调整。

## 5 &lt;div&gt; 与 &lt;span&gt;：通用分组

没有更合适的语义元素，又需要一起设置样式或属性时，可使用通用容器。

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;div&gt; | 通用流内容容器 | 把段落等流内容分在一起 |
| &lt;span&gt; | 通用短语内容容器 | 包住一段文字或短语内容 |

record-meta 给两段附带信息分组；category 为“阅读”这个词提供样式定位点。它们都是本例自定的 class 名称。

```html
<div class="record-meta">
  <p>记录人：小林</p>
  <p>分类：<span class="category">阅读</span></p>
</div>
```

配套文件：[scripts/05-semantic-page-structure/index.html](scripts/05-semantic-page-structure/index.html) · [浏览器预览](http://127.0.0.1:8005/scripts/05-semantic-page-structure/index.html)

默认情况下 &lt;div&gt; 通常形成块级盒，&lt;span&gt; 通常参与行内排版，但 CSS 不能改变它们的内容模型。把 &lt;span&gt; 显示成块状，也不能让它合法包含 &lt;p&gt;。

class="heading" 不会把普通容器变成标题；有明确的标题、强调或导航含义时，使用相应原生元素。

## 6 区域名称与可访问地标

地标（landmark）供辅助技术快速定位主要页面区域；可访问名称（accessible name）说明区域叫什么。HTML 的分节内容和地标角色不是同一分类，例如 &lt;article&gt; 是分节内容，但其 article 角色不是地标。

- aria-labelledby：引用已有标题的 id，值不带 #；本例 reading-plan-title 指向“阅读安排”。
- aria-label：直接给出名称文本，本例导航名称为“页内目录”。

有可见标题时优先引用它。普通 &lt;section&gt; 获得可访问名称后才隐式成为 region；单独写 id 或包含 &lt;h2&gt; 不会自动完成命名。

下面是已命名分节的完整结构，可在 Elements 的 Accessibility 面板检查角色和名称：

```html
<section id="reading-plan" aria-labelledby="reading-plan-title">
  <h2 id="reading-plan-title">阅读安排</h2>
  <p>这次先读完整篇，再回头整理问题。</p>
  <h3>星期六</h3>
  <p>上午读两篇短文，下午记下想继续追问的问题。</p>
  <h3>星期日</h3>
  <p>重读一篇短文，把两次阅读的想法放在一起比较。</p>
</section>
```

配套文件：[scripts/05-semantic-page-structure/index.html](scripts/05-semantic-page-structure/index.html) · [浏览器预览](http://127.0.0.1:8005/scripts/05-semantic-page-structure/index.html#reading-plan)

本页原生元素对应的角色如下，名称与角色是不同信息：

- 页面级 &lt;header&gt;：banner，页面级页头地标。
- &lt;nav&gt;：navigation，名称“页内目录”。
- &lt;main&gt;：main，主要内容地标。
- &lt;aside&gt;：complementary，名称“随手书单”。
- 页面级 &lt;footer&gt;：contentinfo，页面级页脚地标。
- “阅读安排”的 &lt;section&gt;：region，名称“阅读安排”。
- “读后整理”的 &lt;section&gt;：未命名的通用区域，不是 region；内部标题仍是二级标题。

当 &lt;header&gt;/&lt;footer&gt; 位于 &lt;article&gt;、&lt;aside&gt;、&lt;main&gt;、&lt;nav&gt; 或 &lt;section&gt; 的后代位置时，不再提供页面级 banner/contentinfo 地标。文章头尾仍有原来的 HTML 含义。

小型分节可以只保留标题，不必全部命名成地标；多个用途不同的导航应有可区分的名称。通用节点可能在可访问树中被省略，不要求每个 DOM 节点都出现。

## 7 CSS 改变布局

本页用 class="page-layout" 的 &lt;div&gt; 包住 &lt;main&gt; 和 &lt;aside&gt;，只用于布局，不增加主题分节。样式从页面头部引入：

```html
<link rel="stylesheet" href="styles.css">
```

配套文件：[scripts/05-semantic-page-structure/index.html](scripts/05-semantic-page-structure/index.html) · [浏览器预览](http://127.0.0.1:8005/scripts/05-semantic-page-structure/index.html)

styles.css 在宽窗口中设置双栏。窗口宽度不超过 720 CSS 像素时，下面的媒体查询把网格改为一列：

```css
/* 预期：宽视口书单在右侧；不超过 720 CSS 像素时书单在正文后面。 */
@media (max-width: 720px) {
  .page-layout {
    grid-template-columns: minmax(0, 1fr);
  }
}
/* 检查：改变窗口宽度后，main/aside 的职责及标题级别仍保持原样。 */
```

配套文件：[scripts/05-semantic-page-structure/styles.css](scripts/05-semantic-page-structure/styles.css) · [浏览器预览](http://127.0.0.1:8005/scripts/05-semantic-page-structure/index.html)

@media 按视口条件应用样式，grid-template-columns 指定列宽；1fr 按比例取得剩余空间，minmax(0, 1fr) 允许这一列缩小。这里只用它们观察布局，不展开 CSS 课程。

这些规则保留 DOM 内容顺序、标题级别和区域名称。也可在 Elements 中临时删除引入样式的 &lt;link&gt; 作对照，刷新恢复；修改浏览器中的节点不会改写文件。

不能扩大为“任何 CSS 都不影响可访问性”，例如 display:none 会让内容从可访问树中消失。

## 本章小结

- 标题表达层级；&lt;header&gt; 组织介绍内容，不等于标题。
- &lt;article&gt; 表达独立性，&lt;section&gt; 表达主题，&lt;main&gt; 表达本页主线。
- &lt;nav&gt; 和 &lt;aside&gt; 按内容关系选择；&lt;div&gt;/&lt;span&gt; 用于通用分组。
- 地标角色与区域名称分别检查，不能把有标题的分节一律看作已命名地标。

自查：书单移到正文下方后，为什么仍使用 &lt;aside&gt;？文章头尾为什么不增加页面级地标？

复制 scripts/05-semantic-page-structure/index.html 为 scripts/05-semantic-page-structure/practice.html；保存后通过[练习页面](http://127.0.0.1:8005/scripts/05-semantic-page-structure/practice.html)检查。

（3）为文章添加一条独立的读者回应，含标题、内容和署名。检查：能解释使用 &lt;article&gt; 的理由，标题层级正确，页面级 banner/contentinfo 没有因此重复。

### 提示

新增区域名称可用 aria-labelledby 关联可见标题；读者回应可以是与外层文章相关的嵌套文章。练习副本用后可删除。

### 参考解析

1. “下周想读”属于“读后整理”，用 h3 即可；它没有引入新的页面主线，无需另建 main。
2. 给现有 h2 设置唯一 id，例如 review-title，再让 section 的 aria-labelledby 引用这个值。新增的是区域名称与隐式 region 角色，标题仍为 h2；删去命名属性后该区域不再以这个名称成为地标。
3. 读者回应能单独阅读，可作为与原记录相关的嵌套 article；放在原文章的主题层级下安排标题。其内部 header/footer 属于文章，不产生新的页面级 banner/contentinfo。
4. 把基础 .page-layout 规则的 grid-template-columns 改为 minmax(0, 1fr)。原 DOM 中 main 先、aside 后，单栏时书单随之排列在后；aside 的命名与含义无须改变。
