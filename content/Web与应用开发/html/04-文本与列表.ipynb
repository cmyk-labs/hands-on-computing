## 1 段落、换行与主题分隔

一段文字表达一个相对完整的意思时，用 &lt;p&gt;；内容本身需要换行时，用 &lt;br&gt;；话题转折时，用 &lt;hr&gt;。

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;p&gt; | 段落 | 组织一段正文 |
| &lt;br&gt; | 换行 | 标记地址、诗行等内容中的换行 |
| &lt;hr&gt; | 主题分隔 | 标记段落层面的场景或话题转换 |

&lt;br&gt; 与 &lt;hr&gt; 是空元素。不要用连续换行或空段落撑间距；留白交给 CSS。

```html
<!-- 检查：楼层与房间换行，但仍属于同一个段落。地点为教学示例。 -->
<p>学习中心二楼<br>204 室</p>

<!-- hr 表示从地点转到课后安排；不是为了画装饰线。 -->
<hr>
<p>活动结束后，请整理各自的学习笔记。</p>
```

普通段落在默认样式下通常折叠连续空格与源码换行。&lt;p&gt; 只容纳短语内容（phrasing content），即文字和段落内部的文本标记，不能嵌入列表、&lt;pre&gt; 或另一个 &lt;p&gt;；内容类别不同于视觉布局中的“行内”“块级”。

## 2 语气强调与重要性

&lt;em&gt; 表示语气强调（stress emphasis），强调位置影响句子的侧重点；&lt;strong&gt; 表示重要、严肃或紧迫的内容。

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;em&gt; | 语气强调 | 标出朗读时应强调的词句 |
| &lt;strong&gt; | 重要内容 | 标出提醒、警告等重要信息 |

浏览器常用斜体和粗体呈现它们，但 &lt;strong&gt; 不是“更粗的 &lt;em&gt;”。

```html
<!-- 比较两句话：第一句强调由谁复查，第二句强调哪一个版本。 -->
<p><em>我</em>会复查这份笔记。</p>
<p>我会复查<em>这份</em>笔记。</p>

<!-- strong 表达重要性；b 只突出本次要留意的关键词。 -->
<p><strong>修改前请保留原稿副本。</strong></p>
```

### 2.1 其他文本标记

相似的字形可能表达不同含义。先判断为什么标记这段文字，再选择元素。

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;b&gt; | 提请注意 | 突出关键词，不增加重要性 |
| &lt;i&gt; | 区别于普通叙述的文本 | 标记技术术语、外语习语或内心话等 |
| &lt;u&gt; | 非文字注记 | 标出拼写错误等注记 |
| &lt;s&gt; | 已不准确或不再相关的内容 | 保留失效安排等信息 |
| &lt;mark&gt; | 与当前语境相关的标记 | 突出搜索命中词等 |
| &lt;small&gt; | 附注 | 标记简短的小字说明 |

&lt;b&gt;、&lt;i&gt;、&lt;u&gt;、&lt;s&gt; 在现代 HTML 中都有语义。单纯改变字形使用 CSS；&lt;u&gt; 的下划线容易与链接混淆，应有明确注记语境。

```html
<p>本次记录的关键词是 <b>文本</b> 和 <b>列表</b>。</p>

<!-- i 标记人物内心话；u 标记故意写错的英文单词。 -->
<p>我心想：<i>这次应该先读清楚内容。</i></p>
<p>拼写检查：<u lang="en">recieve</u> 应改为 <span lang="en">receive</span>。</p>
```

&lt;s&gt; 不同于记录文档删除操作的 &lt;del&gt;，见第 7 节。&lt;small&gt; 只适合短附注，不会取消内部内容的重要性。

## 3 引用与作品名称

独立引文用 &lt;blockquote&gt;，融入句子的短引文用 &lt;q&gt;，作品标题用 &lt;cite&gt;。

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;blockquote&gt; | 块引用 | 标记来自其他来源的一段内容 |
| &lt;q&gt; | 行内引用 | 标记句子中的引文 |
| &lt;cite&gt; | 作品名称 | 标记书、论文、网页等作品的标题 |

cite 属性保存来源 URL，不会自动生成可见出处。&lt;cite&gt; 元素则标记作品名称，不能用来标记作者姓名。&lt;blockquote&gt; 的署名和出处说明放在元素外；不要把它当成缩进工具。

下面英文是 WHATWG 对段落的定义原文，出处在引文之后：

```html
<!-- 原文摘录；检查 DOM：出处段落应是 blockquote 的后续兄弟元素。 -->
<blockquote cite="https://html.spec.whatwg.org/multipage/grouping-content.html#the-p-element">
  <p lang="en">The p element represents a paragraph.</p>
</blockquote>
<p>—— WHATWG，<cite>HTML Living Standard</cite>，The p element 小节。</p>
```

### 3.1 缩写与全称

&lt;abbr&gt; 表示缩略词或首字母缩略词。title 属性可提供全称；填写时只写全称，不混入其他说明。

陌生缩写首次出现时，把全称直接写进正文，不能只依赖悬停提示。

## 4 代码与预格式化文本

&lt;code&gt; 表示代码，不会执行它或自动高亮语法；&lt;pre&gt; 表示预格式化文本，通常保留空格和换行。多行代码可以组合两者。

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;code&gt; | 代码片段 | 标记计算机代码 |
| &lt;pre&gt; | 预格式化文本 | 保留文本布局 |
| &lt;kbd&gt; | 用户输入 | 标记按键等输入 |
| &lt;samp&gt; | 输出示例 | 标记程序或系统的示例反馈 |
| &lt;var&gt; | 变量 | 标记表达式或代码中的变量名称 |

&lt;pre&gt; 内仍会解析 HTML。显示源码时，用 &amp;lt;、&amp;gt;、&amp;amp; 分别表示 &lt;、&gt;、&amp;，不能把它当作原样显示任意 HTML 的容器。

```html
<!-- 检查：显示三行源码，中间一行保留两个空格；源码标签不被再次解析。 -->
<pre><code>&lt;ul&gt;
  &lt;li&gt;准备笔记&lt;/li&gt;
&lt;/ul&gt;</code></pre>
```

## 5 上下标与注音

&lt;sub&gt;、&lt;sup&gt; 用于有约定含义的上下标，如化学式和指数，不用于单纯调整字形位置。&lt;ruby&gt; 为基础文字附上行间注释，本例用于汉字注音。

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;sub&gt; | 下标 | 表达化学式、变量下标等记法 |
| &lt;sup&gt; | 上标 | 表达指数等记法 |
| &lt;ruby&gt; | 行间注释 | 关联基础文字与注释 |
| &lt;rt&gt; | 注释文本 | 写对应读音等说明 |
| &lt;rp&gt; | 回退括号 | 为不支持 ruby 布局的环境保留括号 |

最简单的注音先写基础文字，再写 &lt;rt&gt;；&lt;rp&gt; 放在注释两侧。读音由作者填写，浏览器不会自动生成。这里的 ruby 是 HTML 标记，与同名编程语言无关。

```html
<!-- 预期：H₂O 中的 2 是下标，2² 中后一个 2 是上标。 -->
<p>化学式：H<sub>2</sub>O；幂的写法：2<sup>2</sup>。</p>

<!-- 预期：支持 ruby 时，hàn 位于“汉”的上方附近，回退括号不显示。 -->
<!-- 在 DOM 中仍可找到 rt 与两侧 rp；不要用空格手工把拼音对齐。 -->
<p>读音：<ruby>汉<rp>（</rp><rt>hàn</rt><rp>）</rp></ruby>字。</p>
```

## 6 时间与机器可读值

&lt;time&gt; 表达日期、时刻或时长，元素内容供读者阅读，datetime 属性提供机器可读形式。它不是日期输入控件，不会自动创建日程。

- 2026-09-12：年、月、日，用连字符分隔。
- 14:30：24 小时制的时刻。
- 2026-09-12T14:30:00+08:00：带时区偏移的日期时间；T 分隔日期与时间，+08:00 表示比 UTC 快 8 小时。
- PT45M：45 分钟的时长；P 引入时长，T 引入时间部分，45M 表示 45 分钟。

datetime 不能写成“下午两点半”等任意文字；应选择 HTML 允许的格式。

```html
<!-- 检查：页面显示中文日期；在 DOM 中查看 datetime 才能看到机器可读值。 -->
<p>示例交流日期：<time datetime="2026-09-12">2026年9月12日</time>。</p>
<p>开始时间：<time datetime="2026-09-12T14:30:00+08:00">北京时间14:30</time>。</p>
<p>计划时长：<time datetime="PT45M">45分钟</time>。</p>
```

## 7 修改痕迹与失效信息

&lt;del&gt;、&lt;ins&gt; 分别标记从文档删除、加入的内容；&lt;s&gt; 表示当前已经失效的信息。选择取决于是否在记录一次编辑。

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;del&gt; | 删除内容 | 保留改动前的文字 |
| &lt;ins&gt; | 新增内容 | 标出改动后的文字 |

- datetime：在编辑标记上记录修改日期，或带时区的完整日期时间；不是活动本身的发生时间。
- cite：可选的修改说明 URL，不会自动生成可见说明。

默认删除线与下划线可以被 CSS 改变。关键修改同时写明“原为”“现为”，不要只靠线条传达。一个段落中的局部修改放在该 &lt;p&gt; 内，不跨越多个段落的局部内容。

```html
<!-- 日期仅是教学数据；datetime 记录此次修订发生于哪一天。 -->
<p>地点修订：原为 <del datetime="2026-09-10">204 室</del>，
现为 <ins datetime="2026-09-10">305 室</ins>。</p>

<!-- s 说明旧信息失效，不表示执行了某个删除操作。 -->
<p><s>旧说明：仍可报名。</s> 当前状态：报名已经截止。</p>
<!-- 检查：旧文字仍在源码和页面中；这些标记不会把文本从文件中擦除。 -->
```

## 8 有序、无序与嵌套列表

交换条目顺序会改变含义时，用 &lt;ol&gt;，如操作步骤；顺序无关时，用 &lt;ul&gt;，如准备物品。条目用 &lt;li&gt; 表达，不能仅在段落前手打数字或圆点。

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;ol&gt; | 有序列表 | 表达有先后关系的条目 |
| &lt;ul&gt; | 无序列表 | 表达顺序无关的条目 |
| &lt;li&gt; | 列表项 | 容纳一条内容 |

嵌套列表放在所属 &lt;li&gt; 内，结束内层列表后再关闭该外层条目。列表本身放在段落之外。

```html
<!-- 检查：外层步骤依次为1、2、3；无序的两件物品属于第一步。 -->
<ol>
  <li>准备材料
    <ul>
      <li>笔记本</li>
      <li>签字笔</li>
    </ul>
  </li>
  <li>记录讨论内容</li>
  <li>整理待办事项</li>
</ol>
```

### 8.1 控制编号

- start：&lt;ol&gt; 的起始整数，普通递增列表默认从 1 开始。
- value：有序列表中某个 &lt;li&gt; 的整数编号，后面的条目接着计数。
- reversed：&lt;ol&gt; 的布尔属性，使编号递减，不会颠倒条目的 DOM 顺序。
- type：&lt;ol&gt; 的编号形式，可用 1、a、A、i、I，分别表示数字、大小写字母及大小写罗马数字。

先看递增列表中途调整编号的写法：

```html
<!-- 预期编号：4、8、9；value 重设当前条目，后面的条目从它继续递增。 -->
<ol start="4">
  <li>第四条记录</li>
  <li value="8">接到第八条记录</li>
  <li>下一条记录</li>
</ol>
```

配套文件：[scripts/04-text-and-lists/index.html](scripts/04-text-and-lists/index.html) · [浏览器预览](http://127.0.0.1:8004/scripts/04-text-and-lists/index.html#lists)

未给 start 的递减列表，从外层自身的 &lt;li&gt; 数量开始，内层列表项不计入。reversed="false" 仍表示该属性存在。

```html
<!-- 预期编号：3、2、1；甲、乙、丙的文本顺序没有改变。 -->
<ol reversed>
  <li>甲</li>
  <li>乙</li>
  <li>丙</li>
</ol>
```

## 9 名称与描述

描述列表（description list）表达名称与值的关联，如术语释义或资料卡；不只是把文字排成两列。

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;dl&gt; | 描述列表 | 组织名称与描述的分组 |
| &lt;dt&gt; | 名称或术语 | 标记被描述的对象 |
| &lt;dd&gt; | 描述或值 | 提供前面名称对应的内容 |

每组先有一个或多个 &lt;dt&gt;，再有一个或多个 &lt;dd&gt;。多个名称可共享描述，一个名称也可对应多个独立值。

```html
<!-- 教学资料卡：两个名称共用一段描述，“记录人”对应两个独立值。 -->
<dl>
  <dt>学习手册</dt>
  <dt>读者指南</dt>
  <dd>这份示例资料面向初次参加读书会的成员。</dd>

  <dt>记录人</dt>
  <dd>小林</dd>
  <dd>小周</dd>
</dl>
```

配套文件：[scripts/04-text-and-lists/index.html](scripts/04-text-and-lists/index.html) · [浏览器预览](http://127.0.0.1:8004/scripts/04-text-and-lists/index.html#descriptions)

同一描述中的多段话仍放在一个 &lt;dd&gt; 内，不因换段就拆成多个值。配套页面的“参与方式”将两个段落和列表放在同一个描述中；&lt;dd&gt; 与 &lt;li&gt; 都可包含列表，但前者是描述，后者是条目。

## 10 样式改变，语义保留

配套页面末尾用 class="plain-text" 选中第二个对照段落。CSS 的 font-weight: normal 与 font-style: normal 分别取消加粗和斜体，HTML 元素仍保留原来的含义。

```css
.plain-text strong {
  font-weight: normal;
}

.plain-text em {
  font-style: normal;
}
```

## 本章小结

- 段落、内容换行与话题转折分别用 &lt;p&gt;、&lt;br&gt;、&lt;hr&gt;。
- 语气、重要性、引用、作品名称各有标记，不能只按默认字形选择。
- 预格式化文本仍解析 HTML；时间属性与编辑时间要按各自允许格式填写。
- 有序列表表达先后，无序列表表达条目，描述列表表达名称与值的关联。

复制 scripts/04-text-and-lists/index.html 为 scripts/04-text-and-lists/practice.html；以下修改都在副本中进行，保存后通过[练习页面](http://127.0.0.1:8004/scripts/04-text-and-lists/practice.html)检查。

（1）改写成一则虚构读书会通知，含两个段落、一处内容换行、一条重要提醒和一处语气强调。检查：没有用空段落或连续 &lt;br&gt; 撑间距，能解释每个文本标记的选择。

（2）让一个递增列表显示 3、6、7，一个递减列表显示 4、3、2、1；给其中一项加入两项无序子列表。检查：编号由浏览器生成，内层列表在所属 &lt;li&gt; 内。

（3）制作资料卡，包含“两个名称共享一个描述”和“一个名称对应两个独立值”。检查：每组先名称后描述；同一描述的多段话仍在一个 &lt;dd&gt; 中。

（4）加入一段已核实的简短引用、一段显示 HTML 源码的预格式化文本，以及含注音、活动时间和修改痕迹的通知。检查：可见出处区分作品与作者，源码显示成文字，注音对应基础文字，活动时间与修改时间分开。

### 提示

先预测编号再刷新；时间和修改说明可在 Elements 查看属性。引用仅摘录必要文字，先核对来源。完成后可删除副本。

### 参考解析

1. 段落用于组织完整意思，br 只保留地址或诗行等内容换行；重要提醒选 strong，语气侧重点选 em。选用理由与字号、粗细无关。
2. 递增列表设 start="3"，第二个 li 设 value="6"，第三项继续到 7。四项递减列表只需 reversed；子列表置于某个 li 内，不能算作外层起点的四项之一。
3. 两个连续 dt 后接一个 dd 构成共享描述；一个 dt 后接两个 dd 表达两个独立值。同一个描述的两个段落放进同一个 dd，避免改变分组含义。
4. 可沿用正文已核查的短引文，把可见出处放在 blockquote 外；HTML 源码先转义特殊字符。time 的 datetime 记录活动时刻，ins/del 的 datetime 记录修订时刻，二者不必相同。
