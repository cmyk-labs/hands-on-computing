# 表单结构与控件

学习目标：能选择合适的表单控件，建立标签与分组，并区分名称、值、默认状态和结果显示。

前置知识：HTML 元素与属性、id、基本文档结构，以及文件和目录操作。

适用范围：WHATWG HTML Living Standard；使用现代浏览器。日期、颜色选择器和输入建议的外观与支持情况可能不同。

环境准备：[环境配置与运行](README.md)。

工作目录：content/Web与应用开发/html；在浏览器运行配套页面。

配套脚本：位于 scripts/10-form-structure-and-controls/。

1. [index.html](scripts/10-form-structure-and-controls/index.html)：标签、分组与本章导航。
2. [buttons.html](scripts/10-form-structure-and-controls/buttons.html)：普通、重置与提交按钮。
3. [text-controls.html](scripts/10-form-structure-and-controls/text-controls.html)、[quantities-and-dates.html](scripts/10-form-structure-and-controls/quantities-and-dates.html)：文字、数量、颜色和日期时间输入。
4. [choices.html](scripts/10-form-structure-and-controls/choices.html)、[list-controls.html](scripts/10-form-structure-and-controls/list-controls.html)：复选、单选、选择列表与输入建议。
5. [file-and-hidden.html](scripts/10-form-structure-and-controls/file-and-hidden.html)、[sample.txt](scripts/10-form-structure-and-controls/sample.txt)：文件选择、隐藏字段及练习素材。
6. [states.html](scripts/10-form-structure-and-controls/states.html)：默认状态、只读与禁用。
7. [results.html](scripts/10-form-structure-and-controls/results.html)：输出结果、度量值与进度；内含同步滑块结果的辅助脚本。

## 2 标签、名称与分组

### 2.1 用 &lt;label&gt; 说明输入用途

&lt;label&gt; 的 for 对应控件的 id。下面 plan-title 是本例自定标识；点击“安排名称”即可把焦点移到输入框。

- id：元素标识符，同一文档内必须唯一。
- for：标签关联的目标 id，值前不加 #。
- name：提交的数据项名称，本例为 title，不用于标签关联。
- value：为本例的文本输入提供默认文字。

普通文本框没有 name 仍能编辑，但不会以该输入值形成提交项；id 不能代替 name。

```html
<label for="plan-title">安排名称</label>
<input id="plan-title" name="title" type="text" value="周末阅读">
```

### 2.2 用 &lt;form&gt; 与 &lt;fieldset&gt; 组织相关控件

&lt;form&gt; 组织一组输入和操作，可以包含段落与分组，不能嵌套另一个 &lt;form&gt;。以下片段位于首页的表单中。

&lt;fieldset&gt; 把相关输入归组，开头的 &lt;legend&gt; 说明整组用途。组标题不能代替每个控件自己的标签。

```html
<fieldset>
  <legend>安排内容</legend>
  <p>
    <label for="plan-title">安排名称</label>
    <input id="plan-title" name="title" type="text" value="周末阅读">
  </p>
  <p>
    <label>虚构地点 <input id="plan-place" name="place" type="text" value="练习室 A"></label>
  </p>
</fieldset>
```

## 3 明确按钮类型

&lt;button&gt; 的文字说明操作，type 决定表单行为：

- button：普通按钮，没有提交或重置的默认动作。
- reset：恢复关联表单中可重置控件的默认状态。
- submit：触发表单提交流程。

本例这些普通表单按钮省略 type 时会作为提交按钮处理，因此显式写明用途。先修改安排名称，再比较三个按钮。

```html
<button type="button">普通按钮（未绑定动作）</button>
<button type="reset">恢复默认安排</button>
<button type="submit">观察提交事件</button>
<!-- 检查：普通按钮不改输入；reset 恢复“周末阅读”；submit 只更新提示。 -->
<!-- 地址应保持不变；该提示是脚本执行的结果，不代表服务器已经接收数据。 -->
```

## 4 文字输入

### 4.1 按用途选择单行输入

&lt;input&gt; 的 type 指定输入类型，省略时默认 text。常用文字类型如下：

- text：普通单行文字。
- password：遮蔽显示的口令文字；遮蔽不代表数据已加密。
- email：邮箱地址，有原生格式约束。
- url：完整 URL，有原生格式约束。
- tel：电话号码，不限定统一电话号码格式。
- search：搜索词。

这些类型不会自动发信、打开网址、拨号或执行搜索。配套页提供六种输入供比较，以下先看普通文字与口令。

```html
<p><label for="text-title">安排名称</label>
  <input id="text-title" name="title" type="text" value="阅读练习"></p>
<p><label for="demo-password">演示口令</label>
  <input id="demo-password" name="demo_password" type="password" value="demo-only-123"></p>
```

## 5 数量、颜色与日期时间

### 5.1 数量与偏好等级

- number：用于可增减的数量。电话、编号即使只含数字，也不因此改用 number。
- range：用于不要求精确录入的范围选择，通常显示为滑块。
- color：用于颜色选择。本例 #336699 是一种不透明蓝色；选择颜色不会自动给页面配色。

滑块的 min、max 指定范围两端，value 给出初始值。

```html
<p><label for="article-count">文章数（篇）</label>
  <input id="article-count" name="article_count" type="number" value="3"></p>
<p><label for="interest-level">兴趣等级（0～5）</label>
  <input id="interest-level" name="interest_level" type="range" min="0" max="5" value="3"></p>
<p><label for="mark-color">标记颜色</label>
  <input id="mark-color" name="mark_color" type="color" value="#336699"></p>
```

### 5.2 日期时间的值与显示

常用类型及本例默认值如下；这些是具体值，不是待替换的占位符。

- date：2026-09-20，年月日。
- time：09:30，24 小时制的时与分。
- datetime-local：2026-09-20T09:30，T 分隔日期和时间，不包含时区。
- month：2026-09，年月。
- week：2026-W38，ISO 周编号年中的第 38 周，表示一整周。

日期选择器的显示格式可能随区域设置变化，控件值仍按对应格式表示。datetime-local 不含 UTC 偏移，不能单独确定全球统一的时间点。

```html
<p><label for="reading-date">阅读日期</label>
  <input id="reading-date" name="reading_date" type="date" value="2026-09-20"></p>
<p><label for="reading-time">开始时间</label>
  <input id="reading-time" name="reading_time" type="time" value="09:30"></p>
<p><label for="local-start">日期与时间（不含时区）</label>
  <input id="local-start" name="local_start" type="datetime-local" value="2026-09-20T09:30"></p>
```

## 6 复选框与单选按钮

### 6.1 checkbox：各项独立选择

checkbox 是复选框，每项可以独立勾选。checked 表示默认勾选，value 表示选项含义；二者职责不同。

checked 是布尔属性，不需要默认选中时省略它，不能写 checked="false"。用户操作改变当前勾选状态，不会自动改写 HTML 的 checked 属性。

```html
<fieldset>
  <legend>练习方式（可多选）</legend>
  <p><input id="method-reading" name="method" type="checkbox" value="reading" checked>
    <label for="method-reading">阅读</label></p>
  <p><input id="method-coding" name="method" type="checkbox" value="coding">
    <label for="method-coding">写代码</label></p>
</fieldset>
```

配套文件：[scripts/10-form-structure-and-controls/choices.html](scripts/10-form-structure-and-controls/choices.html) · [浏览器预览](http://127.0.0.1:8010/scripts/10-form-structure-and-controls/choices.html)

本例两个选项的 name 都是 method，仍可以同时勾选。提交时只收集选中的项，不会为未选项发送 false。

### 6.2 radio：同组最多选一项

在本例同一表单、同一元素树内，radio 使用相同且非空的 name 组成互斥组。period 是自定组名，morning 和 evening 是两个选项的值。

&lt;fieldset&gt; 负责内容分组，radio 的互斥关系依赖 name 等归属条件，不是由边框产生。一个组只给一个默认项写 checked。

```html
<fieldset>
  <legend>阅读时段（选一项）</legend>
  <p><input id="period-morning" name="period" type="radio" value="morning" checked>
    <label for="period-morning">上午</label></p>
  <p><input id="period-evening" name="period" type="radio" value="evening">
    <label for="period-evening">晚上</label></p>
</fieldset>
<!-- 选“晚上”应取消“上午”，但不应改变“纸质材料”。 -->
<!-- 用 Tab 进入单选组后，用方向键切换；复选框获得焦点后可用空格切换。 -->
```

## 9 默认状态、只读与禁用

### 9.1 重置恢复什么

默认状态是重置的起点，不等于用户修改后的当前状态：

- 文本 &lt;input&gt;：默认值来自 value 属性。
- &lt;textarea&gt;：默认文字来自元素内容。
- checkbox、radio：默认勾选来自 checked。
- &lt;option&gt;：默认选择来自 selected。

配套页同时包含这四种默认状态。改完输入、备注、勾选和下拉选项，再重置。

```html
<p><button type="reset">恢复所有默认状态</button></p>
<!-- 改安排、备注、勾选和下拉选项后重置，应回到源码给出的四种默认状态。 -->
<!-- 重置后仍然只读/禁用：reset 不会删除这些属性。 -->
```

配套文件：[scripts/10-form-structure-and-controls/states.html](scripts/10-form-structure-and-controls/states.html) · [浏览器预览](http://127.0.0.1:8010/scripts/10-form-structure-and-controls/states.html)

### 9.2 readonly 与 disabled

- readonly：只读。适用控件不能由用户编辑，但仍可聚焦；本身不排除具名字段的提交。
- disabled：禁用。控件不可操作、不能聚焦，也不参与提交。

readonly 适用于文本、数值、日期时间输入及 &lt;textarea&gt;，不适用于 &lt;select&gt;、checkbox、radio、range、color 或 file。两者都是布尔属性，关闭时删除属性，不能填写字符串 false。

```html
<p><label for="readonly-code">只读编号</label>
  <input id="readonly-code" name="code" type="text" value="DEMO-010" readonly></p>
<p><label for="disabled-place">禁用地点</label>
  <input id="disabled-place" name="place" type="text" value="练习室 B" disabled></p>
```

&lt;fieldset disabled&gt; 会禁用后代控件，但第一个 &lt;legend&gt; 内的控件不受这一个分组的禁用影响。它们仍可能被自己的 disabled 或其他祖先禁用。

```html
<fieldset disabled>
  <legend>暂不可编辑的补充安排
    <label><input id="legend-marker" name="marker" type="checkbox" value="yes">标记稍后查看</label>
  </legend>
  <p><label for="extra-place">补充地点</label>
    <input id="extra-place" name="extra_place" type="text" value="练习室 C"></p>
  <p><label for="extra-mode">补充方式</label>
    <select id="extra-mode" name="extra_mode"><option value="quiet">安静阅读</option></select></p>
</fieldset>
<!-- 第一个 legend 内的“标记稍后查看”仍可勾选；勾选它不会自动启用这一组。 -->
```

## 10 结果、度量与进度

### 10.1 &lt;output&gt;：操作结果

&lt;output&gt; 的 for 列出参与结果的元素 id，多个 id 用空格分隔。它只声明关系，不会自动计算；本例的辅助脚本在滑块变化时同步结果。

这里 reading-minutes 标识滑块，minutes-result 标识结果元素。即使 &lt;output&gt; 有 name，其内容也不会自动作为表单数据提交。

```html
<p><label for="reading-minutes">阅读时长（0～60 分钟）</label>
  <input id="reading-minutes" name="minutes" type="range" min="0" max="60" value="30"></p>
<p><label for="minutes-result">当前选择</label>
  <output id="minutes-result" name="minutes_result" for="reading-minutes">30</output> 分钟</p>
<!-- 检查：滑块初始为 30，调到 45 后 output 应显示 45；重置后两者都恢复 30。 -->
```

### 10.2 &lt;meter&gt; 与 &lt;progress&gt;

- &lt;meter&gt;：表示已知范围内的量值，如电量；min、max 定义范围，value 给出当前值。
- &lt;progress&gt;：表示任务已完成量；起点固定为 0，没有 min 属性，max 为正数，value 在 0 到 max 之间。
- 省略 &lt;progress&gt; 的 value：进度不确定；value="0" 则表示确定的零进度。

这些元素可以放在表单外。本例使用静态示意值，同时提供标签和可见说明，不连接电池或后台任务。

```html
<p><label for="battery-level">电量示意（65%）</label>
  <meter id="battery-level" min="0" max="100" value="65">65%</meter></p>
<p><label for="review-progress">复习进度示意（已完成 3 / 5 项）</label>
  <progress id="review-progress" max="5" value="3">3 / 5 项</progress></p>
<p><label for="unknown-progress">进度不确定的样式示意</label>
  <progress id="unknown-progress">进度不确定</progress></p>
```

## 本章小结

- &lt;label&gt; 的 for 对应 id，name 命名数据项；分组标题与单个控件标签各司其职。
- 按用途选择输入类型。radio 按组互斥，checkbox 独立；&lt;select&gt; 提供选择，&lt;datalist&gt; 只提供建议。
- 默认值、当前值与选中状态要分清；重置恢复默认，readonly 与 disabled 的焦点和提交行为不同。
- &lt;output&gt; 表达结果，&lt;meter&gt; 表达度量，&lt;progress&gt; 表达进度。

### 提示

第一题分别检查 name、value 和 checked。第三题先处理外层 &lt;fieldset&gt; 的 disabled，再设置单个控件状态。练习副本用后可删除，保留原页面与素材。

### 参考解析

（1）新增复选项沿用方式组的 name，使用独立 id 和 value；新增下午单选项沿用时段组 name，并使用独立 id、value 和关联标签。移除旧时段的 checked，再给下午设置 checked。重置恢复标记中的默认状态，不是恢复重置前的最后一次选择；材料组的 name 保持原样。

（2）新 option 写 value="states" 和 selected，并移除原默认选项的 selected。补充主题的 input 通过 list 引用 datalist 的 id；datalist 提供建议，不限制输入必须匹配某项，所以“自己的题目”仍是合法输入。

（3）移除编号的 readonly；先移除补充分组 fieldset 的 disabled，再给地点添加 readonly。readonly 文本框仍可聚焦，但不可由用户编辑；保留外层 disabled 时，设置 readonly 不能使其恢复聚焦。移除分组 disabled 也会启用其中其他控件，包括补充方式，不能误以为只影响地点。

（4）复习 progress 的 value 为 4、max 为 5，并同步修改相邻说明和标签。给不确定 progress 添加 value="0"（可明确 max="5"），其进度变为 0，而不是缺少 value 时的不确定状态。电量是已知范围内的测量值，继续使用 meter，不改成任务完成进度。
