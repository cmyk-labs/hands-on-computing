## 1 元素与标签

用 &lt;p&gt; 标记段落时，&lt;p&gt; 是开始标签，&lt;/p&gt; 是结束标签，中间是内容。整个段落是元素（element）；标签（tag）是源码中标记边界的写法。

下面的 &lt;strong&gt; 表示重要内容，位于段落内部。先结束内层元素，再结束外层元素。

```html
<!-- strong 在 p 内，先结束 strong，再结束 p。 -->
<p>保存前请<strong>检查文件名</strong>。</p>
```

配套文件：[scripts/02-syntax-and-document-structure/syntax.html](scripts/02-syntax-and-document-structure/syntax.html) · [浏览器预览](http://127.0.0.1:8002/scripts/02-syntax-and-document-structure/syntax.html#elements)

HTML 标签名不区分 ASCII 字母大小写，本章统一用小写。缩进方便阅读；普通段落在默认样式下通常合并连续空格和源码换行，不能靠源码换行控制排版。

## 2 文档声明与基本骨架

独立页面以 &lt;!doctype html&gt; 开头，再用 &lt;html&gt; 包住头部和正文。DOCTYPE 是文档类型声明，不是元素；它帮助浏览器使用标准模式，不负责检查源码是否合规。

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;html&gt; | 文档根元素 | 包含头部和正文 |
| &lt;head&gt; | 文档头部 | 保存描述页面的元数据 |
| &lt;body&gt; | 文档正文 | 容纳页面内容 |
| &lt;title&gt; | 文档标题 | 为浏览器标签页等位置提供名称 |
| &lt;meta&gt; | 元数据元素 | 本例声明 UTF-8 编码 |
| &lt;h1&gt; | 一级标题 | 标记正文中的主标题 |

文件本身也按 UTF-8 保存。lang="zh-CN" 声明中文，第 8 节说明其继承关系；编码与资源引入见[元数据与资源引入](03-元数据与资源引入.ipynb)。

```html
<!doctype html>
<html lang="zh-CN">
<head>
  <meta charset="utf-8">
  <title>文档结构实验</title>
</head>
<body>
  <!-- 标签页标题应为“文档结构实验”，正文标题应为“我的语法练习”。 -->
  <h1>我的语法练习</h1>
  <p>保存前请<strong>检查文件名</strong>。</p>
  <!-- 在 Elements 中展开 html：应先看到 head，再看到 body。 -->
  <!-- 在查看页面源代码中能看到这些注释，页面正文中看不到。 -->
</body>
</html>
```

配套文件：[scripts/02-syntax-and-document-structure/document.html](scripts/02-syntax-and-document-structure/document.html) · [浏览器预览](http://127.0.0.1:8002/scripts/02-syntax-and-document-structure/document.html)

初学时显式写全 &lt;html&gt;、&lt;head&gt;、&lt;body&gt; 及其结束标签。HTML 只允许在特定条件下省略某些标签；省略标签不表示元素不存在。

## 3 属性与引号

属性（attribute）写在开始标签内，为元素补充信息。title="HTML syntax" 中，title 是属性名，HTML syntax 是属性值，引号界定值的范围。

- 属性之间用空白分隔；同一开始标签不能重复写同名属性，结束标签不能带属性。
- 属性值可以使用成对的英文双引号或单引号。本章统一加引号，含空格的值不能裸写，中文弯引号不能代替语法引号。
- HTML 属性名不区分 ASCII 字母大小写；属性值是否区分大小写由各属性决定。

title 属性提供补充提示，不同于 &lt;head&gt; 中的 &lt;title&gt; 元素。提示的呈现依赖浏览器与操作方式，必读内容仍需写在正文。

```html
<!-- 悬停前两段可比较提示；单引号、双引号都能包围带空格的值。 -->
<p title="HTML syntax">语法练习：双引号包围属性值。</p>
<p title='HTML syntax'>语法练习：单引号包围属性值。</p>
<p title='请先按 "保存"'>请先按“保存”，再刷新页面。</p>
```

配套文件：[scripts/02-syntax-and-document-structure/syntax.html](scripts/02-syntax-and-document-structure/syntax.html) · [浏览器预览](http://127.0.0.1:8002/scripts/02-syntax-and-document-structure/syntax.html#attributes)

## 4 布尔属性：用存在表示真

布尔属性（boolean attribute）用存在表示真，用缺失表示假。disabled 可禁用本例的独立 &lt;button&gt;，它不是全局属性。type="button" 指定普通按钮，本例没有绑定动作。

以下三种写法都有效：只写 disabled、写 disabled=""、写 disabled="disabled"。显式值只允许空字符串或与属性名匹配且无首尾空白的字符串，匹配时忽略 ASCII 字母大小写。

```html
<!-- 前三种是合规写法；预期三个按钮都被禁用。 -->
<p><button type="button" disabled>只写属性名</button></p>
<p><button type="button" disabled="">属性值为空</button></p>
<p><button type="button" disabled="disabled">属性值等于属性名</button></p>
```

配套文件：[scripts/02-syntax-and-document-structure/boolean-attributes.html](scripts/02-syntax-and-document-structure/boolean-attributes.html) · [浏览器预览](http://127.0.0.1:8002/scripts/02-syntax-and-document-structure/boolean-attributes.html)

false 和 true 都不是布尔属性允许的值。下面的反例虽然不合规，仍因属性存在而被禁用；应删除整个属性来表示假。

```html
<!-- 故意保留反例：false 不是合规布尔属性值，但属性仍存在。 -->
<p><button type="button" disabled='false'>反例：值为 false</button></p>
<!-- 预期：只有第五个按钮可通过 Tab 获得焦点；按钮没有绑定动作。 -->
<p><button type="button">没有 disabled 属性</button></p>
```

配套文件：[scripts/02-syntax-and-document-structure/boolean-attributes.html](scripts/02-syntax-and-document-structure/boolean-attributes.html) · [浏览器预览](http://127.0.0.1:8002/scripts/02-syntax-and-document-structure/boolean-attributes.html)

“布尔”描述取值规则，“全局”描述适用范围。不要把存在性规则推广到所有开关形式的属性，例如全局属性 draggable 是枚举属性，按 true、false 等值解释。表单中由祖先造成禁用的情况留在表单章节。

## 5 空元素与尾随斜杠

空元素（void element）在 HTML 标记中不能容纳内容，也不能写结束标签。常见例子如下；内容暂时为空的 &lt;p&gt;&lt;/p&gt; 仍是普通元素。

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;br&gt; | 换行 | 标记内容中的换行 |
| &lt;hr&gt; | 主题分隔 | 标记话题转折 |
| &lt;img&gt; | 图像 | 嵌入图像 |
| &lt;meta&gt; | 元数据 | 声明编码等信息 |
| &lt;input&gt; | 输入控件 | 接收输入 |

HTML 中 &lt;br /&gt; 的尾随斜杠没有自闭合作用，本章正常写法使用 &lt;br&gt;。

```html
<!-- 两段均分成两行；第二个 br 的尾随斜杠不产生额外效果。 -->
<p>第一行<br>第二行</p>
<p>第一行<br />第二行</p>
```

配套文件：[scripts/02-syntax-and-document-structure/syntax.html](scripts/02-syntax-and-document-structure/syntax.html) · [浏览器预览](http://127.0.0.1:8002/scripts/02-syntax-and-document-structure/syntax.html#void)

普通元素也不能用 &lt;div /&gt; 代替 &lt;div&gt;&lt;/div&gt;。斜杠若紧跟未加引号的属性值，还可能成为值的一部分。SVG、MathML 的外来元素规则不同，这里只讨论 HTML 元素。

## 6 嵌套与内容模型

直接包含另一元素的是父元素，被直接包含的是子元素。嵌套要从内到外关闭，还要满足内容模型（content model），即该元素允许包含什么。

- 流内容（flow content）：正文中广泛使用的内容类别，如 &lt;p&gt;、&lt;div&gt;。
- 短语内容（phrasing content）：文本及段落内部的标记，如 &lt;strong&gt;、&lt;span&gt;。

一个元素可以属于多个内容类别。&lt;p&gt; 要求短语内容，不能直接包含 &lt;div&gt; 或另一个 &lt;p&gt;；通用容器 &lt;div&gt; 则可以包含段落。&lt;span&gt; 用于文本范围内的通用分组。

```html
<!-- 两个 p 同属外层 div；span 在 strong 内，strong 在第一个 p 内。 -->
<div>
  <p>保存前<strong>检查<span>文件名</span></strong>。</p>
  <p>确认后再刷新页面。</p>
</div>
```

配套文件：[scripts/02-syntax-and-document-structure/syntax.html](scripts/02-syntax-and-document-structure/syntax.html) · [浏览器预览](http://127.0.0.1:8002/scripts/02-syntax-and-document-structure/syntax.html#nesting)

“显示时是否换行”不能决定嵌套是否合规。内容关系与视觉布局的区别见[语义化页面结构](05-语义化页面结构.ipynb)。

## 7 注释与字符引用

### 7.1 给源码加注释

注释从 &lt;!-- 开始，到 --&gt; 结束，不成为页面正文。它仍能在源代码中读取，因此不能用来隐藏秘密信息。

注释不能嵌套，也不能插进开始标签的属性列表；临时注释掉元素时，不要包住本来已经含有注释的大段源码。

```html
<!-- 这是给阅读源码的人看的说明。 -->
<p>当前版本。</p>
<!-- <p>暂不显示的备选文字。</p> -->
<!-- 本组正文只有“当前版本。”，源码仍保留备选文字。 -->
```

配套文件：[scripts/02-syntax-and-document-structure/syntax.html](scripts/02-syntax-and-document-structure/syntax.html) · [浏览器预览](http://127.0.0.1:8002/scripts/02-syntax-and-document-structure/syntax.html#comments)

### 7.2 显示尖括号与特殊字符

要显示标签文字，不能直接把它写成标记。字符引用（character reference）以 &amp; 开头，用来表示相应字符；本章统一写全末尾分号。

- &amp;lt;、&amp;gt;：分别表示小于号 &lt;、大于号 &gt;。
- &amp;amp;：表示与号 &amp;。
- &amp;quot;、&amp;apos;：分别表示英文双引号和单引号。
- &amp;nbsp;：不换行空格，不用于堆出页面缩进。

字符引用可用名称或数字码位。&amp;copy;、&amp;#169;、&amp;#xA9; 都表示 ©；169 与 A9 分别是该字符 Unicode 码位的十进制、十六进制表示，x 是十六进制形式的标记。具名引用的名称区分大小写。

```html
<!-- 第一段显示标签文字，不会生成额外的 p 元素。 -->
<p>&lt;p&gt;标签文字&lt;/p&gt;</p>
<p>比较：3 &lt; 5 &amp; 8 &gt; 6。</p>
<!-- 三个版权符号相同；分别使用名称、十进制码位、十六进制码位。 -->
<p>&copy; &#169; &#xA9;</p>
```

配套文件：[scripts/02-syntax-and-document-structure/syntax.html](scripts/02-syntax-and-document-structure/syntax.html) · [浏览器预览](http://127.0.0.1:8002/scripts/02-syntax-and-document-structure/syntax.html#characters)

双引号包围的属性值需要双引号时，可写 &amp;quot;。要显示 &amp;lt; 这串写法本身，就把开头的 &amp; 再转义一次。普通中文可以直接按 UTF-8 保存。

```html
<p title="请先按 &quot;保存&quot;">请先按“保存”，再刷新页面。</p>
<p>显示引用写法本身：&amp;lt;</p>
```

## 8 全局属性

全局属性可以用于所有 HTML 元素，是否产生可见效果取决于场景。

- lang：内容语言，使用 BCP 47 语言标签；本例 zh-CN 表示中国地区的中文，en 表示英语。
- id：元素标识符，本章普通文档中不能重复，值非空且不含 ASCII 空白。
- class：以空白分隔的类名列表；一个元素可有多个类名，不同元素可共享类名。
- data-*：页面自定义数据，星号表示自定后缀，实际写成 data-topic 等名称。
- title：元素的补充说明，见第 3 节。

页面的主要语言写在 &lt;html&gt; 上；本例子元素未单独声明时继承父元素的语言。局部英文另设 lang，不会自动翻译文字，也不改变编码。

```html
<!-- intro 和 follow-up 是两个不同的 id；两段共享 note 类。 -->
<p id="intro" class="note important" data-topic="syntax" data-level="beginner">
  学习记录：<span lang="en">Read the source.</span>
</p>
<p id="follow-up" class="note" title="观察元素属性">检查保存后的页面。</p>
```

配套文件：[scripts/02-syntax-and-document-structure/global-attributes.html](scripts/02-syntax-and-document-structure/global-attributes.html) · [浏览器预览](http://127.0.0.1:8002/scripts/02-syntax-and-document-structure/global-attributes.html)

intro、follow-up 是不同标识符；note、important 是作者自定类名，用空格分开，不用逗号。data-topic 的 syntax 和 data-level 的 beginner 都是自定数据值。没有相应 CSS 或脚本时，class 与 data-* 不会自动增加样式或交互。

自定义数据名使用 data- 前缀和非空后缀，本章采用小写单词与连字符。已有标准属性时优先使用它；data-disabled="false" 不会产生 disabled 的禁用效果。

## 9 浏览器恢复不等于源码合规

浏览器把 HTML 解析成文档对象模型（DOM），并按规则恢复许多错误。看到文字并不证明嵌套正确；比较“查看页面源代码”与 Elements 中的父子关系。

下面故意把 &lt;div&gt; 写进段落：

```html
<!-- 故意保留错误。预期：inner-block 不会成为 intro 的子元素。 -->
<p id="intro">前半句<div id="inner-block">中间容器</div>后半句</p>
<!-- 解析 div 开始标签时，前一个 p 会结束。 -->
<!-- 后半句在该 p 外；多余的 p 结束标签还会使这里产生一个空 p。 -->
```

配套文件：[scripts/02-syntax-and-document-structure/parser-recovery.html](scripts/02-syntax-and-document-structure/parser-recovery.html) · [浏览器预览](http://127.0.0.1:8002/scripts/02-syntax-and-document-structure/parser-recovery.html)

另一处反例把普通元素写成带尾随斜杠的开始标签：

```html
<!-- 故意保留错误：这个 div 不会因尾随斜杠而关闭。 -->
<div id="not-closed" />
<p id="after-slash">我仍然位于 not-closed 容器中。</p>
</div>
```

## 本章小结

- 标签界定元素边界；嵌套还要满足内容模型。
- DOCTYPE、文档标题和正文标题各有职责；语言声明与编码也不同。
- 布尔属性看是否存在；空元素没有结束标签，普通元素不能靠斜杠关闭。
- id 定位、class 分组、data-* 保存自定数据；字符引用让标签文字安全显示。
