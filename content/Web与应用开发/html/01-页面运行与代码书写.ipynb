# 页面运行与代码书写

学习目标：能保存并打开 HTML 页面，区分 HTML、CSS 与 JavaScript 的作用，使用本地 HTTP 服务和开发者工具定位基本问题。

前置知识：会创建文件、使用文本编辑器和浏览器；无需编程基础。

适用范围：HTML Living Standard；浏览器操作以 Windows 下的 Chrome 或 Edge 为例。

环境准备：[环境配置与运行](README.md)。

配套脚本：位于 scripts/01-pages-and-code/。

1. [index.html](scripts/01-pages-and-code/index.html)：保存、打开和修改页面。
2. [roles.html](scripts/01-pages-and-code/roles.html)：三种技术的配合、源代码与当前页面的区别。
3. [styles.css](scripts/01-pages-and-code/styles.css)、[script.js](scripts/01-pages-and-code/script.js)：页面样式与按钮行为。

## 打开配套页面

Step 1：在已激活 Python 环境的终端中，从项目根目录进入本技术目录。

```bash
cd content/Web与应用开发/html
```

Step 2：启动本章预览服务。

```bash
python -m http.server 8000 --bind 127.0.0.1
```

Step 3：打开[本章示例首页](http://127.0.0.1:8000/scripts/01-pages-and-code/index.html)。

127.0.0.1 表示本机，8000 是服务端口；本例服务根目录为 content/Web与应用开发/html。

关闭浏览器标签页不会停止服务；端口占用时先关闭自己已有的同端口预览。

Step 4：在服务终端按 Ctrl+C 停止服务。

## 1 HTML、CSS 与 JavaScript

- HTML（超文本标记语言）：描述页面内容、结构与语义。
- CSS（层叠样式表）：设置颜色、字体和布局。
- JavaScript：编写脚本，响应操作并更新页面。

普通 HTML 页面可以不包含 CSS 和 JavaScript；链接跳转、表单提交等也有 HTML 原生行为。先看标题和段落：

```html
<h1>我的学习笔记</h1>
<p>我正在学习用 HTML 组织网页内容。</p>
<!-- 浏览器显示标题与段落；标签用于描述结构，不作为正文显示。 -->
```


配套文件：[scripts/01-pages-and-code/index.html](scripts/01-pages-and-code/index.html) · [浏览器预览](http://127.0.0.1:8000/scripts/01-pages-and-code/index.html)

## 2 标签总览

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;html&gt; | 根元素 | 包含文档内容 |
| &lt;head&gt; | 文档元数据 | 放置标题、编码和资源声明 |
| &lt;body&gt; | 文档主体 | 放置页面内容 |
| &lt;title&gt; | 文档标题 | 设置浏览器标签页等位置使用的标题 |
| &lt;meta&gt; | 元数据 | 声明编码等文档信息 |
| &lt;h1&gt; | 一级标题 | 标记页面中的一级标题 |
| &lt;p&gt; | 段落 | 组织一段文字 |
| &lt;a&gt; | 超链接元素 | 配合 href 指向目标 |
| &lt;link&gt; | 外部资源关联 | 本例用于引入样式表 |
| &lt;script&gt; | 脚本元素 | 本例用于引入脚本 |
| &lt;button&gt; | 按钮 | 供用户触发操作 |

本章先认识用途，元素、标签与属性的具体语法见[语法与文档结构](02-语法与文档结构.ipynb)。

## 3 保存并打开 HTML 文件

用文本编辑器打开 index.html。完整页面如下，缩进帮助辨认元素的包含关系。

- &lt;!doctype html&gt; 是文档声明，不是普通元素。
- lang="zh-CN" 声明文档语言；charset="utf-8" 声明编码，文件本身也要保存为 UTF-8。
- viewport 声明用于移动设备的视口设置，细节见[元数据与资源引入](03-元数据与资源引入.ipynb)。

```html
<!doctype html>
<html lang="zh-CN">
<head>
  <meta charset="utf-8">
  <meta name="viewport" content="width=device-width, initial-scale=1">
  <title>我的第一个 HTML 页面</title>
</head>
<body>
  <!-- 浏览器标签页看 title，页面内的大标题看 h1。 -->
  <h1>我的学习笔记</h1>
  <p>我正在学习用 HTML 组织网页内容。</p>
  <p><a href="roles.html">查看 HTML、CSS 与 JavaScript 的配合</a></p>
</body>
</html>
```


配套文件：[scripts/01-pages-and-code/index.html](scripts/01-pages-and-code/index.html) · [浏览器预览](http://127.0.0.1:8000/scripts/01-pages-and-code/index.html)

保存与修改步骤：

- 在浏览器按 Ctrl+O，选择这个文件，可以直接查看；此时地址以 file: 开头。
- 修改 &lt;h1&gt; 中的文字，保存，再按 F5 刷新浏览器。
- 没有变化时，先核对保存的文件与打开的路径，再检查文件是否误存为 index.html.txt。

## 4 观察样式与脚本

### 4.1 HTML 引入配套文件

roles.html 通过下面两行分别引入同目录的 styles.css 与 script.js。&lt;link&gt; 放在 &lt;head&gt; 中；本例的 &lt;script&gt; 放在页面元素之后。

### 4.2 分辨三种技术的作用

- HTML 提供标题、段落和按钮。id="status" 标识段落，供脚本查找。
- styles.css 限制正文宽度，并把 &lt;h1&gt; 的文字设为蓝色。
- script.js 监听按钮点击后修改段落文字。本章只观察作用，不要求先掌握 CSS 或 JavaScript 语法。

下面是样式文件中设置标题颜色的部分：

## 5 本地文件与 HTTP 页面

- file: 直接打开磁盘文件，适合先查看简单页面。
- http: 通过服务请求资源；本例的网页、样式和脚本都从本地服务获取。

模块脚本或网络请求等功能可能受到本地文件访问限制，后续实验使用 HTTP 入口。此处的 Python 仅提供本地学习服务，不处理页面中的 JavaScript，也不作为生产服务器。

访问地址中的 /scripts/01-pages-and-code/index.html 对应服务根目录 content/Web与应用开发/html 下的同名相对路径。路径和端口应与启动方式一致。


配套文件：[scripts/01-pages-and-code/index.html](scripts/01-pages-and-code/index.html) · [浏览器预览](http://127.0.0.1:8000/scripts/01-pages-and-code/index.html)

## 6 查看源代码与 DOM

DOM（文档对象模型）用对象和树形关系表示当前文档。浏览器解析源码建立 DOM，脚本还可以修改它。

在 roles.html 中依次操作：

- 按 Ctrl+U 查看源代码，找到下面的段落。
- 回到页面，点击“更新文字”，再右键段落选择“检查”，查看 Elements（元素）中的文字。
- 在 Elements 中临时修改文字，再刷新。未启用本地覆盖或工作区映射时，这类修改不会保存回磁盘文件。

```html
<p id="status">还没有点击按钮。</p>
<!-- 源码保留初始文字；按钮只改变当前 DOM，刷新后重新显示初始文字。 -->
```


配套文件：[scripts/01-pages-and-code/roles.html](scripts/01-pages-and-code/roles.html) · [浏览器预览](http://127.0.0.1:8000/scripts/01-pages-and-code/roles.html)

## 7 定位加载与显示问题

按 F12 打开开发者工具，按问题选择面板：

- Elements（元素）：查看当前 DOM 与生效的样式。
- Console（控制台）：查看脚本输出和错误位置。
- Network（网络）：查看请求地址、状态码和响应内容；先打开面板，再刷新页面。

在 roles.html 中检查 styles.css、script.js 是否加载，并在 Console 中查找“交互脚本已加载”。

- 正常获取资源通常返回 200；缓存复用可能出现 304 或缓存标记。
- 样式缺失时，先检查样式文件请求与 &lt;link&gt; 中的路径。
- 按钮无效时，检查脚本请求与 Console 中的错误。
- [不存在的页面](http://127.0.0.1:8000/scripts/01-pages-and-code/missing-page.html)应返回 404；这与服务未启动导致的连接失败不同。

资源成功返回只说明加载成功，还需要检查代码内容与实际行为。

## 本章小结

- HTML 组织内容，CSS 设置外观，JavaScript 负责脚本行为。
- 保存文件后刷新页面；分清磁盘源码与浏览器当前 DOM。
- Elements 看结构，Console 看脚本，Network 看请求。
- HTTP 地址应与服务根目录对应，用完在终端停止服务。
