# 链接与 URL

学习目标：能编写页面跳转、片段定位和下载链接，读懂 URL，按基准地址判断路径并排查失效链接。

前置知识：HTML 元素与属性、基本文档结构、文件和目录操作。

适用范围：WHATWG HTML Living Standard；使用现代浏览器查看配套页面。

环境准备：[环境配置与运行](README.md)。

配套脚本：位于 scripts/06-links-and-urls/。

1. [index.html](scripts/06-links-and-urls/index.html)、[pages/guide.html](scripts/06-links-and-urls/pages/guide.html)：基本链接、路径与片段定位。
2. [target.html](scripts/06-links-and-urls/target.html)：打开位置与来源信息。
3. [actions.html](scripts/06-links-and-urls/actions.html)、[notes.txt](scripts/06-links-and-urls/notes.txt)：邮件、电话与下载。
4. [url-lab.html](scripts/06-links-and-urls/url-lab.html)：查询参数、编码和 URL 组成。
5. [base.html](scripts/06-links-and-urls/base.html)、[base-target.html](scripts/06-links-and-urls/base-target.html)、[base-fragment.html](scripts/06-links-and-urls/base-fragment.html)：基准 URL、默认打开位置和片段目标。

## 1 标签总览

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;a&gt; | 超链接元素 | 配合 href 链接到页面、文件或指定位置 |
| &lt;base&gt; | 基准 URL 与默认目标 | 设置相对 URL 的解析基准或默认打开位置 |

属性和属性值在下面分别介绍；&lt;base&gt; 作为补充内容放在第 9 节。

## 2 &lt;a&gt;：创建超链接

href 属性指定目标地址；开始标签 &lt;a&gt; 与结束标签 &lt;/a&gt; 之间是链接内容。不写 href 时，该元素表示链接占位，不提供原生跳转功能。

- 链接文字应说明目的地，如“打开出行指南”，避免只写“点击这里”。
- 导航使用链接；提交表单等动作使用 &lt;button&gt;。链接内不要嵌套其他链接、按钮或带 tabindex 的元素。

```html
<p><a href="pages/guide.html">打开出行指南</a></p>
<p><a>扩展阅读尚未提供</a></p>
<!-- 在示例首页按 Tab 聚焦第一条链接，按 Enter 打开指南；第二项是占位文字。 -->
```

## 3 路径与片段

### 3.1 相对与绝对 URL

本例目录如下。没有 &lt;base&gt; 时，普通页面中的相对 URL 以该页面自身的 URL 为基准；基准 URL（base URL）就是解析相对地址时参照的地址。

```text
content/Web与应用开发/html/                       ← 本地服务根目录
└── scripts/06-links-and-urls/
    ├── index.html                 ← 当前页面
    └── pages/
        └── guide.html
```

- pages/guide.html：从当前页面所在目录进入 pages/。
- ./pages/guide.html：这里与上一种写法等价，./ 表示当前目录。
- /scripts/06-links-and-urls/pages/guide.html：从当前站点根路径开始；本例对应 content/Web与应用开发/html。
- 绝对 URL：写明方案、主机等组成，见下面第三条链接。

HTML URL 使用 /，不使用 Windows 路径中的反斜杠。站点根路径开头的 / 不代表磁盘根目录。

```html
<p><a href="pages/guide.html">打开出行指南</a></p>
<p><a href="/scripts/06-links-and-urls/pages/guide.html">打开指南（站点根路径）</a></p>
<p><a href="http://127.0.0.1:8006/scripts/06-links-and-urls/pages/guide.html">打开指南（绝对 URL）</a></p>
<!-- 从示例首页打开：三条链接都到达同一个 guide.html。 -->
```

### 3.2 片段与 id：定位页面中的元素

id 是全局属性，同一文档中的值必须唯一。URL 中 # 后面的部分称为片段标识符（fragment identifier）；本例 #packing 对应 id 为 packing 的元素。

- \#packing：定位当前页面的目标。
- pages/guide.html#packing：先到指南页，再定位目标。
- 片段不随 HTTP 请求发送；修改基准 URL 后，目标页面也可能改变，见第 9 节。

在指南页中：

```html
<a href="#packing">跳到出发清单</a>
<h2 id="packing">出发清单</h2>
<!-- 点击后，标题进入视口，地址以 #packing 结尾。 -->
```

## 4 target：在哪里打开

target 属性指定打开目标的位置。未设置其他默认值时，使用当前页面。

- \_self：当前窗口或框架。
- \_blank：新标签页或新窗口，具体由浏览器设置等因素决定。

```html
<p><a href="pages/guide.html" target="_self">打开指南（当前页）</a></p>
<p><a href="pages/guide.html" target="_blank">打开指南（新窗口或标签页）</a></p>
<!-- 两条链接的目标相同，只改变打开位置。 -->
```


配套文件：[scripts/06-links-and-urls/target.html](scripts/06-links-and-urls/target.html) · [浏览器预览](http://127.0.0.1:8006/scripts/06-links-and-urls/target.html)

补充：\_parent 指父级框架，没有父级时等同 \_self；\_top 指最外层窗口，没有上级时等同 \_self。普通名称可指定同名窗口或框架；嵌入页面的用法见[音视频与嵌入内容](08-音视频与嵌入内容.ipynb)。

## 5 rel：链接关系与来源信息

rel 属性描述链接关系，多个值用空格分隔。下面两个值用于新页面与来源页面之间的信息控制：

- noopener：不向新页面提供原窗口的脚本引用，阻止它通过该引用操作原窗口。
- noreferrer：不向目标提供来源信息（referrer），同时具有 noopener 效果。

来源信息表示从哪里跳转而来，其具体内容还受来源策略和浏览器设置影响。现代浏览器中，&lt;a&gt; 的 target="\_blank" 已隐式采用 noopener；显式写出可表明意图。

```html
<p><a href="target.html" target="_blank" rel="noopener">查看窗口信息（新页面，noopener）</a></p>
<p><a href="target.html" target="_blank" rel="noreferrer">查看窗口信息（新页面，noreferrer）</a></p>
<!-- 比较新页的“窗口信息”：两者均无原窗口引用，第二项还没有来源地址。 -->
```

## 6 href：邮件和电话地址

href 也可以使用其他 URL 方案（scheme），表示地址的处理方式：

- mailto:：交给邮件应用处理。
- tel:：交给支持电话地址的应用处理。

是否打开应用及如何处理，取决于设备和应用配置。以下地址仅用于阅读写法：

```html
<p><a href="mailto:learner@example.invalid">邮件地址示例</a></p>
<p><a href="tel:+12025550123">电话地址示例</a></p>
```

## 7 download：下载文件

download 属性提示浏览器下载目标资源，可以单独写出，也可以给出建议文件名。

- 本例使用同源文件：页面与文件 URL 的方案、主机、端口相同。
- 实际保存方式与文件名还受浏览器和服务器响应头影响；跨源地址不能仅靠 download 保证下载行为。

```html
<p><a href="notes.txt">打开练习材料（TXT）</a></p>
<p><a href="notes.txt" download="link-notes.txt">下载练习材料（TXT）</a></p>
<!-- 第二项建议保存为 link-notes.txt；下载后核对内容中的 html-links-06。 -->
```

## 8 URL 组成与查询参数

### 8.1 读懂一个 URL

下面是普通 HTTP URL 的组成示例，并非所有 URL 方案都采用相同结构。

```text
http://127.0.0.1:8006/scripts/06-links-and-urls/url-lab.html?topic=html&mode=read#details
```

- 方案（scheme）：http，指定本例使用 HTTP。
- 主机（host）：127.0.0.1，本例的本机地址。
- 端口（port）：8006，本地服务监听的端口。
- 路径（path）：/scripts/06-links-and-urls/url-lab.html，目标资源的路径。
- 查询字符串（query）：topic=html&mode=read，传给目标的查询数据。
- 片段（fragment）：details，本例对应页面中的 id。

配套页面的“查看 URL 组成”可展开观察这些部分。


配套文件：[scripts/06-links-and-urls/url-lab.html](scripts/06-links-and-urls/url-lab.html) · [浏览器预览](http://127.0.0.1:8006/scripts/06-links-and-urls/url-lab.html)

### 8.2 查询参数：名称与值

本例采用常见的“名称=值”写法：

- ? 开始查询字符串。
- = 连接名称和值，如 topic=html。
- & 分隔多个参数；写在下面的 HTML 属性中时，用字符引用 &amp;amp; 表示 &。

topic、mode 是本例自定的参数名，具体作用由程序决定；mode=read 不会自动产生阅读模式。下面的链接沿用当前页面路径，替换查询与片段。

```html
<a href="?topic=html&amp;mode=read#details">查看 html 查询参数</a>
<!-- 点击后，页面显示 topic = html、mode = read。 -->
```


配套文件：[scripts/06-links-and-urls/url-lab.html](scripts/06-links-and-urls/url-lab.html) · [浏览器预览](http://127.0.0.1:8006/scripts/06-links-and-urls/url-lab.html)

配套脚本只负责读取并显示参数，使用浏览器的 URL API；本节无需先学习其 JavaScript 实现。

### 8.3 参数中的空格、& 与 +

参数值中的特殊字符需要与分隔符区分：

- %20：空格，例如 HTML%20links 表示 HTML links。
- %26：作为参数内容的 &，例如 A%26B 表示一个完整的 A&B。
- %2B：字面加号。本例使用的 URLSearchParams 会把未编码的 + 解释为空格。

这些写法属于百分号编码（percent-encoding）。&amp;amp; 则是 HTML 字符引用：前者处理 URL 数据，后者在 HTML 中表示字符 &。

```html
<p><a href="?topic=HTML%20links&amp;mode=read#details">查看含空格的参数</a></p>
<p><a href="?topic=A%26B&amp;mode=read#details">查看含 &amp; 的参数</a></p>
<!-- topic 分别显示 HTML links、A&B；两次 mode 都保持 read。 -->
```

## 9 &lt;base&gt;：修改默认值（补充）

&lt;base&gt; 放在 &lt;head&gt; 中，一个文档最多一个，至少设置 href 或 target 之一。

- href：设置文档的基准 URL；带此属性时，应位于其他带 URL 属性的元素之前。
- target：设置链接和表单导航的默认打开位置；带此属性时，应位于表示超链接的元素之前。

通常沿用页面本身的基准即可。下面用三个独立页面观察修改默认值的影响。

### 9.1 href：设置基准目录

在 base.html 的 &lt;head&gt; 中设置基准，页面中的 guide.html 就从 pages/ 解析。

```html
<base href="pages/">
```

下面的链接放在同一文件的 &lt;body&gt; 中：

```html
<a href="guide.html">打开出行指南</a>
<!-- 打开 scripts/06-links-and-urls/pages/guide.html。 -->
```


配套文件：[scripts/06-links-and-urls/base.html](scripts/06-links-and-urls/base.html) · [浏览器预览](http://127.0.0.1:8006/scripts/06-links-and-urls/base.html)

目录基准末尾的 / 会影响解析结果。以上面的 base.html 为起点：

- 基准 pages/ 加上 guide.html → pages/guide.html。
- 基准 pages 加上 guide.html → guide.html，最后的 pages 段被替换。

&lt;base&gt; 也影响图片等相对 URL；url-lab.html 的“补充：基准 URL 末尾的斜杠”可展开对照。

### 9.2 target：设置默认打开位置

在 base-target.html 的 &lt;head&gt; 中只设置 target，基准 URL 保持不变。

```html
<base target="_blank">
```

链接可以沿用默认值，也可以设置自己的 target：

```html
<p><a href="pages/guide.html">打开指南（新窗口或标签页）</a></p>
<p><a href="pages/guide.html" target="_self">打开指南（当前页）</a></p>
<!-- 第一项沿用默认 target，第二项覆盖它。 -->
```

## 10 排查失效链接

- 无法连接本机地址：检查服务是否启动、端口是否正确。
- 返回 404：检查文件是否存在，URL 是否对应服务根目录下的实际位置。
- 页面打开但没有定位：检查片段与 id 是否一致，以及 &lt;base&gt; 是否改变了目标。
- 本机正常，部署后失效：检查路径大小写与部署位置。

Windows 文件名通常不区分大小写，Linux 通常区分；Windows 也支持按目录启用大小写敏感。链接中的路径始终按实际文件名书写。

```html
<a href="pages/missing-guide.html">不存在的文件（404 反例）</a>
<a href="pages/Guide.html">错误大小写路径（平台差异）</a>
<!-- 第一项返回 404；第二项能否打开取决于服务与文件系统的大小写处理。 -->
```

## 本章小结

- href 指定目标；相对地址要先找基准，跨页定位使用“路径#id”。
- target 指定打开位置，rel 描述链接关系，download 提示下载。
- 查询中的分隔符与参数内容要分清；&amp;amp; 与 %26 不能混用。
- &lt;base&gt; 可能改变相对地址和片段目标，排错时同时检查页面与服务目录。
