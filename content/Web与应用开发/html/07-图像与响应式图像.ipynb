# 图像与响应式图像

学习目标：能编写图像替代文本，根据显示空间提供图像候选和不同构图，并区分显示尺寸、资源体积与浏览器实际选择。

前置知识：HTML 元素与属性、相对 URL、基本文件操作和浏览器开发者工具。

适用范围：WHATWG HTML Living Standard；使用现代浏览器查看配套页面。本章 SVG 素材已提供，不需要生成文件或下载图片。

## 1 标签总览

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;img&gt; | 图像元素 | 嵌入图片并提供替代文本 |
| &lt;figure&gt; | 独立内容单元 | 组织插图及其图注，也可用于其他完整内容 |
| &lt;figcaption&gt; | 图注 | 说明所属内容单元 |
| &lt;picture&gt; | 图像候选容器 | 按媒体条件或格式提供图像来源 |
| &lt;source&gt; | 候选来源 | 在 &lt;picture&gt; 中声明一组候选 |

## 2 &lt;img&gt; 与 alt：嵌入有意义的图片

&lt;img&gt; 是空元素，不写结束标签。src 属性指定资源 URL；alt 属性提供图像的替代文本，说明图片在当前上下文中传达的信息。

- 信息图：简述必要内容，不只写文件名或“图片”。复杂图表还需要正文中的完整说明。
- 功能图：图片是链接或按钮的唯一内容时，alt 应说明目的地或动作。
- 装饰图：明确写 alt=""，引号中没有字符。省略 alt 不等于空 alt，可能让屏幕阅读器读出文件名。

下面的插图描述湖边小屋；width、height 在第 4 节说明。

```html
<img id="informative-image" class="scene" src="lake-480.svg"
  width="480" height="270"
  alt="红色小屋位于蓝色湖泊左岸，远处是连绵的山峰。">
```

## 3 &lt;figure&gt; 与 &lt;figcaption&gt;：添加图注

&lt;figure&gt; 把插图及相关内容组织成一个完整单元，不要求每张图都使用它。一个 &lt;figure&gt; 至多有一个 &lt;figcaption&gt;，放在第一个或最后一个子元素位置。

图注是可见说明，alt 是替代文本；有图注不自动免除 alt。本例图注给出图号与主题，alt 补充画面内容。

```html
<figure>
  <img id="informative-image" class="scene" src="lake-480.svg"
    width="480" height="270"
    alt="红色小屋位于蓝色湖泊左岸，远处是连绵的山峰。">
  <figcaption>图 1：湖边小屋，原创场景示意。</figcaption>
</figure>
<!-- 预期：图注可见，alt 不作为图片上的字幕显示。 -->
<!-- 检查：Elements 的 Accessibility 中检查图像名称与链接名称。 -->
```

## 4 width、height 与 CSS：区分资源尺寸和显示尺寸

width、height 属性用不带单位的非负整数给出尺寸；本例按图像比例填写 480 和 270。浏览器可以据此在下载完成前预留空间，减少加载后的内容移动。

CSS 可以改变显示大小。配套样式中的 max-width: 100% 防止图片超出容器，height: auto 让高度随比例计算；它们不会修改磁盘文件或减少下载字节数。

```css
.scene {
  display: block;
  max-width: 100%;
  height: auto;
}
```

配套文件：[scripts/07-images-and-responsive-images/styles.css](scripts/07-images-and-responsive-images/styles.css) · [浏览器预览](http://127.0.0.1:8007/scripts/07-images-and-responsive-images/index.html)

CSS 像素（CSS px）不一定等于一个物理屏幕像素。缩窄示例首页，检查图像是否仍保持 16:9；不要把宽高强制改成不相配的比例。

## 5 loading：选择加载时机

首屏主要图片使用正常加载；较远的下方图片可采用延迟加载，并保留宽高。

- loading="eager"：默认值，不因图片位于视口外而推迟加载。
- loading="lazy"：接近视口时再加载，提前距离由浏览器决定。浏览器需启用 JavaScript，但页面无需另写延迟加载脚本。

配套页面用留白把近景放在下方；下面是下方图像的完整开始标签。

```html
<img id="lazy-image" class="scene" src="cabin-480.svg"
  width="480" height="480" loading="lazy"
  decoding="async" fetchpriority="low"
  alt="湖边红色小屋的近景，浅黄色门位于正面。">
<!-- 检查：Network 开启 Disable cache 后，在页面顶部刷新并向下滚动。 -->
<!-- 观察 cabin-480.svg 的请求时机；浏览器可能在图片进入视口前就请求。 -->
<!-- 预期滚动到下方后近景可以加载；不要求某个固定距离才发起请求。 -->
```

配套文件：[scripts/07-images-and-responsive-images/loading.html](scripts/07-images-and-responsive-images/loading.html) · [浏览器预览](http://127.0.0.1:8007/scripts/07-images-and-responsive-images/loading.html)

补充：获取资源和解码成画面是不同阶段。

- fetchpriority：获取优先级提示；high、low 表示较高、较低，auto 交给浏览器决定。
- decoding：解码提示；async 倾向不推迟其他内容呈现，sync 倾向一起呈现，auto 不指定偏好。

这些提示不保证请求完成顺序，也不减少资源字节数。首屏图像使用 eager 与 high，完整写法见同一文件。

## 6 srcset 的 x 描述符：固定显示宽度

分辨率切换（resolution switching）保持构图一致，提供不同尺寸的资源。设备像素比（device pixel ratio，DPR）表示物理像素与 CSS 像素的比例，页面缩放也可能改变它。

本例用 CSS 将图像显示宽度设为 240 CSS 像素。

- srcset：用逗号分隔候选，每项先写 URL，再写描述符。
- 1x、2x：x 是像素密度描述符的后缀；2x 表示相对于预期显示尺寸的二倍密度。
- 未写描述符的候选按 1x 处理；同一列表不能重复声明相同密度。

```html
<img id="sample-image" class="scene density-image"
  src="lake-240.svg"
  srcset="lake-240.svg 1x, lake-480.svg 2x"
  width="240" height="135"
  alt="红色小屋位于蓝色湖泊左岸，远处是连绵的山峰。">
<!-- 检查：在 390px 和 1000px 视口中，图像显示宽度都应为 240px。 -->
<!-- 在设备工具栏改变 DPR 并重新加载，记录 currentSrc 与 Network 请求。 -->
<!-- 不预先指定浏览器必须下载哪一个候选。 -->
```

配套文件：[scripts/07-images-and-responsive-images/density.html](scripts/07-images-and-responsive-images/density.html) · [浏览器预览](http://127.0.0.1:8007/scripts/07-images-and-responsive-images/density.html)

浏览器结合 DPR、缩放及可能的网络条件选择资源，DPR 为 2 不保证必选某个文件。这里使用可缩放的 SVG，仅观察候选语法；实际矢量图通常无需为清晰度准备这些重复版本，不能用本例比较栅格图的清晰度。

## 7 srcset 的 w 描述符与 sizes：变化的显示空间

显示宽度随布局变化时，w 描述符标注资源的自然宽度，sizes 提供选择候选时参照的显示宽度。

- 480w：w 是宽度描述符后缀；480 要与资源的自然宽度相符，不能填希望显示的宽度。本例两份 SVG 明确声明宽度 480、960。
- 同一 srcset 不能混用 w 与 x；使用 w 时每项都要写宽度，且不能重复。
- sizes：按顺序使用第一个匹配条件；本例最后的 480px 是无条件默认项。长度可以用 px、vw、calc()，不能用百分比。

1vw 是视口宽度的 1%；calc(100vw - 32px) 是视口宽度减去左右各 16 CSS 像素的留白。

```html
<img id="sample-image" class="scene responsive-image"
  src="lake-480.svg"
  srcset="lake-480.svg 480w, lake-960.svg 960w"
  sizes="(max-width: 640px) calc(100vw - 32px), 480px"
  width="960" height="540"
  alt="红色小屋位于蓝色湖泊左岸，远处是连绵的山峰。">
<!-- 检查：390px 视口中显示宽度约 358px，1000px 视口中为 480px。 -->
<!-- 在 Elements 临时把 sizes 改成 100vw，再点按钮；CSS 宽度应保持原值。 -->
<!-- 候选 URL 可能改变，也可能不变；刷新可恢复文件中的 sizes。 -->
```

配套文件：[scripts/07-images-and-responsive-images/responsive.html](scripts/07-images-and-responsive-images/responsive.html) · [浏览器预览](http://127.0.0.1:8007/scripts/07-images-and-responsive-images/responsive.html)

sizes 不设置 CSS 宽度。本例 CSS 在视口不超过 640px 时使用相同的计算式，其他视口显示为 480px；修改布局时应同时核对 sizes。这里是默认 eager 图像，使用 w 候选须写 sizes；漏写会按 100vw 参与选择，可能偏离布局。

浏览器将候选宽度与 sizes 换算成有效密度后自行选择。src 保留兼容入口，不代表每次先下载它，也不保证缩窄窗口后会重新下载小图。

## 8 &lt;picture&gt;：按布局切换构图

art direction（按布局切换图片构图）：宽屏保留湖景，窄屏突出小屋。它改变取景，与上一节保持同一画面的分辨率切换不同。

&lt;picture&gt; 中先放 &lt;source&gt;，最后保留一个 &lt;img&gt;。真正显示图片的仍是 &lt;img&gt;，alt 也写在它上面。

- media：候选适用的媒体条件；本例为视口宽度不超过 640 CSS 像素。
- type：图像类型提示；image/svg+xml 表示 SVG。
- srcset：该条件下的候选。在 &lt;picture&gt; 中使用它，而非音视频场景中的 src。

```html
<picture>
  <source media="(max-width: 640px)" type="image/svg+xml"
    srcset="cabin-480.svg" width="480" height="480">
  <img id="sample-image" class="scene art-image"
    src="lake-960.svg" width="960" height="540"
    alt="一座红色小屋坐落在蓝色湖泊岸边。">
</picture>
<!-- 检查：在 390px 视口重新加载，应显示 cabin-480.svg 的方形近景。 -->
<!-- 在 1000px 视口重新加载，应显示 lake-960.svg 的 16:9 全景。 -->
```

配套文件：[scripts/07-images-and-responsive-images/art-direction.html](scripts/07-images-and-responsive-images/art-direction.html) · [浏览器预览](http://127.0.0.1:8007/scripts/07-images-and-responsive-images/art-direction.html)

浏览器按顺序考虑媒体条件匹配且格式受支持的 &lt;source&gt;，再从其中选择候选；没有合适的来源时使用 &lt;img&gt; 自己的资源。这里每个分支只有一个文件。

近景为 1:1，宽幅为 16:9。&lt;source&gt; 与 &lt;img&gt; 分别提供正确宽高，配合 height: auto 避免拉伸。所选 &lt;source&gt; 的宽高可供 &lt;img&gt; 预留比例；旧浏览器还需检查加载前的布局。共同 alt 只描述两种构图都有的小屋与湖岸。

## 9 图像格式与资源体积

矢量图描述形状和线条，栅格图保存像素网格。选择格式时先看内容、画质与目标浏览器，再比较实际体积。

- SVG：可缩放矢量图形，适合图标和示意图。
- PNG：无损栅格格式，支持透明，适合需要保留细节的图形。
- JPEG：常用于照片的有损压缩，不支持透明背景。
- WebP：支持有损、无损、透明及动画。
- AVIF：AV1 图像文件格式，支持透明及动画等能力；编码选项与浏览器支持需对应实际需求。

&lt;picture&gt; 也可利用 type 提供不同格式的候选。不支持声明的格式时可跳过该来源，但 type 不转换文件，已选 URL 返回 404 也不等于自动尝试下一种格式。本章只提供真实存在的 SVG。

打开[宽幅素材](scripts/07-images-and-responsive-images/lake-960.svg)或[浏览器中的素材](http://127.0.0.1:8007/scripts/07-images-and-responsive-images/lake-960.svg)，对照文件属性与 Network 的 Size。三份宽幅 SVG 的绘图内容相同，只改变声明尺寸；尺寸增大不意味着字节数按比例增大。CSS 缩小显示也不会生成小文件，不能用这些素材给 JPEG、WebP、AVIF 排压缩率名次。

## 10 读取浏览器实际选择

在密度、宽度和构图页面点击“读取当前图像”，配套脚本只显示信息，不控制选图。

- currentSrc：当前选中资源的完整 URL；非空不代表已经成功加载。
- getBoundingClientRect()：元素边界矩形；本例图片没有边框、内边距或变换，可用于观察显示宽高。
- devicePixelRatio：当前 DPR。

这些是浏览器 API，不是 HTML 属性。完整实现见 [scripts/07-images-and-responsive-images/inspect.js](scripts/07-images-and-responsive-images/inspect.js)；[打开响应式示例](http://127.0.0.1:8007/scripts/07-images-and-responsive-images/responsive.html)。

在开发者工具 Network 中开启 Disable cache 并保持工具打开，再调整视口、刷新和读取信息。结合 currentSrc、实际画面和网络请求核对结果；传输量受缓存影响，不等于磁盘文件大小。设备工具栏的模拟不能替代真实设备检查。

## 本章小结

- alt 按图片用途编写，装饰图明确空值；图注与替代文本各有用途。
- 宽高和 CSS 影响布局，加载与解码提示影响处理偏好；它们不会改写资源文件。
- x 描述密度，w 描述资源宽度；sizes 提供显示空间信息，须与实际布局对应。
- &lt;picture&gt; 可切换构图或格式，必须保留 &lt;img&gt;。候选声明与实际选中、成功加载分别判断。
