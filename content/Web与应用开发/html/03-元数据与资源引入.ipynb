# 元数据与资源引入

学习目标：能设置页面元数据，引入图标、样式和脚本，并按加载规则判断脚本执行时机。

前置知识：HTML 元素与属性、head 与 body、文件保存、浏览器开发者工具。

适用范围：WHATWG HTML Living Standard；使用支持 JavaScript 模块的现代浏览器。模块示例通过 HTTP 访问；资源均在本章目录内。

环境准备：[环境配置与运行](README.md)。

工作目录：content/Web与应用开发/html；在浏览器运行配套页面。

配套脚本：位于 scripts/03-metadata-and-resources/。

1. [metadata.html](scripts/03-metadata-and-resources/metadata.html)：标题、编码、描述、视口与样式。
2. [favicon.svg](scripts/03-metadata-and-resources/favicon.svg)、[styles.css](scripts/03-metadata-and-resources/styles.css)：本地图标与共用样式。
3. [loading.html](scripts/03-metadata-and-resources/loading.html)、[classic.js](scripts/03-metadata-and-resources/classic.js)：加载实验与简短执行日志。
4. [defer-first.js](scripts/03-metadata-and-resources/defer-first.js)、[defer-second.js](scripts/03-metadata-and-resources/defer-second.js)：延迟经典脚本的执行顺序。
5. [async-first.js](scripts/03-metadata-and-resources/async-first.js)、[async-second.js](scripts/03-metadata-and-resources/async-second.js)：独立异步脚本。
6. [module-main.js](scripts/03-metadata-and-resources/module-main.js)、[module-message.js](scripts/03-metadata-and-resources/module-message.js)：模块入口与依赖。

## 1 标签总览

| 标签 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| &lt;title&gt; | 文档标题 | 标识整个页面 |
| &lt;meta&gt; | 元数据 | 描述编码、摘要和视口等信息 |
| &lt;link&gt; | 外部资源关系 | 引入图标、样式表等资源 |
| &lt;style&gt; | 内嵌样式表 | 在文档中保存 CSS |
| &lt;script&gt; | 脚本元素 | 本章用于经典脚本与模块脚本 |
| &lt;noscript&gt; | 无脚本替代内容 | 脚本禁用时提供 HTML 内容 |

本章将元数据和资源引用放在 &lt;head&gt;，替代正文放在 &lt;body&gt;。&lt;meta&gt;、&lt;link&gt; 是空元素；&lt;script&gt; 即使使用 src 也要写结束标签。

## 2 文档标题、编码与描述

&lt;head&gt; 保存描述页面的信息。本例独立文档使用一个 &lt;title&gt;，其文字通常出现在浏览器标签页；正文的 &lt;h1&gt; 则是页面内的主标题。

- charset="utf-8"：声明 UTF-8 编码，文件本身也必须按 UTF-8 保存。
- name="description"：声明页面描述，content 保存摘要文字；一个文档最多一条这种描述。
- charset 与 name/content 是不同的 &lt;meta&gt; 写法，不混在同一个元素中。

下面三行均来自 metadata.html 的头部，viewport 在下一节单独说明。

```html
<meta charset="utf-8">
```

配套文件：[scripts/03-metadata-and-resources/metadata.html](scripts/03-metadata-and-resources/metadata.html) · [浏览器预览](http://127.0.0.1:8003/scripts/03-metadata-and-resources/metadata.html)

```html
<meta name="description" content="用一个学习页面观察文档标题、编码、视口和资源引入。">
<title>元数据实验 | 动手学计算机</title>
```

配套文件：[scripts/03-metadata-and-resources/metadata.html](scripts/03-metadata-and-resources/metadata.html) · [浏览器预览](http://127.0.0.1:8003/scripts/03-metadata-and-resources/metadata.html)

&lt;title&gt; 元素和 title 属性不同：前者命名整个文档，后者给某个元素提供补充提示。页面描述也不会自动成为正文段落。

编码声明必须完整位于文件前 1024 字节，不能把“字节”当作汉字个数。声明不会转换已有文件的编码；出现乱码时同时检查保存编码、声明和 HTTP 的 Content-Type。浏览器还考虑 BOM（字节顺序标记）等信息，&lt;meta&gt; 不能保证覆盖错误的服务器编码声明。

## 3 viewport：设置移动端视口

视口（viewport）是浏览器显示页面的区域。移动浏览器常用以下设置，让布局以设备宽度为基础。

- width=device-width：让布局视口宽度对应设备宽度，以 CSS 像素计。
- initial-scale=1：初始缩放比例为 1。

这组设置为窄屏布局提供基础，实际适配仍取决于 CSS；不要用 user-scalable=no 限制用户缩放。

```html
<meta name="viewport" content="width=device-width, initial-scale=1">
```

在开发者工具的设备模式中选择手机预设并刷新，观察文字换行和是否横向溢出。设备模拟不能代替真实手机检查。

## 4 &lt;link&gt;：引入网站图标

&lt;link&gt; 描述文档与资源的关系；它不同于正文中供点击的 &lt;a&gt;，也不同于显示正文图片的 &lt;img&gt;。

- rel="icon"：资源用作网站图标。
- href="favicon.svg"：从当前 HTML 所在目录读取图标。
- type="image/svg+xml"：提示资源为 SVG 图像。
- sizes="any"：提示图标可缩放。

配套 SVG 是蓝底白圆的矢量图标，此处只使用素材，不展开绘图语法。图标通常显示在标签页，位置由浏览器决定。

```html
<link rel="icon" href="favicon.svg" type="image/svg+xml" sizes="any">
```

配套文件：[scripts/03-metadata-and-resources/metadata.html](scripts/03-metadata-and-resources/metadata.html) · [浏览器预览](http://127.0.0.1:8003/scripts/03-metadata-and-resources/metadata.html)

配套素材：[scripts/03-metadata-and-resources/favicon.svg](scripts/03-metadata-and-resources/favicon.svg)。在 Network 中核对请求路径；不要把 Notebook 链接中的 scripts/03-metadata-and-resources/ 再抄进这里的 href，造成目录重复。

## 5 引入 CSS：外部文件与 &lt;style&gt;

共用样式放在外部 CSS 文件，用 rel="stylesheet" 引入；只属于本页的样式也可以写在 &lt;head&gt; 中的 &lt;style&gt;。

外部文件只写 CSS，不包裹 &lt;style&gt; 标签。&lt;style&gt; 元素容纳样式表，单个元素上的 style 属性则直接保存该元素的样式声明。

```html
<link rel="stylesheet" href="styles.css">
<style>
  /* 相同条件下后出现的颜色声明生效；不会抹掉外部样式的边框。 */
  .note { color: #9f1239; }
</style>
```

配套文件：[scripts/03-metadata-and-resources/metadata.html](scripts/03-metadata-and-resources/metadata.html) · [浏览器预览](http://127.0.0.1:8003/scripts/03-metadata-and-resources/metadata.html)

.note 选择 class 中含 note 的元素，color 设置文字颜色。本例外部样式也定义了同一选择器的颜色，并添加蓝色左边框。

本例两条颜色声明的来源、重要性和选择器权重相同，也没有层叠层或作用域差异，因此文档中后出现的深红色生效，边框仍来自外部文件。不要概括成“页面内样式永远优先”；先后位置也不是下载完成顺序。

```html
<p class="note">本段文字来自页面内样式的深红色，左边框来自外部 CSS。</p>
```

配套文件：[scripts/03-metadata-and-resources/metadata.html](scripts/03-metadata-and-resources/metadata.html) · [浏览器预览](http://127.0.0.1:8003/scripts/03-metadata-and-resources/metadata.html)

完整样式：[scripts/03-metadata-and-resources/styles.css](scripts/03-metadata-and-resources/styles.css)。在 Elements 的样式面板中比较颜色和边框的来源。

## 6 普通经典脚本

&lt;script&gt; 可以在标签内直接写 JavaScript，也可以用 src 引入文件。省略 type 时，本章 JavaScript 按经典脚本（classic script）处理；使用 src 后，标签内部不是备用脚本位置。

本章讨论直接写在 HTML 中、随解析遇到的脚本。没有 async 或 defer 的外部经典脚本会使解析等待其获取和执行完成，再处理后面的 HTML。

```html
<!-- 普通经典脚本在这里执行时，正文末尾的 end 尚不存在。 -->
<script src="classic.js"></script>
```

配套文件：[scripts/03-metadata-and-resources/loading.html](scripts/03-metadata-and-resources/loading.html) · [浏览器预览](http://127.0.0.1:8003/scripts/03-metadata-and-resources/loading.html)

classic.js 建立本例自定义的 record 记录函数。后续脚本调用它，在页面显示执行名称、末尾元素是否存在及解析状态；这些日志仅用于对照时机，完整实现留在配套文件。

- 末尾元素：id 为 end 的段落，true 表示调用时已能找到，false 表示尚未找到。
- readyState：浏览器的文档加载状态；loading 表示仍在解析，interactive 表示解析完成，complete 表示文档及相关加载工作完成。

DOM 是文档对象模型。配套脚本中的 document.getElementById、textContent 和 addEventListener 分别用于找元素、更新文字和监听事件，都是浏览器 API，不是 HTML 标签。

## 7 defer：解析后按声明顺序执行

外部经典脚本添加 defer 后，可在解析 HTML 的同时获取文件；等解析结束，再按文档中的声明顺序执行，并在 DOMContentLoaded 事件之前完成。适合需要读取正文或有先后依赖的脚本。

```html
<!-- 两者都在解析后执行；defer-first 先于 defer-second。 -->
<script defer src="defer-first.js"></script>
<script defer src="defer-second.js"></script>
```

配套文件：[scripts/03-metadata-and-resources/loading.html](scripts/03-metadata-and-resources/loading.html) · [浏览器预览](http://127.0.0.1:8003/scripts/03-metadata-and-resources/loading.html)

defer-first.js 与 defer-second.js 各调用一次 record，文件名本身不决定加载方式，HTML 中的 defer 属性才起作用。第二个文件先下载完也不能因此先执行。

defer 对没有 src 的内联经典脚本无效，也不应这样使用；模块脚本的默认行为见第 9 节。

## 8 async：就绪后尽快执行

比较加载方式时，分别问“文件何时获取”和“代码何时执行”。async 允许获取与 HTML 解析重叠，文件就绪后尽快执行；执行仍可能中断解析，不表示脚本自动在另一条线程中运行。

多个 async 脚本不保证声明顺序，适合互不依赖的任务。本例两者只使用已由 classic.js 建立的 record。

![WHATWG 脚本加载时间线：普通、defer、async 经典脚本以及普通与 async 模块脚本，分别标出解析、获取和执行。](image/illustration/03-01-script-loading-timeline.svg)

图 1 引用 WHATWG HTML 的脚本加载图，基于 Peter Beverloo 的原图，按 CC0 1.0 提供，未改绘。绿色 parser 表示 HTML 解析，蓝色 fetch 表示获取脚本，红色 execution 表示执行，时间向右推进。前三行对应本节比较的外部经典脚本；后两行是模块脚本，蓝色分支表示依赖获取，结合下文模块说明阅读。

async 一行只画“解析未结束就已就绪”的一种情况，不表示每次刷新都如此。线段长度是概念示意，不是本课程的实测耗时。

接着把下面两个 async 文件的日志放回时间线上：能判断执行时机可能变化，但不能从文件先后写法推出执行顺序。

```html
<!-- 两个 async 脚本互不依赖，不预设谁先执行。 -->
<script async src="async-first.js"></script>
<script async src="async-second.js"></script>
```

配套文件：[scripts/03-metadata-and-resources/loading.html](scripts/03-metadata-and-resources/loading.html) · [浏览器预览](http://127.0.0.1:8003/scripts/03-metadata-and-resources/loading.html)

- async、defer 都是布尔属性；async="false" 仍表示存在，要关闭就删除属性。
- 外部经典脚本同时写 async 与 defer 时，现代浏览器按 async 处理。
- async 对内联经典脚本无效。

缓存、响应时机和调度会影响本次日志位置。多次刷新恰好顺序相同，也不能证明 async 保证顺序；不要求刷新后一定交换位置。

## 9 模块脚本与依赖

type="module" 将脚本声明为 JavaScript 模块（module script）。未加 async 时，模块及依赖可与解析并行获取，在解析后执行；内联模块也有默认延迟行为，不需要 defer。

模块加 async 后，会在入口和依赖准备好时尽快执行，不再默认等待解析结束。这里保留默认模块示例。

```html
<!-- 本模块未加 async，也没有顶层 await；执行时 end 已存在。 -->
<script type="module" src="module-main.js"></script>
```

配套文件：[scripts/03-metadata-and-resources/loading.html](scripts/03-metadata-and-resources/loading.html) · [浏览器预览](http://127.0.0.1:8003/scripts/03-metadata-and-resources/loading.html)

模块用 import/export 明确导入、导出。下面摘自入口模块：moduleLabel 是从同目录 ./module-message.js 导入的文字，moduleOnly 是本模块自己的变量。const 声明名称，record 是本例已有的记录函数。

```javascript
import { moduleLabel } from "./module-message.js";

const moduleOnly = "模块内变量可读";
// 本页未使用 async 或顶层 await，日志为：
// module-main：已读取导入的文字 | 末尾元素=true | readyState=interactive
record("module-main：" + moduleLabel);
```

配套文件：[scripts/03-metadata-and-resources/module-main.js](scripts/03-metadata-and-resources/module-main.js) · [浏览器预览](http://127.0.0.1:8003/scripts/03-metadata-and-resources/loading.html)

模块自动使用严格模式，声明或导入的名称不会自动成为其他脚本的全局名称。配套页面同时显示模块内文字和经典脚本中的 typeof moduleOnly；后者为 undefined 表示这里没有可读取的同名定义。模块仍能使用 document 等浏览器 API。

模块身份由 type="module" 和导入关系决定，不要求 .mjs 扩展名。通过 HTTP 打开本例；服务器须发送 JavaScript MIME 类型，例如 text/javascript。跨源模块还需要满足 CORS（跨源资源共享）要求，本例资源均为同源。

## 10 对照加载日志与内联反例

普通内联经典脚本在解析遇到时执行。loading.html 在 &lt;head&gt; 中故意保留无效的 defer 用法，以对比外部 defer 脚本：

```html
<!-- 故意反例：无 src 的经典脚本不应加 defer，添加也不会产生延迟效果。 -->
<script defer>
  // 预期仍是 false 和 loading；不要把这个属性照搬到正式页面。
  record("inline-classic-with-defer");
</script>
```

配套文件：[scripts/03-metadata-and-resources/loading.html](scripts/03-metadata-and-resources/loading.html) · [浏览器预览](http://127.0.0.1:8003/scripts/03-metadata-and-resources/loading.html)

正文末尾的内联脚本则位于 end 段落之后，所以已能读取它：

```html
<p id="end">正文结束标记</p>
<script>
  // 预期 end 已存在，但此时解析器尚未处理完文档，readyState 仍为 loading。
  record("body-end");
</script>
```

配套文件：[scripts/03-metadata-and-resources/loading.html](scripts/03-metadata-and-resources/loading.html) · [浏览器预览](http://127.0.0.1:8003/scripts/03-metadata-and-resources/loading.html)

- DOMContentLoaded：本例在 HTML 解析完成、两个 defer 脚本与默认模块执行后触发，不直接等待 async 脚本或图片。样式表可能通过阻塞脚本间接推迟它。
- load：用于观察本页及相关资源的加载完成，不是“刚能读取正文”的最早时刻。

本例没有动态插入脚本，也没有模块顶层 await（模块中的异步等待）。不能把这里的事件关系推广到异步模块的后续工作。

在 Network 中看资源获取，在页面日志中看执行；请求成功或较早开始，都不能直接证明脚本已经执行。

## 11 &lt;noscript&gt;：脚本禁用时的内容

&lt;noscript&gt; 为禁用或不支持脚本的环境提供替代 HTML。本例把说明与普通链接放在 &lt;body&gt; 中，读者无需脚本也能进入学习页。

```html
<noscript>
  <p>当前未启用脚本，加载日志不可用；正文和链接仍可阅读。</p>
  <p><a href="metadata.html">阅读不依赖脚本的学习页</a></p>
</noscript>
```

配套文件：[scripts/03-metadata-and-resources/loading.html](scripts/03-metadata-and-resources/loading.html) · [浏览器预览](http://127.0.0.1:8003/scripts/03-metadata-and-resources/loading.html)

脚本请求失败、JavaScript 报错都不等于关闭脚本支持，&lt;noscript&gt; 不会自动捕获这些故障。

在独立实验页面的 Chrome 开发者工具中按 Ctrl+Shift+P，运行 Disable JavaScript 并刷新；观察后运行 Enable JavaScript，再刷新恢复。

## 本章小结

- 文档标题、描述、编码和视口分别解决命名、摘要、文本解码与显示基础设置问题。
- &lt;link&gt; 的 rel 决定资源关系；CSS 冲突要按层叠条件判断。
- 普通经典脚本阻塞解析；defer 保留声明顺序；async 不承诺顺序；模块默认延迟并有独立作用域。
- 区分文件请求、脚本执行与页面事件；&lt;noscript&gt; 对应脚本禁用环境。

### 参考解析

1. 文档标题出现在标签页，description 保存在元数据中，h1 属于可见正文；三者不会自动互相同步。
2. 只交换两份相同权重规则的位置后，外部样式的蓝色胜出；左边框始终来自外部文件。这个对照固定了其他层叠条件，不能推成“外部样式永远优先”。
3. classic-head 与内联 defer 反例仍处于 loading 且找不到 end；body-end 找到 end 时仍是 loading。两个 defer 按声明顺序、默认模块也在 DOMContentLoaded 之前执行；async 的位置只代表本次观察。
4. 缺失的 async 文件产生加载错误，但脚本支持仍开启，noscript 不因此显示。禁用脚本后日志不再由脚本更新，noscript 中的说明与普通链接可见。
