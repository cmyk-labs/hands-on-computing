## 1 先让布局流动，再添加断点

响应式设计先让容器、轨道和内容适应空间，再在布局确实需要变化的位置添加断点（breakpoint）。断点是内容需要的尺寸，不等同于某个手机型号；本例的 640px 是教学选择。

移动优先（mobile first）先给窄视口可读的单列基础，再逐步增强为多列。默认规则并非只属于手机；宽视口仍会应用它们，条件规则只覆盖需要变化的声明。

width: 100% 配合 max-width 让盒子填满可用宽度并有上限。Grid 的 minmax(0, 1fr) 和长内容换行一起处理窄列；不能仅依赖断点把溢出藏起来。

```html
<div class="fluid-layout">
  <article><h3>基础</h3><p>先保证正文在窄窗口里能够顺序阅读。</p></article>
  <article><h3>实践</h3><p>保存、刷新，再逐渐改变窗口宽度。</p></article>
</div>
```

```css
.fluid-layout {
  display: grid;
  grid-template-columns: minmax(0, 1fr);
  gap: 16px;
  width: 100%;
  max-width: 960px;
  margin: 0 auto;
}
.fluid-layout > article {
  min-width: 0;
  padding: 12px;
  border: 1px solid teal;
  overflow-wrap: anywhere;
}
@media (min-width: 640px) {
  .fluid-layout { grid-template-columns: repeat(2, minmax(0, 1fr)); }
  /* 分别在 639px 与 640px 检查卡片从上下排列变成并排。 */
}
```

## 2 视口与页面缩放

媒体特性 width 查询视口宽度，使用 CSS 像素；它不是某个卡片的宽度，也不是显示面板的物理像素总数。窄窗口的桌面浏览器同样可以匹配窄屏查询。

本章各页的 &lt;head&gt; 都包含下方 viewport 声明。width=device-width 使移动布局视口按设备的 CSS 像素宽度配置；initial-scale=1 指定初始缩放。不要加 user-scalable=no 或过小的 maximum-scale 来禁止放大。

桌面页面缩放会影响以 CSS 像素计的可用视口；移动端双指缩放主要改变视觉视口（visual viewport），不保证触发相同的布局变化。开发者工具里的改宽、浏览器缩放和真实手机手势需要分别检查。

```html
<meta name="viewport" content="width=device-width, initial-scale=1">
```

```html
<p class="viewport-sample">保留缩放能力，正文仍可换行。</p>
```

```css
.viewport-sample {
  max-width: 40rem;
  padding: 1rem;
  border: 1px solid teal;
  /* 放大页面至 200%：检查文字可读、内容未被固定高度裁掉。 */
}
```

## 3 @media、逻辑组合与范围语法

@media 是条件 @ 规则；screen、print 是媒体类型，省略类型相当于 all。width、orientation 等位于条件中时是媒体特性，不是给元素赋值的 CSS 属性。

- and 要求条件同时成立；逗号连接独立查询，只要一项成立即可。not screen and (min-width: 640px) 否定的是整个查询，不能读成“不是屏幕且足够宽”。
- 条件中的 or 也表示或。and 与 or 混用要用括号明确分组，不要在同一层随意混排。only 主要为旧式浏览器提供兼容保护，不会提高现代浏览器中的优先级。
- min-width: 640px 包含 640px，等价于 width >= 640px；max-width 同样包含边界。范围式 640px <= width < 960px 则只包含下界。相邻区间可用严格比较避免空隙或重复。

符合多个查询的规则仍按普通层叠决定结果，没有“更窄区间自动更优先”。本例保留 min/max 写法作为范围语法未实现时的可读回退；边框颜色是额外增强。

```html
<p class="range-sample">区间查询标记</p>
<p class="logic-sample">屏幕宽视口或打印介质标记</p>
<p class="not-sample">整个查询的否定标记</p>
<p class="or-sample">窄于 640px 或达到 960px 的标记</p>
```

```css
.range-sample, .logic-sample, .not-sample, .or-sample {
  border: 3px solid gray;
  padding: 8px;
}
@media (min-width: 640px) and (max-width: 960px) {
  .range-sample { background-color: #e1f1ed; }
}
@media (640px <= width < 960px) {
  .range-sample { border-color: teal; }
  /* 960px 时回退背景仍在，严格范围的蓝绿色边框退出。 */
}
@media screen and (min-width: 640px), print {
  .logic-sample { border-style: dashed; }
}
@media not screen and (min-width: 640px) {
  .not-sample { font-weight: bold; }
  /* 屏幕 639px 时匹配；屏幕 640px 时不匹配；print 时匹配。 */
}
@media (width < 640px) or (width >= 960px) {
  .or-sample { border-style: dotted; }
}
```

## 4 方向、分辨率与尺寸单位

orientation 比较视口宽高：宽大于高为 landscape，宽小于或等于高为 portrait。它不是设备姿态传感器；软键盘等因素改变可用空间，也可能改变查询结果。

resolution 查询输出像素密度。dppx 表示每个 CSS px 对应的设备像素密度，1dppx 等于 96dpi；查询它不意味着图片一定下载了更多细节，也不能用来推断输入方式。

媒体查询里的 em、rem 长度以初始字体值为参照，避免作者样式改变条件后形成循环；不是简单跟随正文中重新设置的字号。本章用 px 让断点边界容易复核，实际设计也可按阅读需求选择相对单位。

```html
<p class="orientation-sample">视口方向标记</p>
<p class="resolution-sample">输出密度标记</p>
```

```css
.orientation-sample, .resolution-sample { border: 2px solid gray; padding: 8px; }
@media (orientation: landscape) {
  .orientation-sample { border-style: dashed; }
}
@media (min-resolution: 2dppx) {
  .resolution-sample { border-width: 6px; }
  /* 结合设备像素比仿真观察边框；本例不据此声称图像更清晰。 */
}
```

## 5 hover、pointer 与键盘入口

hover 判断主要输入机制是否能方便悬停；pointer 的 none、coarse、fine 分别表示没有主要指针、精度有限和精度较高。它们描述能力，不能直接等同于“手机”或“桌面”。

any-hover、any-pointer 汇总任一可用输入机制；混合设备可能同时匹配多种 any-pointer 精度。拥有鼠标不代表用户此刻一定在使用鼠标，查询也不能替代键盘焦点样式。

下面的链接始终存在。只有主要指针既精确又支持悬停时才添加悬停增强；粗略指针增加内边距。:hover 和 :focus-visible 是选择器中的状态伪类，与同名媒体特性 hover 分属不同语法。

```html
<a class="input-link" href="#input-target">打开本页目标</a>
<p id="input-target" tabindex="-1">目标内容始终可用。</p>
```

```css
.input-link {
  display: inline-block;
  padding: 12px 16px;
  border: 2px solid currentColor;
  color: #174f79;
}
.input-link:focus-visible { outline: 3px solid #174f79; outline-offset: 4px; }
@media (hover: hover) and (pointer: fine) {
  .input-link:hover { background-color: #e1f1ed; }
}
@media (pointer: coarse) {
  .input-link { padding: 18px 24px; }
}
/* Tab 聚焦并按 Enter：即使不满足 hover 条件，仍能到达链接目标。 */
```

## 6 用户偏好也是响应条件

prefers-color-scheme 反映用户偏好的明暗方案；prefers-reduced-motion: reduce 表示希望减少非必要运动。查询只给出条件，具体颜色、滚动或动画策略仍需作者编写。

本例默认立即滚动，只有 no-preference 时才启用平滑锚点滚动；不支持该查询时仍保留即时滚动。明暗主题的复用与原生控件方案在“自定义属性与主题”中展开。

prefers-contrast: more 可用于加强边界。forced-colors: active 表示浏览器采用受限的用户颜色方案，和偏好暗色不是同一条件；保留原生颜色调整，本例用系统颜色 CanvasText 补足边框，不取消用户选择。

```html
<div class="preference-panel">
  <p>按偏好调整的内容区域。</p>
  <a href="#reading-end">跳到阅读结束处</a>
</div>
<div class="reading-space">用于观察锚点滚动的阅读空间。</div>
<p id="reading-end">阅读结束。</p>
```

```css
.preference-panel {
  background-color: #ffffff;
  color: #17202a;
  border: 2px solid currentColor;
  padding: 16px;
}
.preference-panel a { color: inherit; }
.reading-space { min-height: 80vh; }
html { scroll-behavior: auto; }
@media (prefers-reduced-motion: no-preference) {
  html { scroll-behavior: smooth; }
}
@media (prefers-color-scheme: dark) {
  .preference-panel { background-color: #17202a; color: #f4f7fa; }
}
@media (prefers-contrast: more) {
  .preference-panel { border-width: 4px; }
}
@media (forced-colors: active) {
  .preference-panel { border: 3px solid CanvasText; }
}
/* 仿真 reduce 后检查根元素的 scroll-behavior 为 auto，并实际点击锚点。 */
```

## 7 图片随布局缩放与固定框裁剪

图像的 CSS 尺寸决定显示盒子，不决定服务器传输哪份文件。srcset、sizes 和 &lt;picture&gt; 处理资源选择，属于 HTML 层面的响应式图像；只缩小 CSS 宽度不会自动减少已下载文件的字节数。

max-width: 100% 限制图片不超过包含块宽度，height: auto 保持固有比例。HTML 的 width、height 同时提供尺寸与比例信息；此处资源是本章自制的 640×360 SVG，避免外部下载条件。

若设计需要固定高度图片框，object-fit: cover 等比例填满但会裁掉部分内容；contain 完整容纳但可能留空白。object-position 决定内容在框内对齐位置，不改变框本身尺寸。

```html
<div class="image-host">
  <img class="fluid-image" src="scene.svg" width="640" height="360" alt="左侧蓝色、中央绿色、右侧橙色的尺寸示意图">
  <img class="crop-image" src="scene.svg" width="640" height="360" alt="相同示意图，使用固定高度裁剪">
</div>
```

```css
.image-host { max-width: 800px; }
.fluid-image {
  display: block;
  max-width: 100%;
  height: auto;
}
.crop-image {
  display: block;
  width: 100%;
  height: 180px;
  object-fit: cover;
  object-position: center;
  margin-top: 16px;
  /* 比较源图的三色区域：固定框可能裁边；上图保持完整比例。 */
}
```

## 8 窄屏、长内容与放大后的检查

无横向滚动不等于内容可用。还要检查长链接换行、表单文字、焦点轮廓、图片比例，以及缩放后控件是否仍可到达。

普通正文的长串可用 overflow-wrap: anywhere。代码等需要保留横向结构的内容可建立局部滚动区，避免让整页一起变宽；为滚动区提供可识别的标签和键盘入口。本例保留代码行，用局部 overflow: auto。

在 320px、640px 和较宽视口观察，再把浏览器缩放到 200%；如果正文有固定高度、最小宽度或 white-space 限制，应逐一定位。不要用 body 的 overflow-x: hidden 代替排查。

```html
<div class="overflow-check">
  <p class="long-address">https://example.invalid/ABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZ</p>
  <pre class="code-scroll" tabindex="0" aria-label="可横向滚动的代码"><code>grid-template-columns: repeat(auto-fit, minmax(min(14rem, 100%), 1fr));</code></pre>
</div>
```

```css
.overflow-check { max-width: 48rem; min-width: 0; }
.long-address { overflow-wrap: anywhere; }
.code-scroll {
  max-width: 100%;
  overflow: auto;
  padding: 12px;
  border: 1px solid gray;
}
.code-scroll:focus-visible { outline: 3px solid teal; outline-offset: 3px; }
/* 320px 下只有代码区允许横向滚动；Tab 聚焦后用方向键检查可操作性。 */
```

## 本章小结

- 先让尺寸流动，内容需要结构变化时再设置断点；移动优先的基础规则适用于所有尺寸。
- @media 匹配环境条件，多个匹配规则仍参与普通层叠；边界是否包含必须明确。
- 视口宽度、像素密度与输入能力是不同条件，不能互相推断。
- 偏好查询需要搭配实际样式，并保留默认可读、键盘可用的基础。
- CSS 图片尺寸和资源选择分工不同；窄屏与缩放检查要同时关注内容和交互。
