## 1 按基础、组件和页面组织

基础样式负责排版默认值和少量统一设置；组件样式负责自身结构、变体与状态；页面样式负责组件之间的布局和页面特有安排。这是本例维护约定，不是 CSS 强制的目录结构。

入口 styles.css 用 @layer 声明 base、components、pages 的顺序，再导入三个同目录文件到相应层。@import 在普通规则之前，只有声明层顺序的 @layer 语句等允许在它之前。这里 URL 从入口样式表所在目录解析。

拆文件便于定位，不会自动产生作用域。后加载文件也必须按完整层叠规则竞争。小页面可以在单文件内按职责分段，规模扩大后再拆；网络请求的代价要结合构建与缓存测量。

```html
<link rel="stylesheet" href="styles.css">
```

```css
@layer base, components, pages;
@import url("base.css") layer(base);
@import url("components.css") layer(components);
@import url("page.css") layer(pages);
/* 先声明层次再导入；同层后续追加规则不会改变该层首次出现的位置。 */
```

## 3 类名、状态与重置的取舍

note-card 是组件名，note-card__title 表示组件内部部件，note-card--featured 表示变体。这借鉴 BEM 命名约定，双下划线和双连字符并没有 CSS 特殊作用。命名表达职责比写成 .blue-box 更便于更换视觉方案。

尽量用组件类选择器，少把 article、页面 ID、层层祖先都写进去。页面负责安排卡片间距，组件不依赖固定的外层 DOM；组件内短而必要的父子关系仍可以明确写出。

状态可以用 :disabled、:checked、:focus-visible 等原生伪类；业务状态需要类或属性时，样式只反映实际状态。给按钮添加灰色类不会真的禁用它，不能替代 HTML 的 disabled。

重置（reset）改变默认外观，统一（normalization）通常更重视保留有用默认值。没有一份重置适合全部项目。本例只统一盒模型与控件字体，保留普通列表标记和焦点提示；不要用全局 all: unset 后才逐个补救。

```html
<article class="note-card note-card--featured">
  <h2 class="note-card__title">组件边界</h2>
  <p>卡片不依赖外层必须是某一种 HTML 标签。</p>
  <button class="note-card__button" type="button" disabled>暂不可用</button>
</article>
<ul><li>普通正文列表保留标记。</li></ul>
<label for="name">姓名</label><input id="name" type="text">
```

```css
.note-card { border: 1px solid #778899; padding: 16px; }
.note-card__title { margin-block: 0 8px; font-size: 1.25rem; }
.note-card--featured { border-width: 3px; }
.note-card__button:disabled { cursor: not-allowed; color: #555; }
/* --featured 是本例的变体约定；disabled 是原生禁用状态，不是仅改变外观的类。 */
```

## 4 @layer 控制覆盖顺序

@layer 是 @ 规则；first、second 是自定层名，不是属性。层的首次出现确定顺序，随后再次写同名层是追加内容，不会把它移动到最后。匿名层不能通过名字再次追加；本例使用命名层方便维护。

同一来源、同等重要性内，层次先于选择器优先级参与比较。普通声明中后层优于前层，层外普通声明优于所有显式层，所以遗留的层外样式可能压过精心安排的组件层。

!important 的层顺序反转：前层重要声明优于后层重要声明，层内重要声明也优于层外重要声明；它没有颠倒全部 CSS 层叠规则。不要靠不断增加 !important 修补组织问题。

层不限制选择器匹配的 DOM 范围，也不阻止继承。若必须服务不支持 @layer 的旧环境，应提供经过验证的平铺样式产物或采用普通文件顺序，不能把核心样式只藏在未知 @ 规则中。本章层示例用于支持该功能的浏览器。

```html
<p id="layered" class="layer-note">后层普通声明</p>
<p class="unlayered-note">层外普通声明</p>
<p class="important-note">重要声明层次反转</p>
```

```css
@layer first, second;
@layer first {
  #layered { color: maroon; }
  .unlayered-note { color: maroon; }
  .important-note { color: maroon !important; }
}
@layer second {
  .layer-note { color: teal; }
  .important-note { color: teal !important; }
}
.unlayered-note { color: navy; }
/* 第一行teal：后层普通声明优先于前层ID；第二行navy：层外普通声明优先。 */
/* 第三行maroon：重要声明的层顺序反转；这里的 !important 仅用于规则对照。 */
```

## 5 原生嵌套与 &

原生嵌套把相关规则写在父规则内，由浏览器解析，不需要 Sass。& 是嵌套选择器，表示父选择器所匹配的元素；&:focus-within 选择处于相应状态的卡片，& > .title 选择它的直接子标题。

空格会改变关系：& .active 选择后代，&.active 选择同一个同时有 active 类的元素。原生 & 不是字符串拼接，不能用 &__title 来生成 .note-card__title；需要写完整类名。

嵌套没有让选择器天然变“轻”。普通样式规则内的 & 按父选择器列表中最高优先级计算，类似 :is()；把 ID 与类放在同一个父列表可能使所有嵌套分支都更难覆盖。这里只嵌套一层，不把 DOM 层级原样抄成 CSS。

本例直接使用原生嵌套，标题与焦点状态集中在同一组件规则内；运行浏览器须支持这些写法。边框和内边距是组件自身声明，嵌套规则再选择直接子标题和获得内部焦点的卡片。

```html
<article class="nested-card">
  <h2 class="title">嵌套组件</h2>
  <button type="button">Tab 聚焦此按钮</button>
</article>
<p class="title">组件外的同名类</p>
```

```css
.nested-card {
  border: 1px solid #778899;
  padding: 16px;
  & > .title { color: navy; text-decoration: underline; }
  &:focus-within { outline: 3px solid #a33b00; background-color: #fff0cf; }
}
/* &与:focus-within之间无空格，选择卡片自身；& > .title选择直接子元素。 */
```

## 6 @scope 的上下边界与继承

@scope (.scope-root) to (.stop) 设置选择器生效的子树范围。.scope-root 是上边界，即作用域根；.stop 是下边界，根包含在范围内，下边界元素及其后代被排除。省略 to 部分则没有额外下边界。

:scope 是伪类，在块内直接选择作用域根。@scope 前导部分的选择器优先级不会自动加到内部选择器上；但显式写出的 :scope 与其他伪类一样贡献一份类级优先级。

分别追踪“规则是否匹配”与“值是否继承”：作用域只限制前者。外部规则仍能影响范围内元素，color 也能沿元素树穿过下边界。本例因此用边框观察匹配范围，用文字颜色观察继承。

本例直接在作用域内设置根颜色与段落边框，要求浏览器支持 @scope。若另外写 .scope-root > .scoped-text，其优先级(0,2,0)高于块内.scoped-text的(0,1,0)，根再近也不能跳过优先级比较；下一节单独演示判定顺序。

本章不把 & 放进 @scope，因为这一组合在不同引擎版本中的优先级处理曾有差异，不能直接套用普通嵌套结论。

![scope-root的直接段落处于匹配范围，stop及其后代被排除，但color继承路径仍能进入stop子树。](image/illustration/19-01-scope-matching-inheritance.svg)

图 1：CMYK Labs（cmyk-labs）依据篇末 CSS Cascade 6 自绘，采用本项目许可；展示下方 @scope 的上下边界。绿色范围包含根，白色区域排除 stop 及后代；颜色继承与边框匹配沿不同规则判断。

运行后先比较两段左边框，再比较文字颜色；不要仅凭 stop 内文字也变绿，就认为作用域下边界失效。

```html
<div class="scope-root">
  <p class="scoped-text">范围内段落</p>
  <div class="stop"><p class="scoped-text">下边界内段落</p></div>
</div>
<p class="scoped-text">范围外段落</p>
```

```css
@scope (.scope-root) to (.stop) {
  :scope { color: darkgreen; }
  .scoped-text { border-left: 4px solid; padding-left: 12px; }
}
/* stop本身及后代不匹配块内规则，但color仍可从scope-root继承进去。 */
```

## 7 先优先级，再作用域接近度

作用域接近度（scoping proximity）比较目标元素沿 DOM 向上到作用域根所需的跳数，跳数少者更近；不看像素距离，也不数选择器写了多少级。

本例只比较同一文档里的作者样式表声明，没有内联、Shadow DOM、动画或过渡。先筛掉未匹配和条件不成立的规则，再依次比较来源与重要性、层顺序、选择器优先级、作用域接近度，最后才比较书写顺序。接近度不能越过更早的判定项。

第一行在 .inner 下方一跳、.outer 下方两跳，两条 .probe 优先级相同，因此较近的 .inner 赢，即使 .outer 规则写得更晚。第二行中较远规则使用 .probe.priority，优先级更高，所以先胜出，不再比较距离。

第三行刻意跨层：late 层的普通声明先胜过 early 层，再近的根和 ID 选择器也无法挽回。未作用域规则在其他条件相同时按无限距离对待；距离相同则继续比较顺序。

设计复用边界时，层回答“谁可以覆盖谁”，作用域回答“规则在哪些子树匹配”，类名回答“这是哪个部件或状态”；三者不能互相替代。

```html
<div class="outer">
  <div class="inner">
    <p class="probe">同优先级比较接近度</p>
    <p class="probe priority">较远但选择器优先级更高</p>
    <p id="layer-proof" class="layer-probe">先比较层，再考虑其他条件</p>
  </div>
</div>
```

```css
.probe { color: #333; }
@scope (.inner) {
  .probe { color: teal; }
}
@scope (.outer) {
  .probe { color: navy; }
  .probe.priority { color: maroon; }
}
@layer early, late;
@layer early {
  @scope (.inner) { #layer-proof { color: maroon; } }
}
@layer late {
  @scope (.outer) { .layer-probe { color: teal; } }
}
/* 支持scope时：第一行teal（近），第二行maroon（更具体），第三行teal（后层）。 */
```

## 本章小结

- 文件职责、类名与简短选择器共同降低结构耦合；拆文件本身没有隔离效果。
- 层次先于优先级，普通声明与重要声明的层顺序不同。
- & 表达父选择器关系，不是类名字符串拼接，嵌套仍需计算优先级。
- @scope 限制匹配，不封锁继承；优先级胜负确定后才轮到接近度。

### 提示

先固定来源与层，再计算选择器权重；排除项和继承要分别检查。

### 重点题解析

第（2）题只把首次层次声明改成second,first，后面层块内容不动。三行依次maroon、navy、teal：普通声明由后层first获胜；层外普通声明仍胜；重要声明由前层second获胜。

第（3）题加空格后选择卡片内满足focus-within的后代。焦点在按钮上时，按钮自身获得背景和轮廓，卡片不再由这条规则突出显示；完整类与DOM不变。

第（4）题先删priority类，第二行不再匹配.probe.priority，和第一行一样由较近inner给出teal。再只在inner内部包一层普通div，两根到段落的距离同时加一，仍是inner更近，颜色不变。前一步改变匹配和优先级竞争，后一步改变距离数值但不改变先后。
