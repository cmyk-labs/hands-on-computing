# 字体与文本排版

学习目标：能为中英文安排字体回退、行距和换行，并加载真实可变字体检查字重与宽度变化。

前置知识：HTML 文本元素与 lang 属性、CSS 类选择器、继承、长度单位和溢出。

适用范围：基础排版面向现代浏览器；可变字体需浏览器支持且本地字体成功加载。字体资源约 489 KB，运行时不依赖外部网络；自动断词还依赖平台词典。

工作目录：`content/Web与应用开发/css/`（以下命令从项目根目录切换到这里执行）。

环境准备：[环境配置与运行](README.md)。

配套脚本：位于 scripts/08-fonts-and-text/。

1. [index.html](scripts/08-fonts-and-text/index.html)：字体族、行高、文本装饰及可变字体。
2. [wrapping.html](scripts/08-fonts-and-text/wrapping.html)：空白、换行、断词和单行省略。

5. [fonts/OFL.txt](scripts/08-fonts-and-text/fonts/OFL.txt)：字体的原始 SIL Open Font License 1.1 及版权声明。

## 2 字体族、字重、字号与回退

font-family 是按顺序尝试的字体族列表，并不是要求同时使用所有字体。匹配还会逐字符检查字形，故同一行中文和英文可能来自不同字体。带空格的具体字体族名加引号；serif、sans-serif、monospace 等通用族不加引号，放在末尾提供回退。

font-size 控制字号，1rem 参照根元素字号；不把文字尺寸固定为屏幕物理像素。font-weight 的 normal 相当于 400、bold 相当于 700；具体粗细取决于字体提供的字面，可变字体还可使用连续字重。

font-style 的 italic 请求斜体，oblique 请求倾斜样式；缺少对应字面时浏览器可能合成样式。font 简写的常用顺序是可选样式和字重、必需字号、可选斜杠行高、必需字体族；省略的分项会重置，不宜只为改字号而覆盖已有字体配置。

```html
<p class="reading">阅读笔记 Reading notes <strong>重点内容</strong></p>
<p class="serif">衬线字体 Serif sample</p>
<p class="mono">等宽字体 Code 012345</p>
```

```css
.reading { font-family: "Microsoft YaHei", "PingFang SC", sans-serif; font-size: 1rem; font-weight: 400; }
.reading strong { font-weight: 700; }
.serif { font-family: Georgia, "Noto Serif CJK SC", serif; font-style: italic; }
.mono { font-family: Consolas, monospace; }
/* 在 Rendered Fonts 中查看实际字体；计算的 font-family 列表不是命中证明。
   系统没有某字体或该字体没有中文字形时，继续使用后续字体。 */
```

## 3 无单位 line-height 的继承

line-height 控制行盒布局使用的高度，不是段落外边距，也不等于可见字形的黑色轮廓高度。同一行有不同字号或行内元素时，最终行盒还会受它们影响。

无单位 1.5 作为数值继承，后代用自己的字号乘以 1.5。150% 则先按声明元素字号计算成长度，后代继承这个长度；本例父元素 16px，所以得到 24px。它们只在字号相同时看起来等价。

正文使用相对行高通常更适合文字缩放；不要为多行段落锁死较小高度。normal 的具体值与字体、浏览器有关，不承诺恒等于某个倍数。

```html
<div class="unitless">父元素 16px
  <p>子元素 24px：无单位行高随字号变化。</p>
</div>
<div class="percentage">父元素 16px
  <p>子元素 24px：继承父元素已计算的行高。</p>
</div>
```

```css
.unitless, .percentage { font-size: 16px; border: 1px solid #777; margin: 12px 0; }
.unitless { line-height: 1.5; }
.percentage { line-height: 150%; }
.unitless p, .percentage p { font-size: 24px; margin: 0; }
/* 子元素使用行高分别为 36px、24px；在 Computed 中核对。
   两个父元素自身均为 24px 行高，差异来自继承的是数值还是计算长度。 */
```

## 4 对齐、装饰与字词间距

text-align 对齐块容器中的行内内容。start/end 跟随文字方向，left/right 是物理方向；center 居中，justify 尝试两端对齐，可能扩大词间空隙，需要检查可读性。

text-decoration 简写设置装饰线。underline 是下划线，line-through 是删除线，none 表示本元素不添加装饰；子元素的 none 不会抹除祖先传播到文本的装饰线。下划线偏移不在该简写内，另用 text-underline-offset。

letter-spacing 添加字符间距，word-spacing 调整单词分隔处的间距；通常用 em 随字号变化。不要假定所有语言都按英文单词间空格排版；过大的正值或负值都可能妨碍阅读。

```html
<p class="aligned">把段落里的行内文字居中。</p>
<p class="decorated">装饰线穿过 <span>后代文字</span>。</p>
<p class="spacing" lang="en">Read a little every day.</p>
```

```css
.aligned { width: 280px; text-align: center; border: 1px solid #777; }
.decorated { text-decoration: underline solid #006666; text-decoration-thickness: 2px; text-underline-offset: 0.2em; }
.decorated span { text-decoration: none; }
.spacing { letter-spacing: 0.04em; word-spacing: 0.25em; }
/* text-align 不会把段落盒本身居中；后代的 none 不抹去祖先画来的装饰线。
   word-spacing 比较英文空格处的距离；中文连续字之间并不是英文单词空格。 */
```

## 5 空白处理与自动换行分开看

white-space 同时涉及空白折叠和软换行。源码换行不必然显示成页面换行；中文等文本中的空白处理也与文字系统有关，下例用拉丁字符方便比较。

- normal 折叠空白并允许自动换行。
- nowrap 折叠空白但禁止自动换行，明确的 &lt;br&gt; 仍可换行。
- pre 保留空白和源码换行，不自动折行。
- pre-wrap 保留空白和源码换行，也允许自动折行。
- pre-line 折叠连续空格，保留换行，并允许自动折行。

这些是属性值，不是五个独立属性。现代规范将 white-space 定义为若干分项的简写；基础教学先使用上述常见单关键字形式。长单词内部能否断开，由下一节属性继续决定。

```html
<p class="whitespace normal">A   B
C   D  E   F</p>
<p class="whitespace pre">A   B
C   D  E   F</p>
<p class="whitespace pre-wrap">A   B
C   D  E   F</p>
<p class="whitespace pre-line">A   B
C   D  E   F</p>
<p class="whitespace nowrap">A   B
C   D  E   F</p>
```

```css
.whitespace { width: 8ch; border: 1px solid #777; font: 20px/1.5 monospace; }
.normal { white-space: normal; }
.pre { white-space: pre; }
.pre-wrap { white-space: pre-wrap; }
.pre-line { white-space: pre-line; }
.nowrap { white-space: nowrap; }
/* 比较源码换行、连续空格与窄盒中的自动换行；pre 与 nowrap 可溢出。
   ch 以字体中数字 0 的字符进宽为参考，不保证等于任意字符的宽度。 */
```

## 6 长单词、中文断行与连字符

overflow-wrap: anywhere 在缺乏合适断点而将溢出时，允许长串在任意字符间应急折行，不插入连字符；它也影响 min-content 尺寸计算。break-word 也能提供应急折行，但新增断点不计入 min-content，布局效果不能完全等同。

word-break: break-all 更积极地允许非中日韩文本词内断行，即使整个单词移到下一行原本能放下；keep-all 抑制中日韩文字的通常词内断行，容易让中文长串溢出。普通文章先保留 normal，有长网址等问题再选合适规则。

hyphens: auto 请求按语言词典断词，必须有正确 lang，且浏览器有相应词典；断点可能因平台而异。manual 使用文本中的软连字符等显式断点，&amp;shy; 在实际发生断行时才显示连字符。none 禁止连字符断词。本例直接观察 auto；没有可用词典时按实际结果说明，不用应急折行掩盖断词条件。

这些规则仍受 white-space 的折行条件制约；不要一边禁止自动换行，一边期待应急折行自动解决溢出。

```html
<p class="long normal-break">engineeringnotebookexamplewithoutspaces</p>
<p class="long emergency">engineeringnotebookexamplewithoutspaces</p>
<p class="long break-all">Read engineering notes.</p>
<p class="long cjk">中文通常允许在字符之间换行。</p>
<p class="long keep-all">中文通常允许在字符之间换行。</p>
<p class="long hyphenated" lang="en">Internationalization improves documentation.</p>
<p class="long manual-hyphen" lang="en">inter&shy;national&shy;ization</p>
```

```css
.long { width: 160px; border: 1px solid #777; }
.normal-break { overflow-wrap: normal; }
.emergency { overflow-wrap: anywhere; }
.break-all { word-break: break-all; }
.keep-all { word-break: keep-all; }
.hyphenated { hyphens: auto; }
.manual-hyphen { hyphens: manual; }
/* 长连续字符串对照溢出与应急换行；keep-all 的中文可能溢出。
   自动断词依赖英文词典，不能预先断言断点；软连字符只在断行处显示。 */
```

## 7 单行省略必须实际发生裁剪

text-overflow: ellipsis 只指定溢出提示，不会自动缩窄元素，也不会主动制造溢出。常见单行方案同时需要有限宽度、禁止换行和裁剪。

它处理行内方向的溢出，不是“多行最后一行省略”的通用开关。盒子太窄时甚至没有空间显示完整省略符。Flexbox 内还可能需要解除项目的自动最小宽度限制，本章先用普通块盒隔离这个条件。

不要只依赖 title 的鼠标提示提供全文；本例用原生 &lt;details&gt; 与 &lt;summary&gt; 提供可键盘展开的完整内容。

```html
<p class="ellipsis">这是一段需要完整阅读但暂时以单行省略展示的课程说明。</p>
<details>
  <summary>阅读完整说明</summary>
  <p>这是一段需要完整阅读但暂时以单行省略展示的课程说明。</p>
</details>
```

```css
.ellipsis {
  width: 240px;
  overflow: hidden;
  white-space: nowrap;
  text-overflow: ellipsis;
}
/* 在有限宽度内显示单行省略；取消 overflow 后对照。
   用键盘展开完整说明，确保被省略内容仍有可操作的阅读入口。 */
```

## 8 @font-face 与加载期间的字体

@font-face 注册字体资源，本身不会给页面文字套用字体；还需在普通规则中通过 font-family 选用。

这里花括号内是描述符，与普通规则中的同名属性职责不同：

- font-family 描述符为该字体取 CSS 名称，本例为 Notebook Roboto。
- src 描述符给出来源；url() 从样式表地址解析，format() 是格式提示。也可用 local() 指定已安装字体，多个来源按顺序尝试。本例只用 URL，避免系统字体替代配套文件。
- font-style 描述符声明字面样式；font-weight 描述符声明可匹配的字重范围。本文件只包含 normal 字面，不把它冒充原生斜体。
- font-display 描述符控制加载期间显示策略。swap 的阻塞期极短，之后可先显示回退字体，字体就绪后替换；block 允许较短阻塞期；fallback 的交换期有限；optional 没有后续交换期；auto 由浏览器选择。不要把具体毫秒数当跨浏览器常量。
- unicode-range 描述符可限定哪些字符使用此字面，但不会给文件补出不存在的字形。

字体替换可能改变文字宽度和行数。通过 Network、实际渲染字体和窄屏布局一起核对；跨源字体还涉及服务器 CORS，不能仅凭 CSS 语法判断加载成功。

```html
<p class="web-font">Reading notes 012345 —— 中文使用回退字体。</p>
<p class="system-font">Reading notes 012345 —— 系统字体对照。</p>
```

```css
@font-face {
  font-family: "Notebook Roboto";
  src: url("fonts/roboto-variable.ttf") format("truetype");
  font-style: normal;
  font-weight: 100 900;
  font-display: swap;
}
.web-font { font-family: "Notebook Roboto", sans-serif; font-size: 24px; }
.system-font { font-family: sans-serif; font-size: 24px; }
/* Network 检查真实字体请求，Rendered Fonts 区分拉丁字形和中文回退。
   禁用缓存并限速刷新可观察加载过程；不要把本地快速加载当作性能结论。 */
```

## 9 用真实可变字体观察变化轴

可变字体（variable font）在一个字体资源中提供若干连续变化轴，并不是给任意普通字体写一个任意数就有变化。本章字体来自 Google Fonts 的 Roboto；资源携带 SIL OFL 1.1，分发时保留 fonts/OFL.txt。

该资源的 wght 轴范围为 100–900，wdth 轴范围为 75–100。字重优先用 font-weight；750 位于可用范围内。宽度对照在这里用 font-variation-settings 的低层形式展示轴标签，四字母标签区分大小写，数值范围由字体定义。

标准轴还包括斜度 slnt、斜体 ital 和光学尺寸 opsz，但不是每个字体都有全部轴。font-optical-sizing: auto 只对有光学尺寸轴的字体有意义；本章 Roboto 文件没有 opsz，不用它声称展示光学尺寸变化。

同一个轴避免同时用低层设置和高层属性争抢控制；可用高层属性时优先高层属性。本例仅改变拉丁字样的真实字体，中文仍会使用回退字体；不能以计算样式中的 font-weight 数字单独证明实际字形来自可变字体。

```html
<p class="variable light">Variable typography 012345</p>
<p class="variable heavy">Variable typography 012345</p>
<p><span class="variable width-normal">Variable typography</span></p>
<p><span class="variable width-narrow">Variable typography</span></p>
```

```css
.variable { font-family: "Notebook Roboto", sans-serif; font-size: 36px; }
.light { font-weight: 300; }
.heavy { font-weight: 750; }
.width-normal { display: inline-block; font-variation-settings: "wdth" 100; }
.width-narrow { display: inline-block; font-variation-settings: "wdth" 75; }
/* 字体完成加载后比较 300/750 的笔画和 100/75 的实际文字宽度。
   不支持可变字体或字体加载失败时保留可读文字，不声称变体效果已出现。 */
```

## 本章小结

- 字体族是回退列表，实际字体要按字形检查；font-family 计算值不是加载证明。
- 无单位行高随后代自身字号计算，百分比行高可能继承已算出的长度。
- 空白、自动折行、词内断点、裁剪与省略分别处理不同问题。
- @font-face 描述字体资源，普通属性选择字体；font-display 控制加载期间显示策略。
- 可变字体必须具备对应轴，字体失败时仍应保留可读文本。

## 练习

（1）把两种行高示例的子元素字号改成 32px，预测并核对其行高；检查行与行是否挤在一起。

（2）给 .emergency 临时加 white-space: nowrap。检查为何应急换行不再解决溢出，再恢复并缩窄盒子。

（3）在字体加载完成后将 .width-narrow 的 wdth 从 75 改成 90，比较相同文本的实际宽度；在 Network 中临时阻止字体请求并刷新，检查回退文字和完整说明入口仍可用。

### 提示

第一题分别考虑继承数值与继承长度；第三题关闭缓存后再观察请求，不能把缓存命中当成未发生加载。

### 重点题解析

第（1）题保持父元素字号 16px 不变。无单位 1.5 随子元素新字号计算为 48px；百分比在父元素上已算为 24px，子元素仍继承 24px。后一组文字变大而行高未变，所以可能拥挤。

第（2）题只添加 nowrap，就禁止了软换行；anywhere 不能越过这个条件。恢复 normal 后才允许应急折行；把宽度从 160px 减小，通常需要更多行。

第（3）题保持字体、字号与文本不变，wdth: 90 的宽度应处于 75 与 100 两组之间；它不是把文本盒机械缩为原来的 90%。字体被阻止后应标为回退状态，不能把该宽度冒称 Roboto 变体结果。
