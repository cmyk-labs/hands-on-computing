# 颜色、背景与装饰

学习目标：能选择清晰可读的颜色，并组合背景、边框和阴影，同时为现代颜色语法保留回退。

## 2 用 RGB 与 HSL 表示颜色

颜色可以写成关键字、十六进制或函数值。类名 hex、rgb、hsl 只是本例选择段落的名称，不是 CSS 属性。

- teal 是命名颜色；#006666 中每两位十六进制数分别表示红、绿、蓝通道，范围 00–ff。
- 三位十六进制形式 #RGB 是短写，每位重复一次，例如 #066 等于 #006666。#RRGGBBAA 和 #RGBA 的末尾部分表示 alpha：00 全透明，ff 不透明；这里 R、G、B、A 是通道占位，不是原样照抄的字母。
- rgb(0 102 102) 按红、绿、蓝排列；常用数字范围 0–255，也可用 0%–100%。
- hsl(180 100% 20%) 依次是色相、饱和度、明度：色相按角度环绕，180 等同 180deg；后两项用百分比。HSL 是 sRGB 的另一种表示，不是更大的色域。

现代函数用空格分隔通道，用斜杠引出 alpha；旧的 rgb(0, 102, 102) 与 rgba(0, 102, 102, 0.5) 仍有效。不要把逗号分隔与斜杠写法混在同一函数里。

```html
<p class="hex">十六进制颜色</p>
<p class="rgb">RGB 颜色</p>
<p class="hsl">HSL 颜色</p>
```

```css
.hex { color: #006666; }
.rgb { color: rgb(0 102 102); }
.hsl { color: hsl(180 100% 20%); }
/* 三段文字应得到相同的 sRGB 颜色；在 Computed 中比较 color。 */
```

## 3 分清颜色透明度、整体透明度与 currentColor

alpha 描述一个颜色的不透明程度，0 或 0% 表示透明，1 或 100% 表示不透明。transparent 是全透明颜色；最终看到的颜色取决于下方背景。

opacity 则作用于元素和后代合成的整体。它不继承，但子元素不能用 opacity: 1 抵消父元素已施加的透明度。只想淡化背景时，在 background-color 的颜色中设置 alpha。

currentColor 是颜色关键字，引用本元素的 color；它不是属性，也不是自定义变量。边框可随文字一起换色。元素完全透明也不等于禁用操作，不能用 opacity: 0 替代表单的 disabled。

```html
<div class="alpha-area">
  <p class="alpha-only"><strong>只让背景半透明</strong></p>
  <p class="whole-opacity"><strong>让整段半透明</strong></p>
</div>
<p class="current-color">文字与边框使用同一个前景色</p>
```

```css
.alpha-area { background-color: #f4d69a; padding: 12px; }
.alpha-only { background-color: rgb(0 102 102 / 25%); }
.whole-opacity { background-color: rgb(0 102 102); opacity: 0.25; }
.whole-opacity strong { opacity: 1; }
/* 第二段子元素设为 1，也不能抵消父元素整体合成时的透明度。 */
.current-color { color: #006666; border: 2px solid currentColor; }
/* 临时修改 color；文字和边框一起变化。 */
```

## 4 现代颜色空间与 color-mix() 的回退

色域是设备或颜色空间可表示的颜色范围。lab()/lch() 使用 CIE Lab；oklab()/oklch() 使用 Oklab，便于按感知明度等维度组织颜色。它们不是 HSL 参数的直接替换。

oklch(92% 0.03 190) 的三项是感知明度、色度和色相：明度常用 0%–100%，色度是从 0 开始的非负数，190 是角度。色度上限取决于明度、色相和目标色域，不能当作 HSL 的百分比饱和度。Oklab 用明度和两个颜色对立轴，Oklch 则把后两项改成色度与色相。

color(display-p3 0.85 0.95 0.9) 指定 Display P3 的红绿蓝通道，常用 0–1；屏幕是否能显示更广色域，还取决于设备和颜色管理。支持语法不保证显示设备覆盖相同色域。

color-mix() 是生成颜色值的函数。本例 in oklab 指定混合空间，teal 占 15%，省略的白色比例补成 85%；两色都省略比例时各占 50%。混合空间会影响结果，混合本身不保证文字对比度合格。

这些功能由 CSS Color 第 4、5 级模块定义，其中仍有演进中的语法。本章只使用已列出兼容资料的基础形式。先写基础色，再写增强声明；@supports 是按语法支持应用规则的 @ 规则，不负责检查视觉效果。

```html
<p class="modern">Oklch 增强色</p>
<p class="wide">Display P3 增强色</p>
<p class="mixed">与白色混合</p>
<p class="fallback">始终显示基础回退色</p>
```

```css
.modern, .wide, .mixed, .fallback { background-color: #e0eeee; color: #163333; }
.modern { background-color: oklch(92% 0.03 190); }
.wide { background-color: color(display-p3 0.85 0.95 0.9); }
@supports (background-color: color-mix(in oklab, teal, white)) {
  .mixed { background-color: color-mix(in oklab, teal 15%, white); }
}
/* 不识别新语法时保留基础背景。禁用增强声明后，与 fallback 对照。 */
```

## 5 背景图像的尺寸、位置与边界

背景是装饰，不替代需要替代文本的 &lt;img&gt;。tile.svg 的相对 URL 从 styles.css 所在目录出发；加载失败时，背景色仍可提供底色。

- cover 保持比例并覆盖定位区域，可能裁剪；contain 保持比例并完整放入，可能留空。两者都不会自动改变元素尺寸。
- background-position: center 将图像居中；百分比位置匹配图像与定位区对应的百分比点，图像恰好铺满该方向时，改百分比可能看不出移动。
- background-origin 选择定位区域，本例 content-box；background-clip 选择绘制边界，本例同样为 content-box，便于直接观察定位区边缘的裁剪。它们分别处理“在哪里定位”和“画到哪里”；padding-box 会把绘制范围扩至内边距，border-box 则扩至边框外缘。
- background-repeat 的 repeat 平铺，no-repeat 只画一份。background-attachment 的默认 scroll 随元素移动；local 随滚动内容移动；fixed 以视口固定背景，移动端和嵌入环境仍需专门检查。

background 简写会重置省略的分项。已有背景图时仅想换底色，应修改 background-color，避免无意清掉图像。

```html
<div class="picture cover">cover</div>
<div class="picture contain">contain</div>
```

```css
.picture {
  width: 240px;
  height: 100px;
  border: 8px dashed #555;
  padding: 12px;
  background-color: #fff4cb;
  background-image: url("tile.svg");
  background-repeat: no-repeat;
  background-position: center;
  background-origin: content-box;
  background-clip: content-box;
}
.cover { background-size: cover; }
.contain { background-size: contain; }
/* 图片固有比例 2:1，内容区为 2.4:1；cover 裁掉上下，contain 左右留空。
   内边距和边框缝隙不绘制背景；改 background-clip 为 border-box 再比较。 */
```

## 6 多重背景按层对应

background-image 用逗号分隔图层，先写的在上面，背景色在所有图片下面。渐变同样是一张背景图像。

位置、尺寸和重复方式的逗号列表按图层对应；列表不足时重复列表，多出的项不参与绘制。本例显式写两项便于观察。使用 background 简写时，底色只能放在最后一层。半透明白色上层会同时覆盖下方图案，不会淡化 HTML 文字。

```html
<div class="layers">文字始终是 HTML 内容</div>
```

```css
.layers {
  width: 240px;
  min-height: 100px;
  background-color: #fff4cb;
  background-image: linear-gradient(rgb(255 255 255 / 70%), rgb(255 255 255 / 70%)), url("tile.svg");
  background-position: center, right bottom;
  background-size: auto, 80px 40px;
  background-repeat: no-repeat, repeat;
}
/* 第一个图层在最上面；取消第一层后，对比平铺图案的清晰程度。 */
```

## 7 渐变是图像值

linear-gradient() 沿直线过渡；radial-gradient() 从一个中心向外过渡；conic-gradient() 围绕中心按角度过渡。它们生成图像值，写在 background-image 中，不能作为 background-color 的颜色值。

色标（color stop）把颜色与位置相连。线性渐变可用百分比或长度标位置，角度 0deg 指向上，90deg 指向右；相邻色标处于相同位置时出现硬边。repeating-linear-gradient() 重复首尾色标之间的区段。

本例 circle 指定圆形径向渐变，at left center 指定中心；锥形渐变从顶部开始绕中心顺时针排列色标。渐变没有固有尺寸；这里由元素和背景尺寸确定绘制范围。用于图表时还需真实文本提供数据含义。

```html
<div class="gradient linear">线性渐变</div>
<div class="gradient radial">径向渐变</div>
<div class="gradient stripes">重复渐变</div>
<div class="gradient conic">锥形渐变</div>
```

```css
.gradient { width: 240px; height: 80px; background-color: #e0eeee; }
.linear { background-image: linear-gradient(to right, #d7eeee 0%, #f5dfb2 100%); }
.radial { background-image: radial-gradient(circle at left center, #f5dfb2, #d7eeee); }
.stripes { background-image: repeating-linear-gradient(90deg, #d7eeee 0px, #d7eeee 10px, #f5dfb2 10px, #f5dfb2 20px); }
.conic { background-image: conic-gradient(#d7eeee 0deg, #f5dfb2 180deg, #d7eeee 360deg); }
/* 90deg 指向右；重复段长 20px，10px 处同位置的两个色标形成硬边。
   锥形渐变绕中心旋转；不支持时仍保留浅色背景和完整文字。 */
```

## 8 边框、圆角与阴影各管什么

border 设置边框宽度、样式、颜色；若样式为 none，仅给宽度与颜色也不会出现可见边框。分项 border-width、border-style、border-color 可以分别修改，四值按上、右、下、左排列。

border-radius 的斜杠前后分别是水平、垂直半径；四值顺序是左上、右上、右下、左下。百分比半径分别参照边框盒的宽、高。圆角影响边框和背景形状，但不自动裁剪所有后代内容，需要裁剪时另外评估 overflow，避免切掉焦点轮廓。

box-shadow 的长度依次是水平偏移、垂直偏移、模糊半径、扩展距离。模糊半径不能为负，扩展距离可为负；inset 改为内阴影，逗号添加多个阴影，先写的在上方。阴影不参与盒尺寸计算，仍可能被祖先裁剪或覆盖邻近内容。

```html
<p class="bordered">边框占据盒模型空间</p>
<p class="rounded">椭圆圆角</p>
<p class="shadowed">阴影不挤开相邻元素</p>
```

```css
.bordered { border: 4px solid #006666; border-left-style: dashed; }
.rounded { border: 4px solid #006666; border-radius: 24px / 12px; }
.shadowed {
  border: 1px solid #006666;
  box-shadow: 4px 6px 8px 1px rgb(0 0 0 / 20%), inset 0 0 0 2px #e0eeee;
}
/* 圆角横半径 24px、纵半径 12px；阴影依次为右移、下移、模糊、扩展。
   取消 box-shadow，观察相邻盒的位置不变；阴影不会给布局增加空间。 */
```

## 9 用实际背景检查颜色对比度

WCAG 2.2 的文本最低对比度要求通常是 4.5:1；大号文字是 3:1。这里“大号”指至少 18pt，或至少 14pt 的粗体；不因使用标题标签就自动满足大号条件。标识、附带文字等有特定例外，普通正文不要套用例外。

检查的是文字与其实际背景的相对亮度对比。背景图像、渐变、半透明叠加会使不同位置的背景不同，需要覆盖文字所在区域的最不利位置。HSL 或 Oklch 的明度数值不等于 WCAG 对比度。

链接、错误和状态提示还要有文字、形状或其他非颜色线索。本例通过“已保存”“待检查”传达状态；没有颜色也能读懂。

```html
<p class="status"><strong>已保存：</strong>笔记保存在本地。</p>
<p class="status warning"><strong>待检查：</strong>请补充标题。</p>
```

```css
.status { color: #163333; background-color: #fff; border-left: 4px solid #006666; padding: 12px; }
.warning { border-left-color: #9b5600; }
/* 在颜色检查器中检查实际前景与背景的对比度；状态同时有文字说明。
   修改颜色或叠加透明背景后重新检查，不沿用原配色的结论。 */
```

## 本章小结

- RGB/HSL 是颜色表示；现代颜色空间还涉及感知维度和设备色域。
- 颜色 alpha 与元素整体 opacity 的作用范围不同；currentColor 跟随本元素前景色。
- 多重背景先写的在上层，定位区域、绘制边界与图像尺寸分别设置。
- 边框参与尺寸，圆角不等于后代裁剪，阴影不为布局预留空间。
- 回退和可读性分别检查：能解析颜色值，不代表对比度合格。
