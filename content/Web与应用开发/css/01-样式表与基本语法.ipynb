## 1 HTML 与 CSS 分别负责什么

HTML 表达内容与结构，CSS（层叠样式表，Cascading Style Sheets）控制颜色、字体和布局等呈现。这里仍使用 HTML 的段落元素，重点转为它的样式。

即使没有添加 CSS，浏览器也会提供默认样式，例如标题字号和段落间距；具体外观可能随浏览器及用户设置变化。
```html
<!-- 未添加作者样式；检查标题和段落仍有浏览器默认的字号与间距。 -->
<h1>周末阅读</h1>
<p>每天留出一点时间读书。</p>
```

配套文件：[unstyled.html](scripts/01-stylesheets-and-syntax/unstyled.html) · [浏览器预览](http://127.0.0.1:8101/scripts/01-stylesheets-and-syntax/unstyled.html)

## 2 读懂一条样式规则

- p 是类型选择器（type selector），选择页面中的 &lt;p&gt; 元素。这里写的是 CSS 选择器，因此不带尖括号。
- 花括号内是声明块；color: navy 由属性 color 和属性值 navy 组成。
- 冒号分隔属性与值，分号分隔声明。本章每条声明末尾都写分号。
- navy 是深蓝色的颜色关键字；20px 表示 20 个 CSS 像素，数字和单位之间不加空格。
```css
/* 保存后刷新 index.html；段落为深蓝色，字号为 20px。 */
p {
  color: navy;
  font-size: 20px;
}
```

/* 与 */ 包围 CSS 注释，可跨行但不能嵌套；CSS 不使用 // 作为注释。缩进和声明换行用于阅读，不能用换行代替分号。


配套文件：[styles.css](scripts/01-stylesheets-and-syntax/styles.css)、[index.html](scripts/01-stylesheets-and-syntax/index.html) · [浏览器预览](http://127.0.0.1:8101/scripts/01-stylesheets-and-syntax/index.html)

## 3 本章使用的属性

| 属性 | 中文名称／含义 | 本章用途 |
| --- | --- | --- |
| color | 前景颜色 | 改变文字颜色 |
| font-size | 字号 | 改变文字大小 |
| border | 边框简写属性 | 一次设置四边的宽度、样式和颜色 |
| background-color | 背景颜色 | 为段落添加背景 |

这些是 CSS 属性。HTML 中的 style 是容纳 CSS 声明的属性，二者不是同一种语法。

## 4 三种引入方式

### 4.1 外部样式表

将 CSS 规则保存在 styles.css 中，在 index.html 的 &lt;head&gt; 中引用：
```html
<link rel="stylesheet" href="styles.css">
```

- rel="stylesheet" 表明链接目标是样式表。
- href="styles.css" 指向与当前 HTML 同目录的文件；本例没有 &lt;base&gt; 改变基准 URL。
- .css 文件直接写 CSS，不包裹 &lt;style&gt; 标签。

多个页面可以引用同一份样式表；需要统一维护页面样式时，优先使用这种方式。


配套文件：[index.html](scripts/01-stylesheets-and-syntax/index.html)、[styles.css](scripts/01-stylesheets-and-syntax/styles.css) · [浏览器预览](http://127.0.0.1:8101/scripts/01-stylesheets-and-syntax/index.html)

### 4.2 内部样式表

在 HTML 的 &lt;head&gt; 中用 &lt;style&gt; 包住 CSS。这份样式写在当前文档中，另一个页面不会自动使用它。
```html
<style>
  /* 与外部样式表示例比较：段落的颜色和字号相同。 */
  p {
    color: navy;
    font-size: 20px;
  }
</style>
```

配套文件：[internal.html](scripts/01-stylesheets-and-syntax/internal.html) · [浏览器预览](http://127.0.0.1:8101/scripts/01-stylesheets-and-syntax/internal.html)

### 4.3 内联样式

在元素的 style 属性中写声明列表，不写选择器和外层花括号。下例把声明直接附加到这个段落。
```html
<p style="color: navy; font-size: 20px;">每天留出一点时间读书。</p>
<!-- 与另两种引入方式比较：这个段落仍为深蓝色，字号为 20px。 -->
```

需要让许多元素保持相同样式时，重复写 style 不便维护。三种方式同时存在时，样式冲突按层叠规则处理；不能简单认为内部样式表总会覆盖外部样式表。


配套文件：[inline.html](scripts/01-stylesheets-and-syntax/inline.html) · [浏览器预览](http://127.0.0.1:8101/scripts/01-stylesheets-and-syntax/inline.html)

## 5 简写属性

border 一条声明可以设置四边的边框宽度、样式和颜色。这里 2px 是宽度，solid 表示实线，teal 表示蓝绿色。

border-demo 是本例自定的类名；选择器 .border-demo 匹配带有这个 class 的元素，开头的点不写进 HTML 的 class 值。
```html
<p class="border-demo">这段文字带有边框。</p>
```

```css
.border-demo {
  border: 2px solid teal;
  /* 在 Computed 中检查：四边均为 2px 实线，颜色为蓝绿色。 */
}
```

简写不是只修改写出的分项：省略的分项通常会重置为初始值。修改已有样式时，先确认自己要改整个简写，还是其中一个具体属性。


配套文件：[syntax.html](scripts/01-stylesheets-and-syntax/syntax.html)、[syntax.css](scripts/01-stylesheets-and-syntax/syntax.css) · [浏览器预览](http://127.0.0.1:8101/scripts/01-stylesheets-and-syntax/syntax.html)

## 6 无效声明怎样处理

浏览器会忽略不认识的属性或当前属性不能接受的值。下例故意保留两处错误，观察其他声明是否仍然生效。

invalid-demo 是本例自定的类名，仅用于选中第二个段落。
```html
<p class="invalid-demo">这段文字仍能应用有效声明。</p>
```

```css
/* 反例：故意保留错误属性和错误值，用来观察浏览器如何处理。 */
.invalid-demo {
  color: teal;
  colour: red;       /* 属性名拼错：这一声明被忽略。 */
  color: 20px;      /* color 不能接受长度：这一声明被忽略。 */
  font-size: 20px;  /* 字号仍生效，文字保持前面的蓝绿色。 */
}
```

这一结论针对独立的无效声明。漏分号、括号不配对或选择器错误，可能影响相邻声明或整条规则；不能概括为“写错什么都只忽略一行”。


配套文件：[syntax.html](scripts/01-stylesheets-and-syntax/syntax.html)、[syntax.css](scripts/01-stylesheets-and-syntax/syntax.css) · [浏览器预览](http://127.0.0.1:8101/scripts/01-stylesheets-and-syntax/syntax.html)

## 7 @ 规则与描述符

@ 规则（at-rule）以 @ 开头；不同规则有不同语法，并不都采用“选择器加声明块”的结构。

### 7.1 @import：导入另一份样式表

rules.html 先通过 &lt;link&gt; 加载 styles/rules.css；后者在开头导入本章已有的 styles.css。
```html
<link rel="stylesheet" href="styles/rules.css">
```

```css
@import url("../styles.css");
/* 相对于本文件导入上一层 styles.css；段落获得深蓝色和 20px 字号。 */
```

url() 表示 URL。外部 CSS 中的相对 URL 以该样式表的 URL 为基准，本例的路径关系是：
```text
scripts/01-stylesheets-and-syntax/
├── rules.html
├── styles.css
└── styles/
    └── rules.css
```

- HTML 的 styles/rules.css：从 rules.html 所在目录进入 styles。
- CSS 的 ../styles.css：从 rules.css 所在的 styles 目录退回一层。
- 内部样式表和 style 属性中的相对 URL，则按所在文档的基准 URL 解析。

本章把 @import 放在 CSS 文件开头，且不放入花括号中。它必须在普通样式规则及其他 @ 规则之前；编码声明 @charset 和声明层顺序的 @layer 语句是允许在它之前出现的例外。


配套文件：[rules.html](scripts/01-stylesheets-and-syntax/rules.html)、[styles/rules.css](scripts/01-stylesheets-and-syntax/styles/rules.css)、[styles.css](scripts/01-stylesheets-and-syntax/styles.css) · [浏览器预览](http://127.0.0.1:8101/scripts/01-stylesheets-and-syntax/rules.html)

### 7.2 @media：按条件应用样式

@media 可以按条件应用内部规则。min-width: 700px 表示浏览器视口宽度至少为 700 个 CSS 像素；它不是某个元素的宽度，也不是屏幕的物理像素数。

lightyellow 是浅黄色的颜色关键字。
```css
@media (min-width: 700px) {
  p {
    background-color: lightyellow;
    /* 视口宽度达到 700px 时出现浅黄色背景；缩窄到 700px 以下后不再应用。 */
  }
}
```

配套文件：[styles/rules.css](scripts/01-stylesheets-and-syntax/styles/rules.css)、[rules.html](scripts/01-stylesheets-and-syntax/rules.html) · [浏览器预览](http://127.0.0.1:8101/scripts/01-stylesheets-and-syntax/rules.html)

### 7.3 描述符与属性的区别

有些 @ 规则通过描述符（descriptor）配置规则自身。例如 @font-face 中的 src 指定字体来源，font-display 控制字体加载期间的显示策略；它们不是写在普通选择器规则中就会生效的 CSS 属性。

本章先区分这两类名称；@font-face 的完整写法在“字体与文本排版”中展开。

## 8 用开发者工具查看和修改

以 Chrome 为例，打开本章首页，在段落上右键选择“检查”。

1. 在 Elements 中确认选中的是 &lt;p&gt;，在 Styles 中找到 styles.css 对应的规则。
2. 临时修改 color 的值，或取消 font-size 前的勾选；在 Computed 中查看浏览器计算后的样式。
3. 点击规则旁的文件位置，确认样式来自哪份源码。普通的 Styles 临时编辑不会自动保存到本地文件；刷新后会重新加载源文件。
4. 打开 syntax.html，在 Styles 中查看故意写错的属性和值如何被标记，并与仍生效的声明区分。
5. 若外部样式未加载，在 Network 中刷新页面，检查 CSS 请求的地址、状态和响应内容。

先确认资源加载，再检查选择器是否匹配、声明是否有效，最后检查覆盖关系。被划线的声明也可能是被其他规则覆盖，不能一律当作语法错误。

## 本章小结

- CSS 规则说明选择哪些元素、设置哪些属性；值必须符合相应属性的要求。
- 外部、内部和内联是三种引入方式，维护范围和书写位置不同。
- 简写会同时设置多个分项；浏览器会忽略独立的无效声明，但结构错误可能影响更大范围。
- @import 的相对 URL 从样式表出发；@media 根据条件应用规则。
- 排错结合实际文件、Network、Styles 与 Computed，区分加载、语法和覆盖问题。
