# 关键帧动画

学习目标：能用关键帧描述多阶段变化，控制播放与组合，并在停止动画时保持界面状态可理解。

前置知识：变换、过渡、正常流、Flexbox 基本排列、状态选择器。

适用范围：现代浏览器；页面不依赖 JavaScript。

工作目录：`content/Web与应用开发/css/`（以下命令从项目根目录切换到这里执行）。

环境准备：[环境配置与运行](README.md)。

配套脚本：位于 scripts/18-keyframe-animations/。

1. [index.html](scripts/18-keyframe-animations/index.html)、[keyframes.css](scripts/18-keyframe-animations/keyframes.css)：关键帧与动画绑定。
2. [playback.html](scripts/18-keyframe-animations/playback.html)、[playback.css](scripts/18-keyframe-animations/playback.css)：方向、轮次、延迟与填充。
3. [composition.html](scripts/18-keyframe-animations/composition.html)、[composition.css](scripts/18-keyframe-animations/composition.css)：暂停和多动画冲突。
4. [state.html](scripts/18-keyframe-animations/state.html)、[state.css](scripts/18-keyframe-animations/state.css)：状态切换、布局与减少运动。

## 1 用 @keyframes 描述一轮变化

过渡根据前后状态产生变化；关键帧动画可以主动定义一轮中的多个阶段。@keyframes 是 @ 规则，travel 是本例自定且区分大小写的动画名称，animation-name 将元素与它关联。

from 与 to 分别对应 0% 与 100%；50% 表示一轮时间线中点，不是父盒宽度。块内写需要变化的属性声明，浏览器按相应动画类型连接关键帧。省略端点时会采用元素未被该动画影响的相应样式；未在某一中间帧写出的属性，不会自动在那一帧重置。

同名关键帧集合不会把所有规则拼起来；重复定义还受层叠位置影响。一个集合内相同百分比可以合并声明，同属性冲突由后写声明决定。关键帧内的 !important 被忽略；不要用它加强动画效果。

这里把动画初始设为 paused，读者勾选后才开始，不自动持续运动。

```html
<input id="play" type="checkbox"><label for="play">播放轨迹</label>
<div class="track"><div class="token">标记</div></div>
```

```css
@keyframes travel {
  from { transform: translateX(0); }
  50% { transform: translateX(80px) translateY(12px); }
  to { transform: translateX(160px); }
}
.track { width: 260px; height: 100px; border: 1px dashed; }
.token { width: 64px; background: #cce8ed; animation: travel 2s linear 1 both paused; }
#play:checked ~ .track .token { animation-play-state: running; }
/* 勾选开始；取消是暂停，不是重放。刷新页面可重新开始。 */
```

## 2 本章使用的属性

| 完整属性名 | 中文名称／含义 | 用途或对象 |
| --- | --- | --- |
| animation | 动画简写 | 组合名称、时间与播放配置 |
| animation-name | 动画名称 | 关联关键帧集合 |
| animation-duration | 单轮时长 | 决定一轮所需时间 |
| animation-timing-function | 动画缓动函数 | 控制关键帧区间的进度变化 |
| animation-delay | 动画延迟 | 推迟开始或进入中途进度 |
| animation-iteration-count | 迭代次数 | 控制播放轮数 |
| animation-direction | 播放方向 | 控制每轮的正反顺序 |
| animation-fill-mode | 填充模式 | 控制活动区间前后的样式影响 |
| animation-play-state | 播放状态 | 暂停或继续 |

## 3 时长、重复次数与播放方向

animation 简写中第一个时间是单轮时长，第二个时间是延迟；遗漏分项会回到初始值，遗漏时长通常为 0s，不能期望可见运动。名称应避免与其他分项关键字冲突。

animation-iteration-count 默认 1，可用非负小数，例如 2.5 表示两轮半，infinite 表示持续重复。延迟只在整个动画开始前应用，不会每轮重新等待。

- normal：每轮从起点到终点。
- reverse：每轮反向，缓动方向也随之反转。
- alternate：第一轮正向，下一轮反向，再交替。
- alternate-reverse：第一轮反向，然后交替。

animation-timing-function 一般应用在相邻关键帧之间；可在某帧内单独指定到下一帧的缓动。不能把一条多段动画的 ease 简单理解为只在全程起止点加减速。

```html
<input id="directions" type="checkbox"><label for="directions">播放两轮</label>
<div class="directions">
  <div class="lane"><div class="runner normal">normal</div></div>
  <div class="lane"><div class="runner reverse">reverse</div></div>
  <div class="lane"><div class="runner alternate">alternate</div></div>
  <div class="lane"><div class="runner alternate-reverse">alternate-reverse</div></div>
</div>
```

```css
@keyframes slide { from { transform: translateX(0); } to { transform: translateX(140px); } }
.lane { width: 320px; height: 52px; border-bottom: 1px dotted; }
.runner { width: 160px; background: #cce8ed; animation: slide 2s linear 2 both paused; }
.reverse { animation-direction: reverse; }
.alternate { animation-direction: alternate; }
.alternate-reverse { animation-direction: alternate-reverse; }
#directions:checked ~ .directions .runner { animation-play-state: running; }
/* alternate 的两轮是“去一次、回一次”；它不是每轮都往返。 */
```

## 4 延迟、填充与最终状态

正 animation-delay 在开始前等待；负值让动画立即处于已经运行一段时间的进度。填充模式描述活动播放区间以外，动画是否仍对样式有影响。

- none：延迟与结束后都采用基础样式。
- backwards：延迟阶段使用播放方向所决定的第一帧。
- forwards：结束后保留实际最后播放的帧。
- both：同时应用 backwards 与 forwards。

“最后播放的帧”受方向与轮数影响：alternate 播放两轮会回到 0% 一端，因此 forwards 不等于总保留 100%。无限循环没有自然结束点。

把时间分成延迟、活动区间、结束后三段，再决定每段由基础声明还是关键帧提供值。填充不会改写 HTML 状态、类名或基础声明；移除 animation 后填充也消失，长期业务状态应有独立状态样式。forwards 还可能保留动画产生的层叠上下文。

下方 normal、单轮 slide 示例在 1s 延迟内的位移依次为 20、20、0、0px，3s 结束后为 20、140、20、140px。勾选播放后分别在这两个阶段观察，解释为何 forwards 没有改变等待姿态；减少运动偏好或暂停状态会改变观察前提。

```html
<input id="fill" type="checkbox"><label for="fill">播放填充对照</label>
<div class="fills">
  <div class="lane"><div class="sample fill-none">none</div></div>
  <div class="lane"><div class="sample fill-forwards">forwards</div></div>
  <div class="lane"><div class="sample fill-backwards">backwards</div></div>
  <div class="lane"><div class="sample fill-both">both</div></div>
</div>
```

```css
.sample {
  width: 120px;
  background: #f6d69b;
  transform: translateX(20px);
  animation: slide 2s linear 1s 1 normal none paused;
}
.fill-forwards { animation-fill-mode: forwards; }
.fill-backwards { animation-fill-mode: backwards; }
.fill-both { animation-fill-mode: both; }
#fill:checked ~ .fills .sample { animation-play-state: running; }
/* 未要求减少运动时，按 none、forwards、backwards、both 的顺序核对横向偏移：
   1s 延迟内为 20、20、0、0px；3s 结束后为 20、140、20、140px。 */
```

## 5 暂停、继续与重新开始

animation-play-state 的 running 和 paused 是属性值。暂停冻结当前播放位置；继续从那里恢复，暂停所花时间不计入动画进度，延迟阶段也可以暂停。

animation: none 是移除动画，与暂停不同。取消再重新应用名称可以创建新的动画，但在同一批样式更新中先删后加可能没有可见重启；本例采用刷新重放，避免隐藏的脚本时序。

把元素或祖先改成 display: none 会终止其 CSS 动画，重新显示时按适用动画启动；不要把隐藏当作保留播放进度的暂停。无限动画只在明确勾选时运行，并提供可用键盘操作的停止入口。

```html
<input id="run" type="checkbox"><label for="run">播放；取消则暂停</label>
<div class="stage"><div class="orb single">单动画</div></div>
```

```css
@keyframes move { from { transform: translateX(0); } to { transform: translateX(160px); } }
.stage { width: 280px; height: 90px; border-bottom: 1px dotted; }
.orb { width: 80px; background: #cce8ed; animation: move 2s linear infinite alternate paused; }
#run:checked ~ .stage .orb { animation-play-state: running; }
/* 连续播放前先勾选；暂停后再勾选，应从停下的位置继续。 */
```

配套文件：[composition.html](scripts/18-keyframe-animations/composition.html)、[composition.css](scripts/18-keyframe-animations/composition.css) · [浏览器预览](http://127.0.0.1:8101/scripts/18-keyframe-animations/composition.html)

## 6 多动画的对应与冲突

用逗号为一个元素设置多个动画；每组都写名称、时长、轮数、方向与播放状态，便于读出对应关系。使用分项列表时，短列表会重复分配给 animation-name 列表。

move 改 transform，fade 改 opacity，可分别影响同一元素。move 与 turn 同时改 transform 时，本例默认的替换合成由列表后面的 turn 取代前面的 transform 效果，不自动追加函数。若希望旋转和平移都存在，可写在同一组 transform 关键帧里，或让内外两层元素分别承担动作。

普通动画值可覆盖普通作者声明，但 !important 声明可能压过动画；调试时仍须追踪层叠。不要把“动画不动”一律判断为关键帧写错。

```html
<div class="stage"><div class="orb combined">位移+透明度</div></div>
<div class="stage"><div class="orb conflict">同属性冲突</div></div>
```

```css
@keyframes fade { from { opacity: 1; } to { opacity: 0.5; } }
@keyframes turn { from { transform: rotate(0deg); } to { transform: rotate(20deg); } }
.combined { animation: move 2s linear infinite alternate paused, fade 2s linear infinite alternate paused; }
.conflict { animation: move 2s linear infinite alternate paused, turn 2s linear infinite alternate paused; }
/* 默认替换合成下，conflict 后列的 turn 接管 transform；不会自动把位移和旋转相加。 */
```

## 7 状态切换与布局影响

状态选择器决定长期的完成样式，acknowledge 只补充短暂提示；动画结束后，即使没有 forwards，完成样式仍在。这里仅定义 50% 关键帧，端点由未受该动画影响的 transform 补足。

动画是否影响布局取决于所改属性，不取决于使用 transition 还是 animation。width 变化会改变盒子尺寸并可能影响邻居；transform 的缩放保留原布局占位，视觉上可能覆盖旁边内容。

两个尺寸例子用 -1s 延迟配合 paused，停在 2s 动画的中点，便于比较静态位置。这是进度构造，不是性能测试；流畅度和渲染成本应在性能工具中实测。

```html
<input id="done" type="checkbox"><label for="done">标记已完成</label>
<p class="status">完成状态由复选框决定。</p>
<div class="row"><div class="layout-box by-width"></div><span>宽度动画的邻居</span></div>
<div class="row"><div class="layout-box by-scale"></div><span>缩放动画的邻居</span></div>
```

```css
.status { border-left: 4px solid #777; padding: 12px; }
@keyframes acknowledge { 50% { transform: translateY(-6px); } }
#done:checked ~ .status { border-color: #146536; background: #d8f0e1; animation: acknowledge 600ms ease-out; }
.row { display: flex; align-items: center; gap: 12px; margin-block: 24px; }
.layout-box { flex: none; width: 80px; height: 40px; background: #cce8ed; }
@keyframes widen { from { width: 80px; } to { width: 160px; } }
@keyframes stretch { from { transform: scaleX(1); } to { transform: scaleX(2); } }
.by-width { animation: widen 2s linear -1s both paused; }
.by-scale { transform-origin: left center; animation: stretch 2s linear -1s both paused; }
/* 两个盒子暂停在中点：width 为120px会推开邻居；scaleX(1.5)保留80px布局宽度。 */
```

## 8 减少非必要动画

prefers-reduced-motion: reduce 表达用户希望减少非必要运动。通常应移除大范围位移、缩放与循环装饰，并保留直接的状态变化；它不等于完全禁止所有视觉反馈。

本章各页为运动元素设置 animation: none。状态页还使用与原状态规则相同的选择器覆盖提示动画，以免较低优先级的减少运动规则失效。只把时长缩短到极小值不等于取消所有副作用，也不能保证界面仍表达完整状态。

不要用连续闪烁制造强调；能用静态文字、边框等表达的信息，不必持续动画。检查减少运动时内容是否仍可读，按钮与复选框是否仍能操作。

```html
<p class="status">完成状态由复选框决定。</p>
```

```css
@media (prefers-reduced-motion: reduce) {
  .token, .runner, .sample, .orb, .layout-box { animation: none; }
}
@media (prefers-reduced-motion: reduce) {
  #done:checked ~ .status { animation: none; }
}
/* 关闭提示动画后，“已完成”的边框和背景仍由状态规则保持。 */
```

## 本章小结

- 关键帧描述一轮变化，animation 把关键帧与元素和播放配置关联。
- 方向、轮数、延迟和填充共同决定何时出现哪一帧；填充不代表状态已持久保存。
- 暂停保留进度，移除动画与 display: none 不等同暂停。
- 多动画默认替换同属性效果；布局影响来自属性本身。
- 先保证状态清楚，再添加必要且可停止的运动。

## 练习

在 scripts/18-keyframe-animations/ 内修改后恢复示例。

（1）将 alternate 的轮数改为 3，再改为 2.5，预测结束位置后观察。检查时同时说明轮数和方向，不只写“停在末帧”。

（2）把 fill-both 改成 reverse，比较延迟阶段与结束后位置；再禁用它的 animation，确认基础偏移恢复。

（3）让 conflict 同时平移与旋转。检查两种效果都存在，并能由复选框暂停。

（4）启用减少运动后勾选完成状态，确认没有位移提示且完成边框、背景仍保留。

### 提示

先列出每一轮的方向；组合变换可合写到同一个关键帧集合，避免两个动画争用 transform。

### 重点题解析

第（1）题保持slide、2s单轮时长、linear和both，只改alternate轮数。三轮方向是正、反、正，结束停在140px；2.5轮在第三轮正向中点结束，停在70px。两者都不是原先两轮的0px。

第（2）题保持单轮、1s延迟和both，reverse的第一帧为100%端点，因此延迟中140px，结束0px。禁用animation后仅基础transform生效，回到20px。

第（3）题可新增一组同时声明translateX与rotate的关键帧，并让conflict只绑定这一组；保留infinite、alternate、paused和已有复选框规则。这样一个动画控制同一transform，不再由后列动画覆盖平移。
