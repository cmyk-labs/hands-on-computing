## 2 容器、项目与两条轴

给元素设置 display: flex，它成为 Flex 容器，直接参与正常流的子元素成为 Flex 项目；后代不会自动全部变成同一层项目。直接文本还可能形成匿名项目，教学示例用明确元素包住文字。绝对定位子元素脱离正常流，不参与通常的伸缩分配。

主轴（main axis）是排列项目的方向，交叉轴（cross axis）与它垂直。row 沿行内方向，column 沿块方向；row-reverse、column-reverse 反转相应主轴。方向还受 writing-mode 和 direction 影响，不能把主轴永久叫作“水平轴”。

本例采用横排、从左到右文字，所以 row 向右，column 向下。display: inline-flex 改变容器在外部的显示方式，其内部仍使用 Flex 布局。

```html
<div class="axis row-axis"><span>A</span><span>B</span><span>C</span></div>
<div class="axis column-axis"><span>A</span><span>B</span><span>C</span></div>
<div class="axis reverse-axis"><span>A</span><span>B</span><span>C</span></div>
<p>行内文字 <span class="inline-axis"><span>A</span><span>B</span></span> 后续文字</p>
```

```css
.axis { display: flex; width: 320px; max-width: 100%; border: 1px solid #777; margin: 16px 0; }
.axis span, .inline-axis span { padding: 8px; background: #e0eeee; border: 1px solid #006666; }
.row-axis { flex-direction: row; }
.column-axis { flex-direction: column; }
.reverse-axis { flex-direction: row-reverse; }
.inline-axis { display: inline-flex; }
/* 横排且从左到右时，row 主轴向右、column 主轴向下；reverse 反转主轴。
   inline-flex 让容器自身参与行内排版，内部仍是 Flex 布局。 */
```

## 3 方向、换行与 gap

默认 flex-direction 为 row，flex-wrap 为 nowrap。nowrap 把项目保留在同一行，是否缩小还取决于项目的收缩因子和最小尺寸；“不换行”不等于“永不溢出”。

wrap 允许增加行；wrap-reverse 反转交叉方向上的行堆放顺序。flex-flow 简写组合方向与换行。本例暂用 flex: 0 0 120px 固定项目基准且关闭伸缩，具体分项在后面解释。

gap 是 row-gap 与 column-gap 的简写。一个长度作用于两个分项，两个长度分别设置行间隙与列间隙；row 主轴下，第一个控制行间、第二个控制同一行项目间，column 主轴下需要重新按方向辨认。间隙不自动在容器外缘增加留白；外缘留白使用 padding。

每条 Flex 行独立分配主轴空间，下一行不是上一行的对齐网格。需要严格跨行对齐多列时，可考虑 Grid。

```html
<div class="wrap-demo"><span>A</span><span>B</span><span>C</span></div>
<div class="wrap-demo no-wrap"><span>A</span><span>B</span><span>C</span></div>
```

```css
.wrap-demo { display: flex; flex-flow: row wrap; width: 320px; max-width: 100%; gap: 8px; border: 1px solid #777; margin: 16px 0; }
.wrap-demo span { flex: 0 0 120px; background: #e0eeee; }
.no-wrap { flex-wrap: nowrap; }
/* 基准宽 120px、既不增长也不收缩：320px 内只能容纳两项和一个 8px 间隙。
   nowrap 仍只排一行；本例关闭收缩，因此第三项溢出。 */
```

## 4 项目对齐与整行对齐

justify-content 处理主轴上伸缩和自动外边距之后的剩余空间；若项目增长已用尽空间，改它可能看不到变化。flex-start、flex-end、center 靠起点、终点或居中；space-between 将余量放在项目之间，space-around 使两端间距为内部的一半，space-evenly 使间隔相等。

align-items 设置项目在各自行中的默认交叉轴对齐。stretch 通常拉伸交叉尺寸为 auto 且没有交叉轴 auto 外边距的项目，仍受最小和最大尺寸约束；center、flex-start、flex-end 不强制拉伸，baseline 可对齐文字基线。align-self 只改一个项目。

align-content 处理允许换行的容器中的行组；要看出空间分配，交叉方向需要额外空间。nowrap 单行容器中该属性无效；wrap 即使碰巧只形成一行，也属于允许多行的容器，不能只数屏幕上实际行数来判断。

Flexbox 没有逐项目的主轴 justify-self 对齐效果，需要让一项远离其他项目时，后面用自动外边距。

```html
<div class="aligned-row"><span>甲</span><span class="self-end">乙</span><span>丙</span></div>
<div class="packed"><span>A</span><span>B</span><span>C</span><span>D</span></div>
```

```css
.aligned-row { display: flex; width: 320px; height: 120px; border: 1px solid #777; justify-content: space-between; align-items: center; gap: 8px; }
.aligned-row span { padding: 8px; background: #e0eeee; }
.self-end { align-self: flex-end; }
.packed { display: flex; flex-wrap: wrap; width: 260px; height: 200px; border: 1px solid #777; gap: 10px; align-items: flex-start; align-content: space-between; margin-top: 16px; }
.packed span { flex: 0 0 100px; height: 30px; background: #f5dfb2; }
/* 上组甲丙位于交叉轴中央，乙单独到底；主轴余量被分到项目之间。
   下组为两行，每行两项；align-content 将整行分到容器交叉轴两端。 */
```

## 5 flex-basis 与 flex-grow：分的是余量

flex-basis 给出用于伸缩计算的主轴基准，不保证最终尺寸。auto 先取相应的主轴尺寸属性：row 通常是 width，column 通常是 height；相应尺寸也是 auto 时，按内容确定。非 auto 的基准在主轴伸缩计算中优先于 width/height。

基准百分比参照容器内部主轴尺寸；该尺寸不确定时，百分比可能按 content 处理。flex-basis: 0 是确定的零长度，不应在不确定高度的 column 情况下与 0% 混为一谈。

flex-grow 是非负增长因子。正余量在可增长项目之间分配；本例因子为 1 和 2，新增空间按 1:2 分配，不表示最终宽度为 1:2，因为各自还有基准尺寸。实际还需处理 gap、边框、内边距和最小、最大尺寸；项目到达限制时需重新分配。

当增长因子总和小于 1 时，可能不会用尽全部余量。本章使用整数因子 1、2，避免把这种特殊情况隐藏在基础算例中。

```html
<div class="basis-row"><span class="basis-explicit">基准 100</span><span class="basis-auto">基准 auto</span></div>
<div class="grow-row"><span class="grow-one">增长因子 1</span><span class="grow-two">增长因子 2</span></div>
```

```css
.basis-row { display: flex; width: 400px; max-width: 100%; outline: 1px solid #777; }
.basis-row span { width: 180px; flex-grow: 0; flex-shrink: 0; background: #e0eeee; }
.basis-explicit { flex-basis: 100px; }
.basis-auto { flex-basis: auto; }
.grow-row { display: flex; width: 500px; max-width: 100%; outline: 1px solid #777; margin-top: 20px; }
.grow-one { flex: 1 1 100px; background: #e0eeee; }
.grow-two { flex: 2 1 100px; background: #f5dfb2; }
/* basis 对照宽度为 100px、180px；前者的非 auto 基准用于主轴尺寸分配。
   grow-row 为 500px 时，两项各从 100px 出发，分到 100px、200px 余量，最终 200px、300px。
   本组没有 padding、border、gap 和宽度上下限干扰，便于核对计算。 */
```

## 6 flex-shrink：因子还要乘基准尺寸

放不下时分配的是负余量。每个仍可收缩项目的权重为“flex-shrink × 内部 flex base size（伸缩基准尺寸）”；不是只按 shrink 数字平均扣宽度。因此相同因子下，基准更大的项目通常承担更多缩减。

本例总基准为 600px，容器为 300px，需要缩减 300px。第一组权重 200:400；第二组权重 400:400。这里没有 padding、border、gap，且用 min-width: 0 排除自动内容下限，所以可直接用该比例核对。

flex-shrink: 0 不参与收缩，其他项目承担余量；如果所有项目的最小尺寸加起来仍然太大，就会溢出。真实算法还会把达到最小或最大约束的项目冻结并重新分配，不应将本例的一次比例运算推广成所有情况。

flex 简写依次是增长、收缩、基准。initial 相当于 0 1 auto，auto 相当于 1 1 auto，none 相当于 0 0 auto。单数字 flex: 1 在浏览器中通常展开为 1 1 0%；需要明确零基准时写出完整的 1 1 0。

```html
<div class="shrink-row proportional"><span class="small-base">200</span><span class="large-base">400</span></div>
<div class="shrink-row weighted"><span class="small-base">200 × 2</span><span class="large-base">400 × 1</span></div>
<div class="shrink-row locked"><span class="small-base">不收缩 200</span><span class="large-base">可收缩 400</span></div>
```

```css
.shrink-row { display: flex; width: 300px; outline: 1px solid #777; margin: 20px 0; }
.shrink-row span { flex-grow: 0; flex-shrink: 1; min-width: 0; }
.small-base { flex-basis: 200px; background: #e0eeee; }
.large-base { flex-basis: 400px; background: #f5dfb2; }
.weighted .small-base { flex-shrink: 2; }
.locked .small-base { flex-shrink: 0; }
/* 600px 基准要放进 300px：同为 shrink:1 时，分别减 100px、200px，剩 100px、200px。
   第二组权重为 2×200 与 1×400，各减 150px，剩 50px、250px。
   第三组第一项不缩，第二项独自减 300px，剩 200px、100px。 */
```

## 7 min-width: auto 为什么阻止收缩

Flex 项目主轴上的自动最小尺寸通常保护内容：对于本例 overflow 为 visible 的非滚动项目，min-width: auto 会产生基于内容的下限。长的不换行文本即使放在子段落中，也可能撑住外层项目。

min-width: 0 允许项目缩到更小，但不会自动修复内容。下例随后由内部段落的 overflow、white-space、text-overflow 负责裁剪和省略。注意这里需要改的是 Flex 项目 .text，不能只改它的内部段落。

若应该完整显示内容，可让长串换行或提供滚动、全文入口；不要为挤进布局无条件隐藏正文。column 主轴常遇到类似的 min-height: auto 问题，可根据滚动区域设计评估 min-height: 0。滚动项目的自动最小尺寸规则又有例外，不能断言所有 auto 都等于内容宽高。

```html
<div class="min-row protected"><span class="label">文件</span><div class="text"><p class="cut">hands-on-computing-super-long-unbroken-filename.css</p></div></div>
<div class="min-row released"><span class="label">文件</span><div class="text"><p class="cut">hands-on-computing-super-long-unbroken-filename.css</p></div></div>
```

```css
.min-row { display: flex; width: 300px; outline: 1px solid #777; margin: 20px 0; }
.label { flex: 0 0 64px; background: #f5dfb2; }
.text { flex: 1 1 0; background: #e0eeee; }
.cut { margin: 0; white-space: nowrap; overflow: hidden; text-overflow: ellipsis; }
.released .text { min-width: 0; }
/* protected 的项目保留基于内容的自动最小宽度，可能把整行撑宽。
   released 才能缩到 236px，内部段落再通过裁剪与省略处理文字。
   只把项目缩小并不会自动让内部长文本换行或省略。 */
```

## 8 自动外边距实现导航分组

主轴上的 auto 外边距在项目伸缩完成后吸收正余量，然后才轮到 justify-content。多个可用 auto 外边距会共享这些余量；因此加了 auto 后，主轴对齐属性可能没有余量可分。

本例固定为从左到右的 row，给帮助项 margin-left: auto 就把它推向另一端。遇到其他书写方向，应选合适的逻辑外边距，而不是照搬物理左边。

当空间不足时，auto 不能创造额外容量；这里允许换行并保留链接本身的可操作区域。不要把所有链接绝对定位到两端来模拟此效果。

```html
<nav class="navigation" aria-label="本章导航演示">
  <a href="#overview">概览</a><a href="#practice">练习</a><a class="help" href="#help">帮助</a>
</nav>
<p id="overview">概览入口</p><p id="practice">练习入口</p><p id="help">帮助入口</p>
```

```css
.navigation { display: flex; flex-wrap: wrap; align-items: center; gap: 12px; border: 1px solid #777; padding: 12px; }
.navigation .help { margin-left: auto; }
/* 在本例从左到右的 row 主轴中，帮助项左侧 auto 外边距吸收正的剩余空间。
   缩窄后允许换行，检查链接均可达；没有正余量时 auto 不会制造额外空间。 */
```

## 9 居中与弹性列表

两轴居中需要主轴与交叉轴都具备可分配空间。给容器 min-height，而不是给文本锁死小高度，可让内容增多时容器继续增长；单个按钮仍保留原生焦点和操作行为。

弹性列表的每项以 12rem 为内容基准，允许增长、收缩和换行。换行判断还要考虑项目外边框、内边距和间隙，不是用容器宽度简单除以 12rem。最后一行会独立分配余量，可能与上一行列宽不同。

这里保留 &lt;ul&gt; 和 &lt;li&gt;；去掉标记后额外保留 role="list"，便于需要该提示的辅助技术继续辨认列表。视觉一致不应以丢失结构为代价。

```html
<div class="centered"><button type="button">居中按钮</button></div>
<ul class="cards" role="list">
  <li><h3>选择器</h3><p>检查目标范围。</p></li>
  <li><h3>层叠</h3><p>检查声明覆盖和继承，补充较长的说明文字。</p></li>
  <li><h3>布局</h3><p>检查可用空间。</p></li>
</ul>
```

```css
.centered { display: flex; justify-content: center; align-items: center; min-height: 180px; border: 1px solid #777; }
.cards { display: flex; flex-wrap: wrap; gap: 12px; padding: 0; list-style: none; }
.cards li { flex: 1 1 12rem; min-width: 0; border: 1px solid #777; padding: 12px; overflow-wrap: anywhere; }
.cards h3 { margin-top: 0; }
/* 居中同时作用于两轴；min-height 允许内容增高，不用固定小高度截断内容。
   卡片以 12rem 内容宽为基准按行分配空间；最后一行的余量在本行重新分配。 */
```

## 10 视觉顺序与阅读顺序

order 将项目按整数分组排序，默认是 0，较小的组先排列，同组保持原有顺序。反转主轴和改变 order 都会改变视觉布局，但不会重写 DOM 顺序，通常也不改变屏幕阅读或顺序键盘导航的逻辑次序。

下面是有意保留的反例：第一步视觉上排到最后，键盘却先访问它，读者容易困惑。有因果关系的步骤、表单字段和正文，应该让 HTML 源顺序本身正确；不要借助正 tabindex 再维护一套焦点排序。

检查不能只截一张图。按 Tab 逐项移动，比较可见排列、编号、焦点路径和实际 DOM 顺序。

```html
<ol class="visual-order">
  <li class="moved"><button id="order-one" type="button">第一步</button></li>
  <li><button id="order-two" type="button">第二步</button></li>
  <li><button id="order-three" type="button">第三步</button></li>
</ol>
```

## 本章小结

- 先判断主轴和交叉轴，再区分容器属性、项目属性、单行与行组对齐。
- flex-basis 是起点，grow 分正余量，shrink 还要乘内部基准尺寸作权重。
- 最小尺寸可阻止收缩；解除项目下限后仍需处理内部文本。
- auto 外边距吸收余量，gap 留项目间距；两者不替代外缘 padding。
- 多行 Flex 各行独立分配，视觉重排不修复 HTML 阅读顺序。
