## 1 内容、内边距、边框与外边距

一个盒子从内向外依次是内容区（content）、内边距区（padding）、边框区（border）与外边距区（margin）。文字和图片放在内容区；padding 在边框内留空，margin 位于边框外。

| 完整属性名 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| width | 宽度 | 给当前尺寸盒指定宽度 |
| height | 高度 | 给当前尺寸盒指定高度 |
| padding | 内边距简写 | 设置内容与边框之间的空间 |
| border | 边框简写 | 设置边框宽度、样式与颜色 |
| margin | 外边距简写 | 设置边框外的布局间距 |
| box-sizing | 尺寸计算方式 | 选择 width/height 使用内容盒还是边框盒 |

padding 与 margin 的一个值作用于四边，两个值依次为上下、左右，三个值为上、左右、下，四个值按上右下左排列。margin 可为负数，padding 不能为负数。

下面默认 content-box，200px 指内容宽；边框外沿宽度为 200 + 20×2 + 4×2 = 248px。margin 不在这个测量内；背景默认延伸到边框区域，外边距自身透明。

```html
<div class="edge-box">内容区里的文字</div>
```

```css
.edge-box {
  width: 200px;
  padding: 12px 20px;
  border: 4px solid teal;
  margin: 16px 10px;
  background-color: lightyellow;
}
/* 在盒模型图核对：内容宽 200px，左右 padding 各 20px，边框各 4px；
   边框外沿宽 248px，左右 margin 另各占 10px。 */
```

## 2 box-sizing 改变宽度包含的区域

content-box 中，width/height 不包含 padding 和 border；border-box 中，它们被计入指定尺寸。两者都不包含 margin，也不会阻止外边距折叠。

先在图中标出 width 的起止边界，再计算两侧额外占用。下面两盒都写 200px：content-box 的外沿为 200 + 2 × 20 + 2 × 4 = 248px；border-box 的内容宽为 200 − 2 × 20 − 2 × 4 = 152px。box-sizing 同样影响 min/max 尺寸属性的尺寸盒。

border-box 不能把内容压到负数：如果指定尺寸比 padding 与 border 的合计还小，内容区下限为 0，外沿仍可能大于指定值。也不要因设置 border-box 就省略长内容与缩放检查。

![content-box 的 200px 仅覆盖内容，边框外沿 248px；border-box 的 200px 包括内边距与边框，内容剩152px。](image/illustration/05-01-box-sizing-widths.svg)

图 1：同一组声明的横向尺寸。示意图按横向尺寸比例绘制，竖向高度只用于标清区域，不是浏览器截图。

运行下方示例后，在开发者工具盒模型中分别找 200、248、152，对照它们到底量到哪条边。

```html
<div class="sizing-demo sizing-content">content-box</div>
<div class="sizing-demo sizing-border">border-box</div>
```

```css
.sizing-demo { width: 200px; padding: 20px; border: 4px solid teal; margin-bottom: 12px; }
.sizing-content { box-sizing: content-box; }
.sizing-border { box-sizing: border-box; }
/* 比较外沿宽度：248px 与 200px；开发者工具中后者内容区为 152px。 */
```

## 3 宽高与最小、最大尺寸

width/height 是首选尺寸，min-* 与 max-* 给使用尺寸加限制。它们是不同属性之间的布局约束，不是选择器优先级竞争。

| 完整属性名 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| min-width | 最小宽度 | 限制宽度下界 |
| max-width | 最大宽度 | 限制宽度上界 |
| min-height | 最小高度 | 限制高度下界 |
| max-height | 最大高度 | 限制高度上界 |

一般先根据 width/height 求尺寸，再受最小和最大尺寸约束。最小尺寸大于最大尺寸时，最小限制优先，不能期待 max-width 永远防溢出。

auto 表示由布局求值；普通流内非替换块的 auto 宽度通常利用可用空间，高度按内容展开。固定高度或 max-height 不会自动让多余内容消失；可增长的正文通常更适合 auto 高度搭配必要的 min-height。

```html
<div class="limit-width">width 与 max-width</div>
<div class="conflict-width">min-width 大于 max-width</div>
<div class="limit-min-height">最小高度</div>
<div class="limit-max-height">最大高度</div>
```

```css
.limit-width { width: 400px; min-width: 120px; max-width: 240px; background-color: lightyellow; }
.conflict-width { width: 200px; min-width: 280px; max-width: 240px; background-color: lightyellow; }
.limit-min-height { height: 40px; min-height: 80px; border: 1px solid teal; }
.limit-max-height { height: 120px; max-height: 60px; border: 1px solid teal; }
/* 检查内容尺寸：宽度分别 240px、280px；高度分别 80px、60px。
   给最后一盒添加多行文字，观察高度限制与内容溢出是两件事。 */
```

## 4 相邻兄弟的外边距折叠

外边距折叠（margin collapsing）把相邻外边距合成一个间距。本章的上下方向对应横排文档中的块方向；普通横向外边距不折叠。

成立条件包括：普通流内块级盒参与同一块级格式化上下文（BFC），相邻边之间没有会隔开的行盒、边框、padding 或清除浮动产生的间隙。浮动盒、绝对定位盒、Flex/Grid 项目的外边距不能套用本例。

两项都为正时取较大值；有正负值时，取最大正值加最负的负值；全为负时取最负者。不是简单把两个值相加，也不是一律取绝对值最大。

下面外框的边框只用来隔开演示与外部布局，不会隔开两个段落之间的外边距。

```html
<div class="margin-well">
  <p class="margin-first">前一段</p>
  <p class="margin-second">后一段</p>
</div>
<div class="margin-well negative-well">
  <p class="margin-first">前一段</p>
  <p class="margin-second">后一段：负上边距</p>
</div>
```

```css
.margin-well { border: 1px solid gray; margin-bottom: 16px; }
.margin-well > p { margin: 0; background-color: lightyellow; }
.margin-well > .margin-first { margin-bottom: 24px; }
.margin-well > .margin-second { margin-top: 40px; }
.negative-well > .margin-second { margin-top: -10px; }
/* 测量两段边框边缘之间的距离：第一组 40px；第二组 24 + (-10) = 14px。 */
```

## 5 父子与空块的折叠边界

父块没有上边框、上内边距、分隔行盒等条件时，首个流内子块的上外边距可与父块折叠，间距出现在父块外面；父 margin 为 0 也可能发生。

末个流内子块与父块的下外边距折叠还要求父 height 为 auto，并满足 min-height 等限制。空块若没有边框、padding、行内内容或有效高度把上下隔开，其自身上下外边距也可能折叠并继续贯通。

本例加 1px 上内边距隔开父子顶部，便于观察条件变化。建立新的 BFC（例如父块 display: flow-root）也能阻止父子折叠；但它不会把内部普通兄弟之间的折叠一并关闭。这里的 flow-root 用于建立独立的块级格式化上下文。

```html
<div class="parent-well"><div class="parent-open"><p>父子可折叠</p></div></div>
<div class="parent-well"><div class="parent-padded"><p>顶部被内边距隔开</p></div></div>
<div class="empty-well">
  <div class="empty-block"></div>
  <p class="after-empty">空块之后</p>
</div>
```

```css
.parent-well, .empty-well { border: 1px solid gray; margin-bottom: 16px; }
.parent-open, .parent-padded { background-color: lightyellow; }
.parent-open > p, .parent-padded > p { margin: 24px 0 0; }
.parent-padded { padding-top: 1px; }
.empty-block { margin: 20px 0 30px; }
.after-empty { margin: 0; background-color: lightyellow; }
/* 第一组父盒顶部与段落顶部重合；第二组段落位于父盒顶部下方 25px。
   空块上下外边距合为 30px；从外框内容顶部到后一段测量，不是 50px。 */
```

## 6 固有尺寸与内容尺寸关键字

固有尺寸（intrinsic size）由内容本身决定，不是先给盒子指定一个固定宽度。文本可根据断行机会形成最小与最大内容尺寸；图像还可能自带自然宽高和比例。

- min-content：尽可能利用允许的断行机会得到的最小内容宽度。本例英文在空格处可断行，因此受最长单词限制；中文和不同断词规则会不同。
- max-content：不进行软换行时所需的宽度；显式换行仍有意义，不代表可以忽略所有换行。它可能超过父容器。
- fit-content：在 min-content 与 max-content 之间，尽量利用可用空间。最小内容本身超过容器时，仍可能溢出。

它们是尺寸关键字，不是属性。fit-content 关键字与 fit-content(...) 函数要区分，函数可用于网格轨道等场景，本章不依赖函数在普通尺寸属性上的支持范围。

```html
<div class="intrinsic-parent">
  <p class="intrinsic-min">CSS notebook layout examples</p>
  <p class="intrinsic-max">CSS notebook layout examples</p>
  <p class="intrinsic-fit">CSS notebook layout examples</p>
</div>
```

```css
.intrinsic-parent { width: 180px; border: 1px solid gray; }
.intrinsic-parent > p { font-family: monospace; margin: 8px 0; background-color: lightyellow; }
.intrinsic-min { width: min-content; }
.intrinsic-max { width: max-content; }
.intrinsic-fit { width: fit-content; }
/* 比较三盒：min 最窄且多行；max 单行越过 180px；fit 利用 180px 并换行。
   改字体或内容后重新测量，不能把某个单词的像素宽度当成规范常量。 */
```

## 7 aspect-ratio 是首选宽高比

aspect-ratio 指定首选宽高比，16 / 9 中左边是宽、右边是高，两个数字表达比例而不是像素。它参与自动尺寸的计算，至少有一边为 auto 才能按这里的方式推导另一边。

宽高都明确指定时，比例不会强行改写它们；min/max 约束和内容也可能让最终盒子偏离比例。自定比例使用 box-sizing 所指定的尺寸盒；替换内容自身的自然比例则依据内容尺寸。

下面使用课程约定的现代 Chrome，直接设置 aspect-ratio；未指定的高度保持 auto，以固定的 240px 宽度推导高度。

```html
<div class="ratio-auto">16 / 9 自动高度</div>
<div class="ratio-fixed">宽高都已指定</div>
<div class="ratio-min">最小高度改变结果</div>
```

```css
.ratio-auto, .ratio-fixed, .ratio-min { width: 240px; border: 1px solid teal; margin-bottom: 8px; }
.ratio-auto, .ratio-min { aspect-ratio: 16 / 9; }
.ratio-fixed { height: 100px; aspect-ratio: 16 / 9; }
.ratio-min { min-height: 160px; }
/* 检查内容高度依次为 135px、100px、160px；后两组没有保持 16:9。 */
```

## 8 替换元素与图像自然尺寸

&lt;img&gt; 属于替换元素（replaced element）：盒中呈现的图像来自外部资源，其内部内容不按普通 HTML 子元素排版。替换元素可能自带自然宽高与宽高比，但不是每一种替换元素都有全部这些信息。

本章自带的 pattern.svg 为 300×150，比例 2:1。HTML 的 width/height 属性给出尺寸信息，有助于在加载前预留比例；CSS 再控制显示宽度，height: auto 用自然比例推导高度。

aspect-ratio: auto 1 / 1 用在具有自然比例的替换元素上时，载入自然比例后用 auto 对应的比例；没有自然比例的普通盒仍用所写比例。这里保留自然比例，不强制把图像变成正方形。

```html
<img class="natural-image" src="pattern.svg" width="300" height="150"
     alt="左边橙色圆形，右边蓝绿色矩形">
<img class="scaled-image" src="pattern.svg" width="300" height="150"
     alt="相同比例的缩小图形">
```

```css
.natural-image { display: block; max-width: 100%; height: auto; }
.scaled-image { display: block; width: 150px; max-width: 100%; height: auto; }
/* 页面可用宽度不少于 300px 时，图像依次显示为 300×150 和 150×75。
   src 从 images.html 出发，指向同目录的 pattern.svg。 */
```

## 9 object-fit 与 object-position 控制盒内图像

先给图片盒相同宽高，再比较里面的图像怎样适配。object-fit 不负责决定元素外部占用尺寸，也不适用于普通文字盒；它对 &lt;iframe&gt; 等嵌入元素也没有这里的作用。

| 完整属性名 | 中文名称／含义 | 用途 |
| --- | --- | --- |
| aspect-ratio | 首选宽高比 | 参与自动尺寸计算 |
| object-fit | 替换内容适配方式 | 决定图像如何装入内容盒 |
| object-position | 替换内容位置 | 决定盒内的对齐与裁切位置 |

- fill 是默认值，让替换内容尺寸与内容盒一致；普通位图在比例不同时会拉伸变形。
- contain 保持比例并完整放入，比例不同时留空。
- cover 保持比例并覆盖整个盒子，多出的部分裁掉。
- none 保持内容自然尺寸；scale-down 在 none 与 contain 的结果中选更小尺寸。

本例 SVG 设置了 preserveAspectRatio="none"，允许图形随视口拉伸；SVG 自身的比例设置也会影响画面，不能只看 object-fit 判断是否变形。

object-position: left center 将图像靠左、垂直居中；right center 改为靠右。默认是 50% 50%。百分比对齐把图像对应的百分比点与盒子的对应点对齐，不能当成相对父元素的移动量。裁切可能隐藏重要信息，需选择合适的焦点位置或改用 contain；必要含义仍放在 HTML 文本或 alt 中。

```html
<p>依次为 fill、contain、cover 靠左、cover 靠右、none、scale-down：</p>
<img class="fit-image fit-fill" src="pattern.svg" alt="fill 对照">
<img class="fit-image fit-contain" src="pattern.svg" alt="contain 对照">
<img class="fit-image fit-cover-left" src="pattern.svg" alt="cover 靠左对照">
<img class="fit-image fit-cover-right" src="pattern.svg" alt="cover 靠右对照">
<img class="fit-image fit-none" src="pattern.svg" alt="none 对照">
<img class="fit-image fit-down" src="pattern.svg" alt="scale-down 对照">
```

```css
.fit-image { width: 150px; height: 120px; border: 2px solid gray; background-color: lightyellow; }
.fit-fill { object-fit: fill; }
.fit-contain { object-fit: contain; }
.fit-cover-left { object-fit: cover; object-position: left center; }
.fit-cover-right { object-fit: cover; object-position: right center; }
.fit-none { object-fit: none; }
.fit-down { object-fit: scale-down; }
/* 六个元素外沿尺寸都为 154×124；观察圆形是否变形、上下留空及左右裁切。
   当前自然尺寸大于盒子，scale-down 与 contain 相同；none 则保持 300×150 后裁切。 */
```

## 本章小结

- width/height 指向哪个尺寸盒由 box-sizing 决定，margin 始终另算。
- min/max 是布局约束；最小限制冲突时可能压过最大限制。
- 折叠只在适用格式化条件下出现，兄弟、父子与空块要分别判断。
- 固有内容尺寸、首选比例与指定尺寸共同参与布局；object-fit 只处理替换内容在盒内的适配。

## 练习

在 scripts/05-box-model-and-sizing/ 的配套文件中操作，先记录原值，练习后恢复。

（1）把 .sizing-border 的 padding 从 20px 改成 30px，检查外沿宽度仍是 200px，内容区缩小；再把 width 改为 20px，检查内容不能被压为负数。

（2）为 .parent-open 增加 padding-top: 1px，检查顶部折叠消失；恢复后改用 display: flow-root，比较两种方式的差异。

（3）把 .intrinsic-parent 宽度改为 40px，检查 fit-content 仍可能因最长单词而溢出。

（4）将两张 cover 图片的 object-position 对调；检查图片盒不移动，但可见图形区域改变。再增大 fit-down 的盒子到 600×300，比较 scale-down 与 contain 是否仍相同。

### 提示

尺寸计算先标清内容边、padding 边和边框边；图片题既检查元素尺寸，也观察图像像素，不能只看属性值。

### 参考解析（第 1 题）

内边距改为 30px 时，内容宽为 200−2×30−2×4=132px。再把 border-box 宽度改为 20px，内容宽下限为 0，外沿仍需容纳 60px 内边距和 8px 边框，因此实际外沿宽为 68px。
