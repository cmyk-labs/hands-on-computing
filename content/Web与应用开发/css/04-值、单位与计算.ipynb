## 1 值的类型与关键字

先看一条声明接受什么。数字、长度、百分比是不同的值类型；属性定义决定允许哪种类型和取值范围。

- 1.5 是没有单位的数字，可用于 line-height，表示字号的倍数；数字可以有小数，但要求整数的语法不能任意写小数。
- 24px 是长度，50% 是百分比；数字与单位或百分号之间不能留空格。
- auto、none 等是关键字，含义取决于属性。width: auto 让布局算法求宽度，不等于宽度为 0；不能把每个关键字都当成通用值。
- rgb()、calc()、rotate() 等是函数形式的值，各自有参数语法，并非任意函数都能写入任意属性。
- 长度的零通常可写 0；时间仍写 0s，角度示例写 0deg，避免把“零可以省单位”推广到全部类型。

下面 width 用长度、line-height 用数字、color 用颜色函数，三者分别服务于尺寸、行高与前景颜色。

```html
<p class="value-types">一条声明先确定属性，再选择它接受的值。</p>
```

```css
.value-types {
  width: 240px;
  font-size: 20px;
  line-height: 1.5;
  color: rgb(0, 0, 128);
  margin: 0;
}
/* 在开发者工具中检查：本例行高为 30px；长度 240px 与数字 1.5 不能互换。 */
```

## 2 px、em 与 rem 的参照

px 是 CSS 像素，不承诺等于一个设备物理像素。页面缩放、设备像素密度会影响两者对应关系。绝对长度之间有固定换算，如 1in = 96px；屏幕上的 1cm 不保证能用实体尺量出一厘米。

em 用于 font-size 时参照父元素字号；用于 padding 等其他属性时，参照元素自身的计算字号。嵌套地反复设置 font-size: 1.5em 会逐层放大。

rem 参照根元素字号，通常是 &lt;html&gt;；它不是永远等于 16px。本例把根字号设为 20px 只是为了算清参照，不是项目通用重置规则。

```html
<div class="font-parent">
  <div class="em-child">相对父字号
    <span class="em-grandchild">再放大一次</span>
  </div>
  <p class="rem-child">相对根字号</p>
</div>
```

```css
html { font-size: 20px; }
.font-parent { font-size: 24px; }
.em-child { font-size: 1.5em; padding-left: 0.5em; }
.em-grandchild { font-size: 1.5em; }
.rem-child { font-size: 1.5rem; }
/* 检查：em-child 字号 36px、左内边距 18px；孙元素 54px；rem-child 为 30px。 */
```

## 3 百分比必须问参照对象

百分比不是一种固定长度，参照由属性及布局条件决定。包含块（containing block）是布局时计算尺寸和位置的参照区域；本章的普通块级例子中，它来自父块的内容区域。

- width 的百分比参照包含块宽度。
- 本章横排布局中，padding 的上下左右百分比都参照包含块宽度；padding-top: 10% 不是父高度的 10%。
- font-size 的百分比参照父元素字号。
- height 的百分比需要能确定的参照高度。本例普通流内子块的父高度为 auto 且依赖内容，不能靠 height: 50% 得到屏幕的一半；显式设置父高度后才有这里需要的参照。

其他布局如 Flex、Grid 对确定尺寸另有规则，不把“父元素必须写 px 高度”当成百分比高度的唯一条件。

```html
<div class="percent-parent">
  <div class="percent-size">宽度与上内边距</div>
</div>
<div class="height-auto"><div class="half-height">父高度 auto</div></div>
<div class="height-fixed"><div class="half-height">父高度 120px</div></div>
```

```css
.percent-parent { width: 320px; max-width: 100%; border: 1px solid gray; }
.percent-size { width: 50%; padding-top: 10%; background-color: lightyellow; }
.height-auto, .height-fixed { width: 240px; border: 1px solid teal; }
.height-fixed { height: 120px; }
.half-height { height: 50%; background-color: lightyellow; }
/* 容器宽度为 320px 时：子宽 160px，上内边距 32px。
   两组高度对照：auto 父框按内容展开；固定父框中子高为 60px。 */
```

## 4 视口单位与动态浏览器界面

视口是浏览器用于布局页面的区域。1vw 是视口基准宽度的 1%，1vh 是基准高度的 1%；vmin 取两者较小值，vmax 取较大值。它们参照视口，不参照当前父元素。

移动浏览器工具栏展开与收起时，可用区域会变化。svh 按小视口，lvh 按大视口，dvh 随动态视口变化；当前规范中的 vh 对应大视口基准，因此工具栏展开时 100vh 可能大于可见高度。

动态视口不保证与软键盘出现后的区域完全相同，浏览器可能把部分界面作为覆盖层。dvh 也可能在滚动中改变尺寸；需要稳定排版时可考虑 svh，而不是盲目把 vh 全部替换为 dvh。

本例先写 vh，再写 dvh。旧浏览器会忽略不认识的单位并保留前一条有效声明。桌面窗口中几种视口可能恰好相同，不能据此断言手机也相同。

```html
<div class="viewport-width">宽度为 50vw</div>
<div class="viewport-height">最小高度为 20dvh，旧环境回退 20vh</div>
```

```css
.viewport-width { width: 50vw; background-color: lightyellow; }
.viewport-height {
  min-height: 20vh;
  min-height: 20dvh;
  border: 1px solid teal;
}
/* 调整视口宽高分别观察两组；长内容仍可把 min-height 对应的框撑高。 */
```

## 5 calc() 混合长度与百分比

calc() 在属性值中做运算，结果必须符合该属性接受的类型。width 可接受长度与百分比的组合，因此能写 calc(100% - 24px)。

加减两边要留空格；乘除按通常先后顺序，也可用括号分组。本章采用兼容性较稳的“尺寸乘以无单位数字”“尺寸除以非零无单位数字”，不依赖新扩展中的带单位相除。

长度不能与时间相加；auto 也不能当普通数字参与本例 calc() 运算。百分比仍使用属性规定的参照对象，函数不会替你换参照。

```html
<div class="calc-parent">
  <div class="calc-width">留出 24px</div>
  <div class="calc-scale">先加长度再除以 2</div>
</div>
```

```css
.calc-parent { width: 320px; max-width: 100%; border: 1px solid gray; }
.calc-width { width: calc(100% - 24px); background-color: lightyellow; }
.calc-scale { width: calc((100px + 20px) / 2); border: 1px solid teal; }
/* 父内容宽 320px 时，第一个内容宽 296px；第二个内容宽 60px，不含边框。 */
```

## 6 min()、max() 与 clamp() 控制范围

min() 返回参数中的较小值，因此常用来给宽度设上限；max() 返回较大值，常用来保证下限。参数须能按同一种兼容类型比较，百分比是否可用仍由属性决定。

clamp(160px, 50vw, 400px) 的三个参数依次是下限、偏好值和上限：先尽量采用 50vw，超出范围时取边界。一般写法 clamp(MIN, VAL, MAX) 中，MIN、VAL、MAX 分别代表这三项；等价于 max(MIN, min(VAL, MAX))。如果下限大于上限，下限优先，不能认为浏览器自动交换了参数。

尺寸下限可能让极窄容器溢出，max() 不会自动防止这一点。用这些函数给字号加边界时，还要检查文字放大和用户默认字号；本例只演示尺寸，字体基准的选择另行检查。

```html
<div class="bounded-min">min：宽度不超过 240px</div>
<div class="bounded-max">max：左右内边距至少 8px</div>
<div class="bounded-clamp">clamp：160px 至 400px</div>
```

```css
.bounded-min { width: 100%; max-width: 240px; width: min(100%, 240px); background-color: lightyellow; }
.bounded-max { padding-left: max(8px, 2vw); padding-right: max(8px, 2vw); border: 1px solid teal; }
.bounded-clamp { width: 240px; width: clamp(160px, 50vw, 400px); background-color: lightyellow; }
/* 视口 900px 时 clamp 宽度为 400px；400px 时为 200px；280px 时触及下限 160px。
   min 与 clamp 前面保留普通声明作旧环境回退；极窄视口仍需检查内容溢出。 */
```

## 7 角度和时间也是带单位的值

角度可用 deg、grad、rad、turn：一周分别是 360deg、400grad、2πrad、1turn，π 是圆周率。下面 rotate(0.125turn) 表示八分之一周；transform 是属性，rotate() 是它的函数值。

时间单位 s 表示秒，ms 表示毫秒，1s = 1000ms。transition-duration: 200ms 是 0.2 秒；该属性不接受负持续时间。某些延迟属性可接受负数，不能只按“时间类型允许负号”推断每个属性都接受。

下面按钮用 :hover/:focus 改变背景，以观察时间值。减少动态效果偏好为 reduce 时把持续时间改为 0s；这里只使用条件，完整过渡与偏好设计在后续章节展开。

```html
<div class="angle-frame"><div class="angle-sample">转</div></div>
<button class="time-sample" type="button">悬停或聚焦观察颜色过渡</button>
```

```css
.angle-frame { padding: 24px; }
.angle-sample { width: 40px; height: 40px; border: 1px solid teal; transform: rotate(0.125turn); }
.time-sample { background-color: white; transition-property: background-color; transition-duration: 200ms; }
.time-sample:hover, .time-sample:focus { background-color: gold; }
@media (prefers-reduced-motion: reduce) {
  .time-sample { transition-duration: 0s; }
}
/* 检查旋转角度为 45 度；颜色过渡在 200ms 后到达 gold，减少动态效果时直接切换。 */
```

## 8 指定值、计算值与使用值

指定值（specified value）来自层叠及默认处理，不等于“源码里出现的每条值”。计算值（computed value）按属性定义解析继承、相对单位等依赖，例如父字号 20px 时，font-size: 150% 计算为 30px。

使用值（used value）在布局需要的参照确定后得到。width 的计算值可以保留百分比；本例 320px 包含块中的 width: 50% 使用值为 160px。

浏览器的 getComputedStyle() API 返回解析值（resolved values），按属性及布局条件可能是计算值，也可能是使用值。即使工具把宽度显示成 px，也不能倒推出源码写的是 px。本章无需写 JavaScript，使用 Styles 看原声明、盒模型看布局尺寸来对照。

```html
<div class="processing-parent">
  <p class="processing-child">150% 字号与 50% 宽度</p>
</div>
```

```css
.processing-parent { width: 320px; max-width: 100%; font-size: 20px; }
.processing-child { font-size: 150%; width: 50%; margin: 0; background-color: lightyellow; }
/* 父内容宽 320px 时：Styles 仍显示 150% 与 50%；布局宽度为 160px，字号为 30px。 */
```

## 9 无效值与单位选择

能解析成某个类型不等于该属性接受它。width: 12 缺少长度单位，padding: -4px 超出允许范围，width: calc(20px + 1s) 的类型无法相加。独立无效声明在解析时被忽略，前面的有效声明仍可参与层叠。

这里说的是能在解析时判定的错误。包含 var() 的声明可能到计算值阶段才发现无效，此时不会简单重跑层叠去捡上一条声明；相关机制在自定义属性章节展开。

选择单位时先问参照：随文字大小变化用 em/rem，随容器比例变化看百分比，随视口变化用视口单位，小边框等可用 px。无论选哪种，都需检查长文本、窄视口和缩放，不能靠单位名称保证可访问性。

```html
<div class="invalid-value">保留解析阶段的无效值反例</div>
```

```css
.invalid-value {
  width: 180px;
  width: 12;                    /* 反例：非零宽度缺少单位。 */
  width: calc(20px + 1s);       /* 反例：长度与时间不能相加。 */
  padding: 8px;
  padding: -4px;                /* 反例：内边距不接受负值。 */
  transition-duration: 200ms;
  transition-duration: 0;      /* 反例：时间的零也需单位。 */
  border: 1px solid teal;
}
/* 检查有效值保留：内容宽 180px、padding 8px、transition-duration 0.2s。 */
```

## 本章小结

- 数字、长度、百分比、角度和时间各有语法与适用范围，关键字的含义也由属性决定。
- em、rem、百分比和视口单位的参照不同；百分比高度还依赖布局中的确定尺寸。
- 数学函数组合与约束值，不消除类型要求，也不自动修复溢出。
- Styles 中的原声明、规范计算值与布局使用值需区分；getComputedStyle() 返回 resolved values。
