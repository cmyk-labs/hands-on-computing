## 1 inline 与 block 不是固定的左右上下

行内方向（inline direction）沿一行文字前进；块方向（block direction）沿下一行或下一个块排列。每条轴都有 start 与 end，它们表示起点和终点，不是固定屏幕边缘。

- horizontal-tb：行横向排列，新行在下方；在 ltr 下，inline-start 是左，block-start 是上。
- vertical-rl：行竖向排列，新列在左边；本例 ltr 下，inline-start 是上，block-start 是右。
- vertical-lr：行竖向排列，新列在右边；本例 ltr 下，inline-start 是上，block-start 是左。

这些是 writing-mode 的属性值。三幅规范原图用编号 1、2、3 表示内容顺序，粗箭头表示块推进方向；细线表示文本行，方框表示其他内容。

下方三段保持相同结构与 direction: ltr，只切换书写模式。观察“第一行”“第二行”，先指出第二行沿哪个块方向出现，再判断 inline-size 对应屏幕宽还是高；改变 direction 后还要重新判断行内起点。

```html
<div class="writing-sample horizontal" dir="ltr"><p>第一行</p><p>第二行</p></div>
<div class="writing-sample vertical-right" dir="ltr"><p>第一行</p><p>第二行</p></div>
<div class="writing-sample vertical-left" dir="ltr"><p>第一行</p><p>第二行</p></div>
```

```css
.writing-sample {
  inline-size: 160px;
  block-size: 120px;
  border: 1px solid teal;
  margin: 12px 0;
}
.writing-sample p { margin: 0; padding: 4px; }
.horizontal { writing-mode: horizontal-tb; }
.vertical-right { writing-mode: vertical-rl; }
.vertical-left { writing-mode: vertical-lr; }
/* 横排第二段在下；vertical-rl 的第二段在左，vertical-lr 则在右。 */
```

## 2 direction、HTML dir 与混合文字

direction 的 ltr、rtl 设置基本方向，不是把字符串中的每一个字符简单倒序。横排 rtl 时行内起始边在右；竖排的起止还受 writing-mode 与字形朝向影响。本章的竖排示例明确使用 ltr。

内容的方向通常应写在 HTML 的 dir 属性中，让语义不依赖 CSS。lang 说明语言，并不自动等于 dir；不确定输入文字方向时可以用 dir="auto"。&lt;bdi&gt; 隔离一段双向文字，减少用户名、数字和标点相互干扰。

下面 Arabic 文字用 dir="rtl" 声明内容方向；css-rtl 段只演示 CSS 的对应属性。:root 等选择器也可控制继承，但不要为了移动盒子而随意更改整页文字方向。text-align: start 表示向本行起始侧对齐。

```html
<p class="direction-sample" dir="ltr">Reading 2026</p>
<p class="direction-sample" lang="ar" dir="rtl">مرحبا بالعالم</p>
<p class="direction-sample css-rtl">ABC 123</p>
<p class="direction-sample" dir="rtl">用户：<bdi dir="auto">Alice-42</bdi>，编号 <bdi>2026</bdi></p>
```

```css
.direction-sample { border: 1px solid teal; padding: 8px; text-align: start; }
.css-rtl {
  direction: rtl;
  /* 第一段靠左；其余三段靠右，ABC 和 123 内部仍按原顺序阅读。 */
}
```

## 3 逻辑尺寸与物理尺寸的映射

inline-size 对应行内轴长度，block-size 对应块轴长度。horizontal-tb 下分别映射 width、height；vertical-rl 和 vertical-lr 下分别映射 height、width。

min-inline-size、max-inline-size 给行内尺寸设置上下限；块方向也有 min-block-size、max-block-size。它们控制的是尺寸映射，盒模型仍受 box-sizing 约束，百分比是否可解析仍取决于相应包含块尺寸。

本例采用 border-box，使测量包含边框。逻辑盒在两种模式下交换物理宽高；固定 width、height 的对照盒不交换。跨模式设计还要检查内容能否装下，不能只交换数字。

```html
<div class="logical-size horizontal">逻辑盒：横排</div>
<div class="logical-size vertical-right" dir="ltr">逻辑盒：竖排</div>
<div class="physical-size vertical-right" dir="ltr">物理盒：竖排</div>
```

```css
.logical-size {
  box-sizing: border-box;
  inline-size: 180px;
  block-size: 100px;
  border: 2px solid teal;
  margin: 12px 0;
}
.physical-size {
  box-sizing: border-box;
  width: 180px;
  height: 100px;
  border: 2px solid gray;
}
/* 横排逻辑盒为 180×100，竖排逻辑盒为 100×180；物理盒仍为 180×100。 */
```

## 4 逻辑边距、内边距与边框

margin-inline 与 padding-inline 控制行内两侧；margin-block 与 padding-block 控制块方向两侧。一个值设置同轴两侧，两个值依次表示 start、end，不是“上右”。

border-inline-start 把提示边框放在文字起始侧：横排 ltr 在左，横排 rtl 在右。同样，margin-inline-start 是起始外边距；这些逻辑长属性按元素自身计算后的书写模式映射。

margin: 8px 16px 仍是物理简写，按上、右、下、左展开，不会因为用了逻辑属性就自动改义。要支持不同方向，明确自己需要随内容变化还是固定在屏幕一侧。

```html
<div class="note-host" dir="ltr"><p class="logical-note">LTR 起始侧提示</p></div>
<div class="note-host" dir="rtl"><p class="logical-note">RTL 起始侧提示</p></div>
```

```css
.note-host { border: 1px solid gray; margin-block: 12px; }
.logical-note {
  margin-block: 8px;
  margin-inline: 20px 8px;
  padding-block: 8px;
  padding-inline: 16px;
  border-inline-start: 6px solid teal;
  text-align: start;
  /* LTR 的 6px 边框和 20px 外边距在左；RTL 时两者都移到右侧。 */
}
```

## 5 竖排字形朝向与阅读边界

writing-mode 决定排版方向，text-orientation 主要决定竖排中的字形朝向。默认 mixed 会让适合竖排的字符直立、部分横排文字侧转；upright 让字符分别直立，并会影响用于双向文字处理的方向。

本例只比较 ltr 下的中文与短拉丁字母。不要把 upright 当作对任意 RTL 内容都无影响的外观开关。字体和字符的竖排字形表现可能不同，检查真实目标字体中的标点与混排。

书写模式改变视觉排布，DOM 中的段落顺序仍应服务于理解。竖排阅读区要能滚动或随内容扩展，不用裁剪隐藏多出的文字。

```html
<div class="orientation-frame">
  <p class="vertical-copy mixed-copy" dir="ltr">中文 CSS ABC，标点。</p>
  <p class="vertical-copy upright-copy" dir="ltr">中文 CSS ABC，标点。</p>
</div>
```

```css
.orientation-frame { max-width: 100%; overflow: auto; }
.vertical-copy {
  writing-mode: vertical-rl;
  inline-size: 220px;
  min-block-size: 80px;
  margin: 12px 0;
  padding: 8px;
  border: 1px solid teal;
}
.mixed-copy { text-orientation: mixed; }
.upright-copy { text-orientation: upright; }
/* 对比 ABC 是侧转还是逐字直立；检查中文标点，而非只看盒子旋转。 */
```

## 6 Grid 和 Flexbox 都需要重新辨认轴

Grid 的列沿行内轴展开、行沿块轴展开；justify 对应行内轴，align 对应块轴。竖排时，Grid 的“列”不一定是屏幕上的竖直分栏。

Flexbox 的主轴由 flex-direction 决定：row 沿行内轴，column 沿块轴；交叉轴与主轴垂直。justify-content 在 Flexbox 中沿主轴，不能把 Grid 的行内轴解释不加条件地移植过来。

下方两组都使用竖排 ltr，Grid 两列与 Flexbox 的 row 都沿上到下展开。这里只改变布局方法；真实应用应保证源码阅读顺序合理。

```html
<div class="vertical-grid" dir="ltr"><span>A</span><span>B</span></div>
<div class="vertical-flex" dir="ltr"><span>A</span><span>B</span></div>
```

```css
.vertical-grid, .vertical-flex {
  writing-mode: vertical-rl;
  inline-size: 180px;
  block-size: 80px;
  gap: 8px;
  border: 1px solid gray;
  margin: 12px 0;
}
.vertical-grid { display: grid; grid-template-columns: repeat(2, 1fr); }
.vertical-flex { display: flex; flex-direction: row; }
.vertical-flex > span { flex: 1; }
.vertical-grid > span, .vertical-flex > span { background-color: #dfefea; padding: 4px; }
/* 两组 B 均在 A 的下方；把 flex-direction 改成 column 后再判断主轴。 */
```

## 7 逻辑偏移跟随方向，定位参照仍需成立

inset-inline-start、inset-inline-end 与 inset-block-start、inset-block-end 分别对应两条逻辑轴的起止偏移；inset-inline、inset-block 是同轴简写。inset 本身仍是 top、right、bottom、left 的物理四边简写。

这些属性不会自行建立定位。绝对定位仍需合适的包含块；本例由 position: relative 的盒子提供参照，标签用 position: absolute。为了观察纯方向差异，标签继承父盒子的 writing-mode 和 direction。

这里 block-start 为 8px、inline-end 为 8px：横排 ltr 靠右上，横排 rtl 靠左上，竖排 vertical-rl 且 ltr 靠右下。

```html
<div class="position-host" dir="ltr"><span class="logical-badge">A</span></div>
<div class="position-host" dir="rtl"><span class="logical-badge">A</span></div>
<div class="position-host vertical-right" dir="ltr"><span class="logical-badge">A</span></div>
```

```css
.position-host {
  position: relative;
  width: 260px;
  max-width: 100%;
  height: 140px;
  border: 1px solid gray;
  margin: 12px 0;
}
.logical-badge {
  position: absolute;
  inset-block-start: 8px;
  inset-inline-end: 8px;
  padding: 4px;
  background-color: #dfefea;
  /* 三个 A 依次靠右上、左上、右下，两条对应边均偏移 8px。
     偏移以包含块内边缘为参照，外框还有 1px 边框。 */
}
```

## 8 物理属性与逻辑属性混用时的层叠

当逻辑属性映射到某个物理属性时，它们共同参与层叠，没有“逻辑属性天然优先”。来源、重要性、层、优先级等相同后，声明顺序才决定同一边的结果。

同一条 mixed-margin 规则中先写 margin-left: 10px，再写 margin-inline-start: 30px。横排 ltr 下两者竞争左边；rtl 下后者映射右边，所以两边可能同时留下外边距。仅换书写方向就可能暴露原来不明显的混用问题。

本节故意混写来观察映射后的层叠；普通方向自适应示例直接选择所需逻辑属性，不额外添加物理属性兜底。不能把在 LTR 下看起来一样的声明视为所有模式下等价。

```html
<div class="mix-host" dir="ltr"><p class="mixed-margin">LTR：逻辑声明在后</p></div>
<div class="mix-host" dir="rtl"><p class="mixed-margin">RTL：两边分别有值</p></div>
<div class="mix-host" dir="ltr"><p class="physical-last">LTR：物理声明在后</p></div>
```

```css
.mix-host { border: 1px solid gray; margin: 12px 0; }
.mixed-margin, .physical-last { padding: 8px; border: 1px solid teal; }
.mixed-margin {
  margin: 0;
  margin-left: 10px;
  margin-inline-start: 30px;
}
.physical-last {
  margin: 0;
  margin-inline-start: 30px;
  margin-left: 10px;
}
/* Computed 检查：前两组分别为左30/右0、左10/右30；最后一组左10。 */
```

## 本章小结

- inline 沿一行文字前进，block 沿块与行的排列方向；物理映射必须结合 writing-mode 和 direction。
- 内容方向优先使用 HTML dir；lang 不自动指定方向，混排可用 bdi 隔离。
- 逻辑尺寸、间距和偏移跟随书写模式；物理属性仍固定在物理边。
- Grid 的轴与 Flexbox 的主交叉轴需分别辨认，不能统一叫水平和垂直。
- 逻辑属性和物理属性映射到同一边时共同参与层叠，混用需跨方向检查。
