# 列表、表格与表单样式

学习目标：能调整列表、表格和表单的外观，处理长内容，并保留语义、键盘操作与清晰焦点。

## 2 列表标记与 ::marker

&lt;ul&gt;、&lt;ol&gt; 表达列表结构，&lt;li&gt; 是列表项；这里调整它们的外观。list-style-type 可用 disc、circle、square 等形状，或 decimal、lower-alpha 等编号形式。

list-style-position: outside 把标记放在主要内容盒外，inside 则让标记成为首行行内内容的一部分。外侧标记需要足够留白，祖先裁剪也可能切掉它；不要为了对齐直接清掉全部内边距。

list-style-image 指定图像，图像不可用时回到 list-style-type。list-style 是上述分项的简写，会设置省略项；若重要序号属于内容含义，优先保留原生有序列表。

::marker 是标记伪元素，不是属性；它只支持有限属性，例如 color、字体属性和 content，不可当普通盒子任意设置 padding、背景或尺寸。只修改颜色和字体时保留原标记作为回退。

```html
<ul class="outside"><li>这是一条较长的学习任务，用来观察换行后的文字起点。</li><li>复习笔记</li></ul>
<ul class="inside"><li>这是一条较长的学习任务，用来观察换行后的文字起点。</li><li>复习笔记</li></ul>
<ul class="image-list"><li>图像标记与类型回退</li></ul>
```

```css
.outside, .inside { width: 240px; padding-left: 32px; list-style-type: square; }
.outside { list-style-position: outside; }
.inside { list-style-position: inside; }
.image-list { list-style-image: url("marker.svg"); list-style-type: square; }
.outside li::marker { color: #006666; font-size: 1.2em; }
/* outside 的续行与正文起点对齐；inside 的标记进入首行内容。
   图像不可用时使用 square；::marker 不支持时仍保留普通列表标记。 */
```

## 3 计数器与生成内容

counter-reset 在元素上创建命名计数器，省略起始数值时为 0；counter-increment 默认加 1，本例先递增再显示。topic 是自定的计数器名称。

counter() 输出当前计数器值；嵌套编号可用 counters() 把同名嵌套计数器按指定分隔符组合。counter-set 改变当前计数器值，没有可用计数器时会创建；它和在嵌套范围用 counter-reset 新建计数器不是同一件事。

content 在 ::before、::after 等伪元素里生成文本。这里 ::before 出现在元素内容之前；元素需实际生成盒子，display: none 的元素不会像可见条目一样更新计数器。

生成内容的辅助技术呈现可能不同，不能用它作为唯一的错误消息、按钮名称或必要业务内容。本例真实标题完整表达主题；如果要表达有顺序的任务，使用 &lt;ol&gt;，不靠伪元素替代结构。

```html
<div class="topics">
  <h3>选择器复习</h3>
  <p>检查匹配范围。</p>
  <h3>布局练习</h3>
  <p>检查窄屏布局。</p>
</div>
<div class="topics">
  <h3>新一组专题</h3>
</div>
```

```css
.topics { counter-reset: topic; }
.topics h3 { counter-increment: topic; }
.topics h3::before { content: "专题 " counter(topic) " · "; color: #006666; }
/* 每个 topics 容器重建 topic 计数器，第二组从 1 开始。
   删除生成内容后，真实标题仍能表达主题；编号不承担唯一的任务含义。 */
```

## 4 分离边框与合并边框

border-collapse: separate 为每个单元格保留独立边框，border-spacing 的两个值分别是水平、垂直间距；单值作用于两个方向。单元格内容与边框之间的空间由 padding 控制。

collapse 合并相邻边框。相邻声明冲突会按表格边框冲突规则选择，不是两条线简单相加；通常更宽边框优先，hidden 还有特殊的抑制规则。此时 border-spacing 不生效。

保留 &lt;caption&gt;、&lt;th&gt; 与 scope，读者和辅助技术仍可理解标题及表头关系；CSS 不负责补出缺失的表格语义。

```html
<table class="separate">
  <caption>分离边框</caption>
  <thead><tr><th scope="col">主题</th><th scope="col">安排</th></tr></thead>
  <tbody><tr><th scope="row">CSS</th><td>周二</td></tr><tr><th scope="row">HTML</th><td>周四</td></tr></tbody>
</table>
<table class="collapsed">
  <caption>合并边框</caption>
  <thead><tr><th scope="col">主题</th><th scope="col">安排</th></tr></thead>
  <tbody><tr><th scope="row">CSS</th><td>周二</td></tr><tr><th scope="row">HTML</th><td>周四</td></tr></tbody>
</table>
```

```css
.separate, .collapsed { width: 320px; max-width: 100%; border: 2px solid #333; }
.separate th, .separate td, .collapsed th, .collapsed td { border: 1px solid #777; padding: 8px; text-align: left; }
.separate { border-collapse: separate; border-spacing: 8px 4px; }
.collapsed { border-collapse: collapse; border-spacing: 8px 4px; }
/* separate 的列间 8px、行间 4px；collapse 中 border-spacing 不起作用。
   caption、表头和行头保留真实表格语义，不用 div 伪造表格。 */
```

## 5 自动布局、固定布局与长内容

table-layout: auto 会考虑单元格内容，长的不可断字符串可能撑宽表格。table-layout: fixed 在给定表格宽度时，主要按列宽、首行单元格宽度和剩余空间确定列宽；后续行内容不会重新决定这些列宽。

固定算法先使用明确的列宽，没有列宽时再看首行相应单元格，余下列分配剩余空间。本例 &lt;col&gt; 为第一列指定 100px，表格给定 360px；只写 fixed 而表宽仍不确定，不应承诺相同布局。

固定列不等于内容会自动适配：仍需折行、裁剪或滚动策略。本例在固定表中让长串折行，在自动表外提供有名称、可聚焦的滚动区域。缩窄视口时保留完整表格关系，不把表格当页面布局工具。

```html
<div class="table-region" tabindex="0" role="region" aria-label="自动布局表格，可横向滚动">
<table class="auto-table">
  <caption>自动布局</caption>
  <colgroup><col class="name-column"><col></colgroup>
  <thead><tr><th scope="col">项目</th><th scope="col">记录</th></tr></thead>
  <tbody><tr><th scope="row">练习</th><td>engineeringnotebookexamplewithoutspaces</td></tr></tbody>
</table>
</div>
<table class="fixed-table">
  <caption>固定布局</caption>
  <colgroup><col class="name-column"><col></colgroup>
  <thead><tr><th scope="col">项目</th><th scope="col">记录</th></tr></thead>
  <tbody><tr><th scope="row">练习</th><td>engineeringnotebookexamplewithoutspaces</td></tr></tbody>
</table>
```

```css
.table-region { max-width: 100%; overflow-x: auto; }
.auto-table, .fixed-table { width: 360px; border-collapse: collapse; }
.auto-table { table-layout: auto; }
.fixed-table { table-layout: fixed; max-width: 100%; }
.name-column { width: 100px; }
.auto-table th, .auto-table td, .fixed-table th, .fixed-table td { border: 1px solid #777; padding: 8px; text-align: left; }
.fixed-table td { overflow-wrap: anywhere; }
/* auto 会参考长内容；fixed 先按表宽和列宽安排，长串在单元格内换行。
   窄视口时检查自动表的区域可滚动，不能把整个页面撑宽。 */
```

## 6 控件尺寸与有效、禁用、只读状态

浏览器默认控件的字体和盒模型不完全相同。font: inherit 让控件跟随页面字体；box-sizing: border-box 让设置的宽度包含边框和内边距。只给文字输入类控件统一宽度，避免把复选框拉成整行。

下面的 :required、:invalid、:valid、:disabled 都是状态伪类，不是属性值。required、readonly、disabled 则是 HTML 属性：

- required 参与原生约束验证，空的必填输入也可能在刚打开页面时匹配 :invalid；实际产品可另评估 :user-invalid 的支持与交互时机。
- disabled 禁止操作，通常不进入 Tab 顺序，且不提交该控件值；只画成灰色不能替代它。
- readonly 适用于支持它的文本类控件，仍可聚焦与选择，也不同于 disabled；不能把它用于不支持的控件来假装只读。

这里的 required 与状态选择器专门用于讲解浏览器约束验证状态，不是给其他示例添加输入防护。placeholder 不能替代可见标签，状态也不能只用红绿颜色表达。

```html
<form class="sized-form" action="forms.html" method="get">
  <p><label for="email">电子邮箱（必填）</label><br><input id="email" name="email" type="email" required aria-describedby="email-help"></p>
  <p id="email-help">请输入完整邮箱，例如 reader@example.com；虚线边框表示当前值不符合要求。</p>
  <p><label for="topic">主题</label><br><select id="topic" name="topic"><option>CSS</option><option>HTML</option></select></p>
  <p><label for="notes">备注</label><br><textarea id="notes" name="notes" rows="3"></textarea></p>
  <p><label for="reference">只读编号</label><br><input id="reference" value="NOTE-09" readonly></p>
  <button type="reset">重置</button>
  <button type="button" disabled>暂不可用</button>
</form>
```

```css
input, select, textarea, button { font: inherit; box-sizing: border-box; }
.sized-form input, .sized-form select, .sized-form textarea { width: 100%; max-width: 28rem; padding: 8px; border: 2px solid #777; }
textarea { resize: vertical; }
input:required { border-left-width: 5px; }
input:invalid { border-style: dashed; border-color: #9b2600; }
input:valid { border-style: solid; }
input[readonly] { background-color: #eee; }
button:disabled { color: #555; background-color: #eee; border: 1px solid #999; }
/* 初始必填空值也匹配 :invalid；输入合法邮箱后边框从虚线变为实线。
   只读输入仍可聚焦和选择；disabled 按钮不参与正常 Tab 导航。 */
```

## 7 accent-color 与 appearance 的取舍

accent-color 只调整支持控件和状态的强调色，例如复选框、单选按钮、范围输入和进度条的部分区域；不保证重绘所有控件。浏览器不支持时继续使用原生配色。

appearance: auto 使用平台外观；none 抑制部分原生绘制，不会把元素变成失去行为的普通盒子。复选框取消原生外观后可能没有可辨认图形，因此必须补齐尺寸、边框、选中标记、焦点和禁用状态；能用强调色满足需求时通常更简单。

本例保留真实 &lt;input type="checkbox"&gt; 和 &lt;label&gt;，使用图像勾号区分选中状态。单选按钮共享 name 后仍按原生分组工作。控件内部样式因平台而异，不保证 &lt;option&gt;、日期选择器等都能统一装饰。

@media (forced-colors: active) 是针对用户强制颜色模式的小幅调整：本例回到原生复选框，避免自绘标记与系统配色冲突。回退选择器同时覆盖 :checked，确保选中状态的背景图也被清除。媒体查询的一般写法在响应式主题展开。

```html
<p><label><input class="native-check" type="checkbox">接收更新（原生外观）</label></p>
<p><label><input class="custom-check" type="checkbox">接收更新（自定义外观）</label></p>
<fieldset>
  <legend>阅读方式</legend>
  <label><input class="native-check" type="radio" name="mode" value="online" checked>在线</label>
  <label><input class="native-check" type="radio" name="mode" value="offline">离线</label>
</fieldset>
```

```css
.native-check { accent-color: #006666; }
.custom-check {
  appearance: none;
  width: 24px;
  height: 24px;
  vertical-align: middle;
  border: 2px solid #006666;
  background-color: white;
}
.custom-check:checked { background-color: #006666; background-image: url("check.svg"); background-size: contain; }
@media (forced-colors: active) {
  .custom-check, .custom-check:checked { appearance: auto; background-image: none; }
}
/* 点击文字标签或按空格切换；自定义样式仍作用在真实 checkbox 上。
   强制颜色模式回到原生外观；不能仅用背景颜色表示选中。 */
```

## 8 链接、悬停与键盘焦点

链接保留下划线，使它不只靠颜色可辨认。:link 匹配未访问链接，:visited 匹配已访问链接；等优先级规则通常按 link、visited、hover、active 的顺序写，避免后面的普通状态盖掉交互反馈。

:hover 是悬停，:active 是激活期间，:focus 表示元素当前有焦点；:focus-visible 还考虑浏览器是否判断应显示焦点指示，常见于键盘导航，不能简单等同于“只在键盘下匹配”。

outline 不占布局空间。不要全局移除焦点轮廓；下面先为 :focus 提供基础指示，再增强 :focus-visible，旧浏览器也有回退。悬停效果不能承担触屏和键盘用户唯一可获得的信息。

为保护浏览历史，:visited 允许的样式与脚本可读信息受到限制；getComputedStyle 不提供可靠的访问历史结果，不用它判断用户访问过哪里。

```html
<p><a class="example-link" href="#destination">阅读本页说明</a></p>
<button class="action" type="button">可聚焦按钮</button>
<p id="destination">链接目标：样式保留了可辨认的下划线。</p>
```

```css
.example-link:link { color: #005b70; }
.example-link:visited { color: #67408a; }
.example-link { text-decoration: underline; text-underline-offset: 0.2em; }
.example-link:hover { text-decoration-thickness: 3px; }
.example-link:active { color: #9b2600; }
a:focus, button:focus, input:focus, select:focus, textarea:focus, .table-region:focus {
  outline: 3px solid #006666;
  outline-offset: 3px;
}
a:focus-visible, button:focus-visible, input:focus-visible, select:focus-visible, textarea:focus-visible, .table-region:focus-visible { outline-width: 4px; }
.action:hover { background-color: #e0eeee; }
/* Tab 导航时检查轮廓；保留 :focus 作为基础，不支持 :focus-visible 也可见。
   hover 是悬停状态，不假定触屏有鼠标；访问历史颜色不作脚本读取断言。 */
```

## 本章小结

- 列表标记外观与列表语义分开；::marker 的可用属性有限。
- 计数器负责 CSS 编号，生成内容不承担唯一的关键信息。
- 表格边框模型与列宽算法是两件事；固定列还要处理长内容。
- 控件外观可调整，标签、原生状态与操作能力必须保留。
- 状态和焦点要看实际交互，hover 与颜色不替代键盘可访问性。

（3）只用 Tab、空格和方向键访问表单：切换复选框、切换单选项、进入只读输入、跳过禁用按钮。检查每一步是否都有可见焦点。

### 提示

第二题分别检查表格外框、单元格与滚动区域；第三题不能只看截图，原生控件的操作必须实际完成。

### 重点题解析

第（2）题只改变后续数据行的文本，保持 fixed、表格宽度和第一列 100px 不变；首行及列定义仍决定列宽，长串通过 anywhere 在第二列内折行。360px 视口下，固定表受 max-width 限制，自动表的超宽内容由自身滚动区域承载；比较文档 scrollWidth 与 clientWidth，不能只看到滚动条就断言整页没有溢出。

第（3）题中只读编号可以聚焦，但不能编辑；Tab 从重置按钮继续时跳过禁用按钮。复选框在焦点内按空格切换，单选组用方向键切换；自定义外观仍沿用原生 input 的状态。
