## 1 从两列认识网格线、轨道与区域

Grid 同时组织行和列。display: grid 建立网格容器；正常流中的直接子元素成为网格项目，孙元素不会自动加入外层网格。grid 是 display 的值，不是选择器。

读布局先数边界线，再数线之间的轨道（track）。轨道可以是行或列，行列相交是单元格（cell）；一个或多个相邻单元格组成的矩形是网格区域（area）。因此“两列”对应三条列线。

![W3C Grid 原图：两列对应三条列线，三行对应四条行线。](image/illustration/12-01-grid-lines-tracks.png)

图 1：引用 W3C CSS Grid 1 原图，保留 477×308 像素与原内容。顶部 line 1～3 是列线，右侧 line 1～4 是行线；图中有两列三行。原图的 150px、50px 与 1fr 用于说明轨道，本例使用的尺寸与项目数量不同，不能混用。Copyright © 2025 World Wide Web Consortium；许可及来源见篇末。

下面示例横排且从左到右：两列仍有三条列线，正数从起始边的 1 开始；负数从显式网格末端倒数，-1 是最后一条显式线。grid-column: 1 / -1 占满显式列，不自动表示所有后来生成的隐式列。斜杠左、右分别是起止线；span 2 表示跨两条轨道。

打开 Grid 覆盖层后先找 1 与 -1 的位置，再把 A 改为只占第一列，观察“跨两条轨道”和“跨两条线”为什么不是同一说法。本例轨道实际宽度由容器、120px、1fr 与 gap 共同决定。

```html
<div class="line-grid">
  <div class="wide-item">A：跨两列</div>
  <div>B：第一列</div>
  <div>C：第二列</div>
</div>
```

```css
/* index：先观察轨道和跨列项目，再比较比例分配与长内容。 */
.line-grid {
  display: grid;
  grid-template-columns: 120px minmax(0, 1fr);
  gap: 12px;
}
.line-grid > div { border: 1px solid teal; padding: 8px; }
.wide-item {
  grid-column: 1 / -1;
  /* Grid 覆盖层中检查 A 的左右边分别落在列线 1 和 3。 */
}
```

## 2 fr、repeat() 与 minmax() 控制尺寸

fr 是轨道分配剩余空间的弹性单位。先考虑固定尺寸、间距和轨道的最小要求，再分配可用空间；它不是容器宽度的百分比。这里所有弹性系数之和大于等于 1，1fr 与 2fr 在最小值不限制时按 1:2 分配。

repeat(3, 1fr) 表示重复三次 1fr；repeat() 不能嵌套。minmax(0, 1fr) 把最小轨道尺寸设为 0，最大端参与弹性分配；最小端不能写 fr。minmax(80px, auto) 则允许行在至少 80px 的基础上随内容增高。

单独的 1fr 隐含自动最小值；长的不可断字符串可能撑宽轨道。把轨道最小值设为 0，并按内容选择换行或局部滚动。min-width: 0 只允许盒子缩小，不会自动让文字断行。

```html
<div class="fraction-grid">
  <div>固定 100px</div>
  <div>剩余空间的一份</div>
  <div>剩余空间的两份</div>
</div>
<div class="safe-grid">
  <div>ABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZ</div>
  <div>另一列仍可阅读</div>
</div>
```

```css
.fraction-grid {
  display: grid;
  grid-template-columns: 100px minmax(0, 1fr) minmax(0, 2fr);
  gap: 10px;
}
.fraction-grid > div { border: 1px solid teal; overflow-wrap: anywhere; }
.safe-grid {
  display: grid;
  grid-template-columns: repeat(2, minmax(0, 1fr));
  gap: 12px;
  margin-top: 16px;
}
.safe-grid > div {
  min-width: 0;
  overflow-wrap: anywhere;
  border: 1px solid teal;
  /* 缩窄页面：长串换行；不要用隐藏整个页面的溢出来遮住问题。 */
}
```

## 3 显式、隐式网格与自动放置

grid-template-columns、grid-template-rows 与 grid-template-areas 定义显式网格。项目多于现有格子，或指定位置超出边界时，浏览器会建立隐式轨道；其尺寸用 grid-auto-rows、grid-auto-columns 控制。

自动放置先避开已经占用的位置，再处理尚未确定位置的项目。默认 grid-auto-flow: row 逐行寻找空间，需要时增加行；column 逐列放置，需要时增加列，通常还需明确行数。自动放置使用考虑 order 后的文档顺序，本章不设置 order。

dense 会回头把后来的小项目填进空洞，可能改变视觉顺序，却不会改写 DOM 阅读顺序和常规 Tab 顺序。叙事内容和交互步骤不要靠填洞重新排序。下面两组只改变 dense，均保留 A、B、C 的源码顺序。

```html
<div class="packing sparse">
  <a class="span-two" href="#end">A：跨两列</a>
  <a class="span-two" href="#end">B：跨两列</a>
  <a class="small" href="#end">C</a>
</div>
<div class="packing dense">
  <a class="span-two" href="#end">A：跨两列</a>
  <a class="span-two" href="#end">B：跨两列</a>
  <a class="small" href="#end">C</a>
</div>
<div class="column-flow"><span>1</span><span>2</span><span>3</span></div>
<p id="end">链接目标：用于检查键盘顺序。</p>
```

```css
/* placement：固定列数，比较普通自动放置、dense 填空和按列放置。 */
.packing {
  display: grid;
  grid-template-columns: repeat(3, minmax(0, 1fr));
  grid-template-rows: 48px;
  grid-auto-rows: minmax(48px, auto);
  gap: 8px;
  margin-bottom: 16px;
}
.packing > a { border: 1px solid teal; padding: 4px; }
.span-two { grid-column: span 2; }
.dense { grid-auto-flow: row dense; }
/* C 在 dense 组填入首行空位；第二行仍是隐式行。Tab 检查源码顺序。 */
.column-flow {
  display: grid;
  grid-template-rows: repeat(2, 40px);
  grid-auto-flow: column;
  grid-auto-columns: 100px;
  gap: 8px;
}
.column-flow > span { border: 1px solid teal; }
/* 1、2 在首列，3 在第二列；这里的列由隐式网格生成。 */
```

## 4 auto-fill 与 auto-fit 的空轨道差异

auto-fill、auto-fit 是 repeat() 的重复次数关键字。浏览器根据容器可用尺寸、轨道的确定尺寸部分与 gap 计算能放几列。不能写 repeat(auto-fit, 1fr)，本例用 minmax() 提供可计算的最小尺寸。

auto-fill 保留可容纳的空轨道；auto-fit 在项目放置后折叠没有项目占用或跨越的轨道及相邻间距，让其余弹性轨道获得更多空间。项目把所有轨道占满时，两者往往没有可见差别。

min(140px, 100%) 让最小轨道尺寸在容器不足 140px 时缩到容器宽度；100% 的参照是此网格容器。测试空轨道差异需让容器至少能放三列，并只保留两个项目。

```html
<div class="repeat-grid fill"><div>一</div><div>二</div></div>
<div class="repeat-grid fit"><div>一</div><div>二</div></div>
```

```css
/* 两组使用相同内容与最小列宽，仅切换 auto-fill / auto-fit。 */
.repeat-grid {
  display: grid;
  gap: 12px;
  max-width: 700px;
  margin-bottom: 16px;
}
.fill { grid-template-columns: repeat(auto-fill, minmax(min(140px, 100%), 1fr)); }
.fit { grid-template-columns: repeat(auto-fit, minmax(min(140px, 100%), 1fr)); }
.repeat-grid > div {
  border: 1px solid teal;
  padding: 8px;
  /* 宽容器里比较两组项目宽度；Grid 覆盖层显示 fill 保留的空列。 */
}
```

## 5 命名区域与 Grid、Flexbox 配合

grid-template-areas 的每个字符串代表一行，各行格子数必须相同。同名区域必须组成完整矩形；点号表示空格子。header、menu、body 是本例区域名称，既不是选择器，也不自动等于 HTML 标签。

grid-area: menu 将项目放入命名区域。另一种写法是用 [sidebar-start] 这样的方括号给网格线命名，再把该名称写入 grid-column 的起止位置；区域命名和线命名服务于不同放置方式。

页面需要行列同时对齐时可用 Grid；同一卡片内的按钮只需沿一条轴排列并换行，可用 Flexbox。嵌套容器各自管理直接子元素：页面网格不会把工具条的按钮当作外层项目。

```html
<div class="page-grid">
  <header class="page-heading">阅读计划</header>
  <nav class="page-menu" aria-label="阅读分类">基础 · 进阶</nav>
  <article class="page-body">
    <h3>布局笔记</h3>
    <div class="actions"><button type="button">收藏</button><button type="button">标记已读</button></div>
  </article>
</div>
```

```css
.page-grid {
  display: grid;
  grid-template-columns: [sidebar-start] minmax(0, 1fr) [body-start] minmax(0, 2fr) [page-end];
  grid-template-areas: "header header" "menu body";
  gap: 12px;
}
.page-heading { grid-area: header; }
.page-menu { grid-area: menu; margin: 0; }
.page-body { grid-area: body; min-width: 0; }
.page-grid > * { border: 1px solid teal; padding: 8px; }
.actions {
  display: flex;
  flex-wrap: wrap;
  gap: 8px;
  /* 外层检查区域边界，内层检查按钮换行；Tab 顺序沿 HTML 前进。 */
}
```

## 6 项目对齐与轨道整体对齐

本章横排时，justify 方向是左右的行内轴，align 方向是上下的块轴；竖排时不能继续把它们固定记成水平和垂直。

justify-items、align-items 为项目在各自区域内设置对齐；justify-self、align-self 只改变一个项目。普通自动尺寸项目的默认对齐通常表现为拉伸，有固有宽高比的项目等存在例外。

justify-content、align-content 则移动或分布整个轨道组，必须有可分配的剩余空间才看得到整体移动。本例固定轨道宽高并给容器高度；如果所有宽度都交给 fr 填满，justify-content 的空余分配就可能看不出变化。

```html
<div class="alignment-grid">
  <div>A</div><div class="own-alignment">B</div>
</div>
```

```css
/* alignment：先设置整组轨道的位置，再设置项目在各自格子内的位置。 */
.alignment-grid {
  display: grid;
  grid-template-columns: repeat(2, 90px);
  grid-template-rows: 80px;
  gap: 12px;
  height: 200px;
  border: 1px solid gray;
  justify-content: center;
  align-content: center;
  justify-items: start;
  align-items: start;
}
.alignment-grid > div { padding: 8px; border: 1px solid teal; }
.own-alignment {
  justify-self: end;
  align-self: end;
  /* A 靠自身格子的起始角，B 靠结束角；整个 192px × 80px 轨道组居中。 */
}
```

## 7 在同一格内重叠

明确把两个项目放在相同网格区域时，它们可以重叠；这与自动放置避开已占用格子是两种情况。网格项目可在不设置 position 的情况下使用非 auto 的 z-index，且会建立自己的层叠上下文。

不指定 z-index 时按相关绘制顺序堆叠，后面的项目通常盖住前面的项目。重叠只改变呈现，不能据此删除有意义的替代文字或遮住可操作控件。

```html
<div class="overlap-grid">
  <div class="back-layer">背景区域</div>
  <p class="front-layer">同一格内的说明</p>
</div>
```

```css
/* 把两项放到同一格，观察项目重叠和 z-index。 */
.overlap-grid { display: grid; min-height: 120px; }
.overlap-grid > * { grid-column: 1; grid-row: 1; }
.back-layer { background: #dceeea; padding: 8px; }
.front-layer {
  z-index: 1;
  justify-self: center;
  align-self: center;
  margin: 0;
  padding: 8px;
  background: white;
  /* 说明盖在浅色背景上，仍参与网格尺寸计算。 */
}
```

## 8 subgrid 共享父网格轨道

普通嵌套网格独立计算自己的轨道。subgrid 是 grid-template-columns 或 grid-template-rows 的值，让该方向复用它所跨越的父网格轨道；子网格自身必须既是父网格项目，又建立 Grid。

本例每张卡片跨三条父行，用 subgrid 让两张卡片的标题、摘要、按钮行对齐。子内容仍可参与父轨道的内容尺寸计算；子网格线从自己的 1 重新编号，gap 默认衔接父网格，也可以显式修改。

设为 subgrid 的方向不会再扩展隐式轨道。不要在只有三条共享行的卡片里无限追加自动放置项目；需要任意行数时保留该方向为普通嵌套网格。

两组使用相同内容、列数和间距，只比较共享父行与独立计算行的差别。子网格组直接使用 subgrid，运行条件已在开篇说明；独立网格组保留自己的 auto 1fr auto 行定义。

```html
<div class="cards use-subgrid">
  <article class="course-card"><h3>短标题</h3><p>短摘要。</p><button type="button">阅读</button></article>
  <article class="course-card"><h3>较长标题<br>占用两行</h3><p>这张卡片也有摘要。</p><button type="button">阅读</button></article>
</div>
<h3>独立网格对照</h3>
<div class="cards independent-cards">
  <article class="course-card"><h3>短标题</h3><p>短摘要。</p><button type="button">阅读</button></article>
  <article class="course-card"><h3>较长标题<br>占用两行</h3><p>这张卡片也有摘要。</p><button type="button">阅读</button></article>
</div>
```

```css
.cards { display: grid; grid-template-columns: repeat(2, minmax(0, 1fr)); gap: 12px; }
.course-card {
  display: grid;
  grid-template-rows: auto 1fr auto;
  gap: 12px;
  min-width: 0;
  border: 1px solid teal;
  padding: 8px;
}
.course-card h3, .course-card p { margin: 0; }
.course-card button { justify-self: start; }
.use-subgrid > .course-card {
  grid-row: span 3;
  grid-template-rows: subgrid;
  /* 比较两卡摘要的顶部坐标；独立网格对照组的行仍各自计算。 */
}
```

## 本章小结

- 网格线界定轨道，项目占据矩形区域；显式网格之外还能生成隐式轨道。
- fr 分配可用空间，minmax() 表达最小约束；长内容仍需合适的换行或滚动策略。
- 自动放置、dense 与显式重叠各有条件，视觉顺序不能替代内容顺序。
- auto-fit 与 auto-fill 的差异在空轨道；项目对齐与整体轨道对齐不是同一对象。
- 页面 Grid 可以容纳 Flexbox 工具条，subgrid 用来共享父轨道，并与独立网格比较。

### 重点题解析

第（2）题在同宽容器中给两组各加入相同项目。所有自动重复列都有项目占用时，auto-fit 没有空轨道可折叠，因此与 auto-fill 的列宽一致；再次缩窄时按新的可用宽度重新计算列数。

第（3）题保持每卡只有标题、摘要、按钮三个直接子项目。子网格的三行由父网格共同定高，某卡摘要增高会撑高共享摘要行，所以按钮仍对齐；独立网格分别计算标题和摘要行，内容仍完整，但不能保证对应行起点一致。
