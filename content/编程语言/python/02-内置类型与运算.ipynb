## 1 名称、对象与类型

Python 用对象表示数据，每个对象都有类型。type(值) 可以查看类型，例如 int 表示整数，str 表示字符串。

赋值让名称指向一个对象，称为名称绑定。再次赋值会重新绑定这个名称，其他名称仍保留原来的绑定。

名称不需要预先声明固定类型，也可以改绑到另一种类型的对象，这体现了 Python 的动态类型。下面变化的是 count 所指向的对象，并不是把原整数对象改成了字符串。

```python
count = 3
previous_count = count
print(count, type(count))  # 预期：3 <class 'int'>

count = 5
print(count, previous_count)  # 预期：5 3；重新绑定 count 不影响 previous_count。

count = "五"
print(count, type(count))  # 预期：五 <class 'str'>
```

## 2 常见内置类型

| 类型 | 中文名称／含义 | 常见用途 | 可变性 |

| float | 浮点数 | 近似数值计算 | 不可变 |
| complex | 复数 | 包含实部和虚部的数值 | 不可变 |
| bool | 布尔值 | 表示真或假 | 不可变 |
| NoneType | None 的类型 | 表示未提供值或没有结果 | 不可变 |
| str | 字符串 | 文本 | 不可变 |
| list | 列表 | 存放多项有序数据 | 可变 |
| tuple | 元组 | 按顺序组合多项数据 | 不可变 |
| range | 整数序列 | 表示指定范围和步长的整数序列 | 不可变 |
| dict | 字典 | 保存键与值的对应关系 | 可变 |
| set | 集合 | 保存不重复的元素 | 可变 |
| frozenset | 不可变集合 | 保存不重复的元素 | 不可变 |

字符串在第 3 章展开，容器与对象引用在第 4 章展开。tuple 自身不可变，但它引用的可变对象仍可能变化，届时会用例子区分。

这不是全部类型清单：还有缓冲区视图 memoryview，以及函数、类、模块等对象的类型；二进制数据与 memoryview 在第 12 章介绍。

## 3 数值、布尔值与 None

### 3.1 整数与字面值

int 表示整数，其大小没有固定的 32 位或 64 位上限，实际仍受可用内存限制。

直接在代码中写出的 42、1.5、"Python" 等表示法称为字面值（literal）。整数默认用十进制书写，也可用 0b、0o、0x 前缀分别表示二进制、八进制、十六进制；数字间的下划线用于分组，不改变数值。

非零十进制整数字面值不能加前导零，例如 012 不合法。负数写作 -12，其中负号是运算符。

```python
# 四种进制表示同一个整数；预期：42 42 42 42
print(42, 0b101010, 0o52, 0x2A)

print(1_000_000)  # 预期：1000000；下划线不会出现在数值输出中。
print(123_456_789_012_345_678_901_234_567_890)  # 观察较大的整数。
```

### 3.2 浮点数与精度

float 用有限精度表示数值。带小数点或 e/E 指数的数字写法会生成 float，例如 1.5e3 表示 1.5 乘以 10 的 3 次方。

多数十进制小数不能精确地表示为有限位二进制小数，计算时可能出现细小的舍入误差。需要判断近似相等时，可以用 math.isclose，并按问题选择允许误差，即容差。

下面的 import math 引入标准库数学工具，math.isclose 通过点号调用其中的函数。模块在第 8 章介绍，数值精度在第 20 章继续讨论。

```python
import math

print(1.5e3, type(1.5e3))  # 预期：1500.0 <class 'float'>
print(0.1 + 0.2)  # 预期：0.30000000000000004；观察舍入误差。
print(math.isclose(0.1 + 0.2, 0.3))  # 预期：True，表示在默认容差内接近。
```

### 3.3 复数

complex 包含实部与虚部，Python 用 j 或 J 表示虚数单位。4j 是虚数字面值，3 + 4j 是相加得到复数的表达式；complex(3, 4) 也能构造这个复数。

用 .real 和 .imag 读取实部与虚部；复数在复平面上到原点的距离称为模，可以用 abs 得到。

复数支持加减乘除，但不支持大小排序、整除或取余。

```python
signal = 3 + 4j
print(signal, type(signal))  # 预期：(3+4j) <class 'complex'>
print(signal.real, signal.imag)  # 预期：3.0 4.0
print(abs(signal))  # 预期：5.0
print(complex(3, 4))  # 预期：(3+4j)
```

### 3.4 布尔值

bool 只有 True 和 False 两个值，用于表示真与假。比较运算可以得到布尔值，例如 >= 判断是否大于或等于，== 判断是否相等；= 则用于赋值。

bool 是 int 的子类型。isinstance(值, 类型) 检查时会接受该类型及其子类型，因此它与直接查看 type 的用途不同。比较与逻辑运算会在第 5 章系统介绍。

```python
is_passing = 75 >= 60
print(is_passing, type(is_passing))  # 预期：True <class 'bool'>
print(3 == 4)  # 预期：False

# 两项均为 True；bool 同时属于 int 的子类型。
print(isinstance(is_passing, bool), isinstance(is_passing, int))
```

### 3.5 None

None 是表示“没有值”的单独对象，它的类型是 NoneType。可以用它表达尚未提供或尚未得到结果；数值 0 和空字符串则是已经有具体含义的值。

检查是否为 None，使用 is None 或 is not None。is 检查是否为同一对象，== 检查是否相等，两者的完整区别在第 4 章展开。

```python
answer = None
print(type(answer))  # 预期：<class 'NoneType'>
print(answer is None)  # 预期：True

answer = 0
print(answer is None)  # 预期：False；0 与“没有值”不同。
```

## 4 数值运算

### 4.1 算术运算符

| 运算符 | 用途 |
| --- | --- |
| +、- | 加、减；放在单个数前表示正号、负号 |
| \* | 乘法 |
| / | 除法 |
| // | 向下取整除法 |
| % | 取余 |
| \*\* | 乘方 |

整数使用 / 相除会得到 float，即使结果没有小数部分。在支持的混合算术运算中，操作数按 int → float → complex 的方向转换。

除法、整除和取余的除数不能为零，否则会触发 ZeroDivisionError。

```python
print(7 + 2, 7 - 2, 7 * 2)  # 预期：9 5 14
print(-7, +7)  # 预期：-7 7
print(8 / 2, type(8 / 2))  # 预期：4.0 <class 'float'>
print(2 ** 3, 2 ** -2)  # 预期：8 0.25；整数的负整数次幂产生 float。
print(2 + 0.5, type(2 + 0.5))  # 预期：2.5 <class 'float'>
print(2 + 0.5j)  # 预期：(2+0.5j)
```

### 4.2 整除与取余

// 向负无穷方向取整，因此对负数不能简单理解为“去掉小数部分”。参与整除的数包含 float 时，结果仍可为 float。

用整数 x 表示被除数、非零整数 y 表示除数，商与余数满足数学关系 x = (x // y) × y + x % y。非零余数与除数同号；divmod(x, y) 同时返回商与余数。

```python
print(7 // 3, 7 % 3)  # 预期：2 1
print(-7 // 3, -7 % 3)  # 预期：-3 2；-3 × 3 + 2 = -7
print(7 // -3, 7 % -3)  # 预期：-3 -2；-3 × (-3) - 2 = 7
print(7.0 // 3)  # 预期：2.0
print(divmod(7, 3))  # 预期：(2, 1)，依次为商与余数。
```

### 4.3 增强赋值

+=、-=、\*=、/= 等把运算与赋值写在一起，称为增强赋值（augmented assignment）。其他算术和位运算也有相应形式，例如 //=、%=、\*\*=、&=。

对于下面的整数变量，score += 5 会先计算再重新绑定 score，原来的整数对象不变。运算结果也可能是另一种类型。

对于列表等可变对象，增强赋值可能直接修改原对象，因此不能把所有 += 都当成普通加法后赋值；具体区别在第 4 章演示。

```python
score = 10
previous_score = score
score += 5
print(score, previous_score)  # 预期：15 10；原整数未被修改。

score /= 3
print(score, type(score))  # 预期：5.0 <class 'float'>
```

### 4.4 整数位运算

位运算按整数的二进制位进行计算，常用于标志位和掩码。bin 可以把整数转换为带 0b 前缀的二进制字符串，方便观察。

| 运算符 | 对二进制位的作用 |
| --- | --- |
| & | 按位与：两位都是 1 时结果为 1 |
| \| | 按位或：至少一位是 1 时结果为 1 |
| ^ | 按位异或：两位不同时结果为 1 |
| \~ | 按位取反 |
| << | 左移指定的位数 |
| >> | 右移指定的位数 |

用 n 表示移位位数，它必须是非负整数。左移 n 位等价于乘以 2 的 n 次方，右移 n 位等价于除以 2 的 n 次方并向下取整。

Python 整数的位运算不限定在显示出来的几位；对整数 x，\~x 等于 -x - 1。本节操作 int，布尔逻辑使用的 and、or、not 在第 5 章介绍。

```python
left = 0b1100
right = 0b1010
print(bin(left & right))  # 预期：0b1000
print(bin(left | right))  # 预期：0b1110
print(bin(left ^ right))  # 预期：0b110
print(bin(left << 1), bin(left >> 1))  # 预期：0b11000 0b110
print(~left)  # 预期：-13；不是只把显示出来的四位翻转。
print(-7 >> 1)  # 预期：-4，与 -7 // 2 一致。
```

## 5 类型转换

### 5.1 从文字得到数字

input 返回字符串。需要做数值计算时，先用 int 或 float 转换；complex 可以构造复数，str 可以得到用于显示的字符串。

转换必须符合目标类型的输入规则。int 默认按十进制解析文字，也可以通过第二个参数指定进制。int("3.5") 不符合整数字符串的规则，会触发 ValueError。

```python
count_text = "12"  # 与 input 读到的结果一样，这里是字符串。
count = int(count_text)
print(count + 3, type(count))  # 预期：15 <class 'int'>
print(float("1.5"))  # 预期：1.5
print(complex("3+4j"))  # 预期：(3+4j)
print(str(count), type(str(count)))  # 预期：12 <class 'str'>
print(int("1010", 2))  # 预期：10；第二个参数指定二进制。
```

### 5.2 整数转换与舍入

int 将有限浮点数向零方向截断；round 则取最近值，恰好处于中间时选择偶数。round 不指定小数位数时返回整数，第二个参数可以指定保留的小数位数。

round 仍然受到浮点表示误差的影响，不能把它当作修复浮点精度的工具。需要精确十进制运算时，第 20 章会介绍 decimal。

```python
print(int(3.9), int(-3.9))  # 预期：3 -3，都是向零截断。
print(round(2.5), round(3.5))  # 预期：2 4，中点取偶数。
print(round(1.234, 2))  # 预期：1.23
print(round(2.675, 2))  # 预期：2.67；传入的 float 是十进制值的近似。
```

### 5.3 转为布尔值

bool 按对象的真值规则进行转换：False、None、数值零、空字符串和空容器为假值。

非空字符串为真值，因此 bool("False") 不会按单词含义得到 False。读取开关配置时，需要明确解析约定的文字，不能直接把它交给 bool。完整的真值判断在第 5 章展开。

```python
print(bool(None), bool(0), bool(0.0))  # 预期：False False False
print(bool(""), bool([]))  # 预期：False False
print(bool("False"), bool("0"))  # 预期：True True；两者都是非空字符串。
print(int(True), int(False))  # 预期：1 0；显式转换布尔值。
```

## 6 运算符优先级与求值顺序

### 6.1 运算符如何组合

下面是本章算术与位运算符的优先级，从高到低排列；同一行具有相同优先级。括号可以明确改变组合方式。

表中的 x 代表参与运算的数值，称为操作数，可换成具体数字或已定义的变量。+x、-x、\~x 都只有一个操作数，称为一元运算；x - y 则表示两个数相减，其中 y 是另一个操作数。

| 写法 | 运算 |
| --- | --- |
| \*\* | 乘方 |
| +x、-x、\~x | 保持数值、取相反数、整数按位取反 |
| \*、/、//、% | 乘、除、整除、取余 |
| +、- | 加、减 |
| <<、>> | 移位 |
| & | 按位与 |
| ^ | 按位异或 |
| \| | 按位或 |

同级运算通常从左组合，乘方从右组合。乘方与一元运算符有一个边界：左侧的负号在乘方之后作用，右侧的负号可以直接构成负指数。

```python
print(2 + 3 * 4, (2 + 3) * 4)  # 预期：14 20
print(2 ** 3 ** 2, (2 ** 3) ** 2)  # 预期：512 64
print(-2 ** 2, (-2) ** 2)  # 预期：-4 4
print(2 ** -2)  # 预期：0.25
print(6 & 3 + 1, (6 & 3) + 1)  # 预期：4 3；加法优先于按位与。
```

### 6.2 操作数何时求值

优先级决定运算如何组合，求值顺序决定各个操作数何时取得。Python 从左到右对表达式求值；下面两次 input 也按书写顺序调用，随后按乘法优先于加法的规则组合数值。

普通赋值先计算右侧，再把结果赋给左侧名称。不要把“乘法优先”理解成“先调用乘号旁边的 input”。

```python
# 示例输入：先 2，后 3；确认提示顺序，预期结果为 11。
total = int(input("先输入加数：")) + int(input("再输入乘数：")) * 3
print(total)
```
