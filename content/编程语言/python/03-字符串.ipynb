## 1 字符串的表示

### 1.1 引号与多行文本

str 表示字符串，用于保存文本。单引号和双引号的作用相同；需要在内容中保留一种引号时，可以用另一种引号包住整个字符串。

三重引号可以直接包含换行，内部的空格和换行都会保留。两引号之间没有内容时，得到空字符串。

```python
name = "Python"
same_name = 'Python'  # 此处对比两种引号；通常沿用双引号。
message = '他说："开始学习。"'
notice = """课程：Python
主题：字符串"""

print(name == same_name)  # 预期：True；引号不同不影响字符串内容。
print(message)  # 预期：他说："开始学习。"
print(notice)  # 预期分两行显示，换行属于字符串内容。
print("左" + "" + "右")  # 预期：左右；中间是空字符串。
```

### 1.2 转义与内容观察

转义序列用反斜杠引入特殊字符。下面列出常用写法；左列是代码中的写法，右列是它表示的内容。

| 写法 | 中文名称／含义 |
| --- | --- |
| \\n | 换行符 |
| \\t | 水平制表符 |
| \\r | 回车符 |
| \\\\ | 一个反斜杠 |
| \\" | 一个双引号 |
| \\' | 一个单引号 |
| \\u4e2d | 字符“中”，对应 Unicode 码位 U+4E2D |

print 显示字符串内容；repr 返回带引号和必要转义的表示，便于看清换行、空格和反斜杠。repr 中显示的转义写法不等于原字符串里真的包含这些反斜杠。

```python
text = "第一行\n第二行"
print(text)  # 预期分两行显示。
print(repr(text))  # 预期：'第一行\n第二行'；此行把换行显示为转义写法。
print(repr("姓名\t课程"))  # 观察 \t；它是制表符，不是固定数量的空格。
print("他说：\"你好\"")  # 预期：他说："你好"
print("\u4e2d")  # 预期：中
```

### 1.3 原始字符串

在引号前加 r 或 R，得到原始字符串字面值（raw string literal）。例如 r"\\n" 保留反斜杠和字母 n，不把它们解释为换行。结果仍是 str，没有新增一种字符串类型。

原始字符串适合书写含反斜杠的文本，但不能以奇数个反斜杠结束。需要末尾的单个反斜杠时，可以另行拼接；不要把“原始”理解为所有字符都不受引号语法限制。

```python
raw_path = r"C:\new\text.txt"
escaped_path = "C:\\new\\text.txt"

print(raw_path)  # 预期：C:\new\text.txt；\n 和 \t 没有变成换行或制表符。
print(raw_path == escaped_path)  # 预期：True；两种写法得到相同内容。
print(type(raw_path))  # 预期：<class 'str'>
print(r"C:\new" + "\\")  # 预期：C:\new\；用普通字符串补上末尾反斜杠。
```

## 2 读取与组合字符串

### 2.1 长度与索引

str 是由 Unicode 码位构成的序列；码位是 Unicode 为字符分配的编号。len 返回码位数量，索引读取其中一项，结果仍是长度为 1 的字符串。

索引从 0 开始，负数索引从末尾计数。以 "A中B" 为例：

| 字符 | 从开头计数的索引 | 从末尾计数的索引 |
| --- | --- | --- |
| A | 0 | -3 |
| 中 | 1 | -2 |
| B | 2 | -1 |

空字符串没有可读取的位置，索引越界会触发 IndexError。长度也不等于视觉上的字符数：一个带重音的字母可能由两个码位组合显示；编码与 Unicode 在第 12 章继续介绍。

```python
text = "A中B"
print(len(text))  # 预期：3
print(text[0], text[1], text[-1])  # 预期：A 中 B
print(type(text[1]), len(text[1]))  # 预期：<class 'str'> 1

# 两种写法通常显示为同一个带重音的字母，但码位数量不同。
print("é", "e\u0301")  # \u0301 表示组合重音符。
print(len("é"), len("e\u0301"))  # 预期：1 2
```

### 2.2 切片与步长

切片写作 text[start:stop:step]。text 是字符串，start 是开始索引，stop 是停止索引，step 是每次移动的索引间隔，称为步长。结果包含开始位置，不包含停止位置。

步长默认是 1；正数向右取，负数向左取，不能为 0。省略起止位置时，默认边界随步长方向确定，所以 text[::-1] 可以倒序取出整个字符串。

切片允许边界越界；范围内没有可取内容时返回空字符串。不要把切片的边界处理与单个索引越界混为一谈。

```python
text = "notebook"
print(text[1:5])  # 预期：oteb；读取索引 1、2、3、4。
print(text[:4], text[4:])  # 预期：note book
print(text[-4:])  # 预期：book；从倒数第 4 项取到末尾。
print(text[::2])  # 预期：ntbo；读取索引 0、2、4、6。
print(text[::-1])  # 预期：koobeton
print(text[6:1:-2])  # 预期：obt；读取索引 6、4、2，不包含停止位置 1。
print(repr(text[50:]))  # 预期：''；越界切片不等于越界索引。
print(repr(text[:-1:-1]))  # 预期：''；显式的 -1 是最后一项，不是末尾之外。
```

### 2.3 拼接、重复与不可变性

+ 拼接字符串，字符串乘以整数表示重复；重复次数小于或等于 0 时得到空字符串。字符串拼接不会自动把数字转成文字，需要先转换或使用后面的格式化写法。

字符串不可变，不能对某个索引或切片直接赋值。需要改变内容时，构造字符串再重新绑定名称，其他名称仍保留原来的内容。

相邻的字符串字面值可以自动拼接，这条规则不适用于变量。

```python
word = "cat"
original = word
word = "b" + word[1:]

print(word, original)  # 预期：bat cat；原来的字符串没有被修改。
print("-" * 4, repr("-" * 0))  # 预期：---- ''
print("第" + str(3) + "章")  # 预期：第3章；先把整数转换成字符串。
print("学习" "字符串")  # 预期：学习字符串；相邻字面值自动拼接。
```

## 3 常用字符串方法

### 3.1 方法调用与大小写转换

字符串提供的操作可以通过方法调用，例如 "python".upper()。点号连接字符串与方法名，括号内按需要传入参数。

| 方法 | 中文名称／含义 |
| --- | --- |
| str.lower | 转为小写 |
| str.upper | 转为大写 |
| str.casefold | 转为适合忽略大小写比较的形式 |

这些方法返回处理后的字符串，不修改原字符串。casefold 对 Unicode 大小写差异的处理比 lower 更充分，例如德语字母 ß 会变成 ss。

```python
title = "PyThOn"
print(title.lower(), title.upper())  # 预期：python PYTHON
print(title)  # 预期：PyThOn；原字符串不变。

print("Maße".lower(), "Maße".casefold())  # 预期：maße masse
print("Maße".casefold() == "MASSE".casefold())  # 预期：True
```

### 3.2 清理、删除与替换

| 方法 | 中文名称／含义 |
| --- | --- |
| str.strip | 去除两端指定字符，默认去除空白字符 |
| str.lstrip | 只处理开头一端 |
| str.rstrip | 只处理末尾一端 |
| str.removeprefix | 删除匹配的完整前缀 |
| str.removesuffix | 删除匹配的完整后缀 |
| str.replace | 替换匹配的子串，可指定最多替换几次 |

strip、lstrip、rstrip 传入的是“可移除的字符”，不按一个完整前缀或后缀匹配，也不处理字符串中间的字符。

删除固定前后缀时，用 removeprefix 或 removesuffix；不匹配时内容保持不变。replace 默认替换所有匹配项，第三个参数限制替换次数。这些操作都需要使用返回值。

```python
text = " \tPython 3.12 \n"
print(repr(text.strip()))  # 预期：'Python 3.12'
print(repr(text.lstrip()))  # 预期：'Python 3.12 \n'
print(repr(text.rstrip()))  # 预期：' \tPython 3.12'

filename = "report.txt"
print(filename.rstrip(".txt"))  # 预期：repor；末尾属于 .、t、x 的字符均被移除。
print(filename.removesuffix(".txt"))  # 预期：report；只删除完整后缀。
print("v3.12".removeprefix("v"))  # 预期：3.12
print("red-red".replace("red", "blue", 1))  # 预期：blue-red
```

### 3.3 查找、统计与前后缀判断

下表中，sub 表示要查找的子串，text 表示被搜索的字符串。

| 操作或方法 | 中文名称／含义 |
| --- | --- |
| sub in text | 判断是否包含子串，返回布尔值 |
| str.find | 返回首次匹配的索引，未找到时返回 -1 |
| str.index | 返回首次匹配的索引，未找到时触发 ValueError |
| str.count | 统计子串不重叠出现的次数 |
| str.startswith | 判断是否以指定前缀开头 |
| str.endswith | 判断是否以指定后缀结尾 |

只判断是否包含时使用 in；需要位置时再用 find 或 index。索引 0 表示在开头找到，-1 表示未找到，不能直接把 find 的结果当作“是否找到”的布尔值。

```python
filename = "data/data.csv"
print("csv" in filename)  # 预期：True
print(filename.find("data"), filename.find("json"))  # 预期：0 -1
print(filename.index("csv"))  # 预期：10
print(filename.count("data"))  # 预期：2
print(filename.startswith("data/"), filename.endswith(".csv"))  # 均为 True。
print("aaaa".count("aa"))  # 预期：2；匹配不重叠。
```

### 3.4 拆分文本

| 方法 | 中文名称／含义 |
| --- | --- |
| str.split | 从左边拆分，返回列表 |
| str.rsplit | 从右边拆分，返回列表 |
| str.splitlines | 按行边界拆分，默认不保留换行符 |
| str.partition | 按首次出现的分隔符拆成三部分 |

split 不传分隔符时，按连续空白字符拆分并忽略两端空白；显式传入分隔符时，相邻分隔符之间会保留空字符串。第二个参数限制拆分次数，rsplit 则从右侧开始计数。

列表用方括号显示多项结果。partition 返回由“前段、分隔符、后段”组成的元组；找不到分隔符时，后两项为空字符串。列表和元组的操作在第 4 章展开。

```python
print("  A \t B  ".split())  # 预期：['A', 'B']
print(" A  B ".split(" "))  # 预期：['', 'A', '', 'B', '']
print("A,,B,".split(","))  # 预期：['A', '', 'B', '']
print("a-b-c".split("-", 1))  # 预期：['a', 'b-c']
print("a-b-c".rsplit("-", 1))  # 预期：['a-b', 'c']
print("甲\n乙\n".splitlines())  # 预期：['甲', '乙']；末尾换行不增加空行。
print("mode=fast=safe".partition("="))  # 预期：('mode', '=', 'fast=safe')
print("mode".partition("="))  # 预期：('mode', '', '')
```

### 3.5 拼接多个字符串

str.join 表示用指定分隔符连接多个字符串，写作“分隔符.join(字符串序列)”。分隔符放在各项之间，不会自动加在开头或末尾。

序列中的每一项都必须是 str，不能直接混入整数或 bytes，否则触发 TypeError。拼接大量片段时，先收集片段再用 join，避免反复构造越来越长的字符串。

```python
parts = ["Python", "字符串", "练习"]
print(" / ".join(parts))  # 预期：Python / 字符串 / 练习
print("".join(["A", "中", "B"]))  # 预期：A中B；空字符串表示不插入分隔符。
print(",".join(["章节", str(3)]))  # 预期：章节,3；数字先转换成字符串。
```

### 3.6 字符类别判断

以下方法都要求字符串非空，且每个字符符合条件。判断依据是 Unicode 字符类别，不局限于英文字母和 0–9。

| 方法 | 中文名称／含义 |
| --- | --- |
| str.isalpha | 全部是字母类别字符，包含汉字等 |
| str.isalnum | 全部是字母或数值字符 |
| str.isspace | 全部是空白字符 |
| str.isdecimal | 全部是十进制数字字符 |
| str.isdigit | 数字字符，范围还包括上标数字等 |
| str.isnumeric | 数值字符，范围还包括分数字符等 |

数字字符的范围依次扩大。它们判断字符类别，不能直接替代数值转换规则："-12" 可以转为整数，但不是全部由数字字符组成；"²" 的 isdigit 为 True，却不能直接用 int 转换。

```python
print("中文".isalpha(), "A3".isalnum(), " \t\n".isspace())  # 均为 True。

# 每行依次比较十进制数字、数字、数值字符三种范围。
print("12".isdecimal(), "12".isdigit(), "12".isnumeric())  # True True True
print("²".isdecimal(), "²".isdigit(), "²".isnumeric())  # False True True
print("½".isdecimal(), "½".isdigit(), "½".isnumeric())  # False False True

print("-12".isdecimal(), int("-12"))  # 预期：False -12；类别判断与转换不同。
print("".isalpha(), "".isdecimal())  # 预期：False False；空字符串不满足条件。
```

## 4 字符串格式化

格式化把值组织成指定样式的字符串。本章先用 f-string 学习，再对比另外三种写法。

| 名称 | 中文名称／含义 |
| --- | --- |
| f-string | 格式字符串字面值，在字符串内嵌入表达式 |
| str.format | 字符串方法，把传入的值填入模板 |
| format | 内置函数，按规则格式化单个值 |
| % 格式化 | 旧式字符串格式化，本章掌握基本阅读方式 |

### 4.1 f-string 与替换字段

在引号前加 f，花括号内写需要求值的 Python 表达式。字符串创建时就完成求值，变量随后改变不会自动更新已有字符串。

!r 使用 repr 显示值，适合观察引号与转义；表达式后加 = 可以同时显示表达式文本与值。要输出花括号本身，写成 {{ 或 }}。

```python
name = "小林"
score = 88
message = f"{name}：{score + 2} 分"
score = 95

print(message)  # 预期：小林：90 分；已经生成的字符串不会跟着 score 改变。
print(f"{score=}")  # 预期：score=95；用于观察当前值。
print(f"姓名：{name!r}")  # 预期：姓名：'小林'
print(f"字面花括号：{{课程}}")  # 预期：字面花括号：{课程}
```

### 4.2 数值格式

在替换字段的表达式后加冒号，冒号后写格式说明符。例如 f"{price:.2f}" 中，price 是待显示的数值，.2f 表示用定点形式保留两位小数。

| 格式说明符 | 中文名称／含义 |
| --- | --- |
| .2f | 定点形式，小数点后显示 2 位 |
| .1% | 乘以 100 后显示 1 位小数，并添加百分号 |
| 04d | 十进制整数，至少 4 位宽，不足时在符号之后补零 |
| , | 对十进制整数等支持的数值格式使用千位分隔符 |
| #x | 十六进制整数，包含 0x 前缀 |
| .2e | 科学计数法，系数的小数点后显示 2 位 |

格式化产生字符串，不会改变原数值，也不会修复浮点表示误差。格式要与类型匹配，例如 d 用于整数格式。

```python
price = 12.375
print(f"{price:.2f}")  # 预期：12.38；按格式舍入显示。
print(price)  # 预期：12.375；原数值未变。
print(f"{0.375:.1%}")  # 预期：37.5%；输入已经是比例，不要再乘以 100。
print(f"{7:04d}")  # 预期：0007
print(f"{1234567:,}")  # 预期：1,234,567
print(f"{255:#x}")  # 预期：0xff
print(f"{1234.0:.2e}")  # 预期：1.23e+03；指数表示 10 的 3 次方。
```

### 4.3 宽度、对齐与动态精度

宽度规定字段至少占多少个字符位置；内容更长时不会被截断。对齐符号决定内容放在哪一侧，前面还可以指定一个填充字符。

| 格式说明符 | 中文名称／含义 |
| --- | --- |
| \<8 | 至少 8 位宽，左对齐 |
| \>8 | 至少 8 位宽，右对齐 |
| ^8 | 至少 8 位宽，居中 |
| .\>8 | 至少 8 位宽，右对齐，用点号填充 |

格式说明符里还可以嵌入替换字段。下面 precision 是保存小数位数的变量，.{precision}f 会先取得位数，再按定点形式格式化。

```python
label = "AI"
print(f"|{label:<8}|")  # 预期：|AI      |；竖线帮助观察两端空格。
print(f"|{label:>8}|")  # 预期：|      AI|
print(f"|{label:^8}|")  # 预期：|   AI   |
print(f"|{label:.>8}|")  # 预期：|......AI|
print(f"|{label:1}|")  # 预期：|AI|；最小宽度 1 不会截断已有内容。

precision = 3
print(f"{1 / 7:.{precision}f}")  # 预期：0.143；小数位数来自 precision。
```

### 4.4 str.format 与 format

str.format 在模板字符串上调用。{} 按传入顺序取值，{0}、{1} 按从 0 开始的位置取值，{name} 按调用时提供的同名参数取值。自动编号和手动数字编号不能混用。

花括号内可以使用前面的格式说明符，但不能像 f-string 一样随意写算术表达式；需要计算时，先计算再传入。

内置函数 format 接收“单个值、格式说明符”两个参数，第二个参数直接写格式，不加冒号或花括号。

```python
print("{} / {}".format("Python", "字符串"))  # 预期：Python / 字符串
print("{1} / {0}".format("Python", "字符串"))  # 预期：字符串 / Python

template = "{name}：{score:.1f} 分"
print(template.format(name="小林", score=88.5))  # 预期：小林：88.5 分

formatted = format(12.375, ".2f")
print(formatted, type(formatted))  # 预期：12.38 <class 'str'>
print(formatted == f"{12.375:.2f}")  # 预期：True；这里使用相同的格式规则。
```

### 4.5 了解旧式 % 格式化

当 % 运算符左侧是字符串时，这里表示格式化，作用与数字取余不同。格式字符串内用 % 引入转换标记，右侧提供待填入的值。

| 标记 | 中文名称／含义 |
| --- | --- |
| %s | 用 str 转换并填入值 |
| %d | 十进制整数形式 |
| %.2f | 小数点后显示 2 位 |
| %% | 输出一个百分号，不消耗待填入的值 |

填入多个值时，用元组按标记顺序提供，数量需要对应。这里的浮点格式不会像 f-string 的百分比格式那样自动乘以 100。

```python
print("课程：%s" % "Python")  # 预期：课程：Python
print("章节：%d，得分：%.2f" % (3, 88.5))  # 预期：章节：3，得分：88.50
print("完成 %.1f%%" % 37.5)  # 预期：完成 37.5%；%% 输出一个百分号。
```

## 5 综合应用：整理一条文本记录

把清理、拆分、类型转换和格式化连起来，生成一行摘要。

输入约定为“课程名、数量、单价”三列，列内不包含逗号，数量和单价是可转换的数值文字。split 的结果是列表，下面用 fields[0]、fields[1]、fields[2] 读取各列，索引同样从 0 开始。

```python
record = "  python , 3 , 12.50  "

# 1. 拆分约定的三列，单独清理课程名两端的空白。
fields = record.strip().split(",")
course = fields[0].strip().upper()

# 2. 数值转换后再计算；int 和 float 可以接受文字两端的空白。
quantity = int(fields[1])
unit_price = float(fields[2])
total_price = quantity * unit_price

# 3. 组织摘要；预期：PYTHON | 003 项 | 合计 37.50
print(f"{course} | {quantity:03d} 项 | 合计 {total_price:.2f}")
```
