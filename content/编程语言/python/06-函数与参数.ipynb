## 1 定义、调用与返回

### 1.1 形参与实参

def 定义函数，函数名后写形参列表，冒号后的缩进代码组成函数体。执行定义会创建函数；函数体在调用时执行。

| 名称 | 中文含义 | 下例对应内容 |
| --- | --- | --- |
| parameter | 形参，定义中用于接收传入值的名称 | price、quantity |
| argument | 实参，调用时实际传入的值 | 12、3 |

下面的 total_price 表示计算总价，price 表示单价，quantity 表示数量。return 将计算结果交给调用方。函数体开头的三引号字符串是文档字符串，用一句话说明用途。

```python
def total_price(price, quantity):
    """返回单价与数量的乘积。"""
    return price * quantity


amount = total_price(12, 3)
print(amount)  # 36：调用时将 12、3 分别绑定给 price、quantity。
print(total_price(5 + 1, 2))  # 12：先计算实参表达式，再进入函数。
```

### 1.2 return 与 print

return 结束当前函数调用，并提供返回值；print 负责输出文本。打印出一个值，不代表函数返回了这个值。

没有执行 return 就到达函数末尾，或者执行不带值的 return，都会返回 None。调用方可以保存返回值、参与后续计算，或者忽略它。

```python
def show_total(price, quantity):
    """打印总价。"""
    print(price * quantity)


shown = show_total(12, 3)  # 打印 36。
print(shown)  # None：show_total 没有显式返回计算结果。

returned = total_price(12, 3)  # 沿用上一示例的函数。
print(returned + 4)  # 40：返回的数值可以继续参与计算。
```

### 1.3 提前返回与返回元组

return 可以放在条件或循环内部，一旦执行，就退出整个函数调用。

下面的 score_bounds 表示取得成绩上下界；输入为空时提前返回 None，否则把最小值、最大值组成一个元组返回。习惯上说的“返回多个值”，在这种写法中实际返回的是一个元组。

```python
def score_bounds(scores):
    """返回成绩的最小值和最大值；空输入返回 None。"""
    if not scores:
        return None
    return min(scores), max(scores)


bounds = score_bounds([80, 60, 90])
lowest, highest = bounds  # 对返回的元组进行解包。
print(bounds)  # (60, 90)
print(lowest, highest)  # 60 90
print(score_bounds([]))  # None：不会继续执行 min 和 max。
```

## 2 常用传参方式与默认值

### 2.1 位置实参与关键字实参

| 方式 | 中文含义 | 对应关系 |
| --- | --- | --- |
| positional argument | 位置实参 | 按出现顺序匹配形参 |
| keyword argument | 关键字实参 | 按“形参名=值”匹配形参 |

普通调用中，直接写出的位置实参放在关键字实参之前。关键字实参之间的顺序可以调整，但一个形参不能重复接收值。

缺少必需实参、重复传值、传入函数不接受的关键字或多余的位置实参，都会引发 TypeError。

```python
# 沿用 total_price(price, quantity)。
print(total_price(12, 3))  # 36：两个位置实参。
print(total_price(price=12, quantity=3))  # 36：两个关键字实参。
print(total_price(quantity=3, price=12))  # 36：按名称匹配。
print(total_price(12, quantity=3))  # 36：混合使用。

# total_price(12, price=3) 会重复给 price 传值，不能这样调用。
```

### 2.2 默认值何时确定

给形参指定默认值后，调用时可以省略对应实参。默认值在执行 def 时求值，此后省略实参的调用复用这个值；它不会在每次调用时重新计算。

在位置形参中，没有默认值的形参应位于有默认值的形参之前。仅限关键字形参不受这一顺序限制，下一节会介绍。

下面的 default_quantity 是定义函数时使用的默认数量。修改这个名称的绑定，不会改写函数已经保存的默认值。

```python
default_quantity = 2


def calculate_total(price, quantity=default_quantity):
    """返回总价；省略数量时使用定义时确定的默认值。"""
    return price * quantity


default_quantity = 5

print(calculate_total(10))  # 20：默认数量仍为 2。
print(calculate_total(10, 3))  # 30：显式实参取代默认值。
print(calculate_total(10, quantity=0))  # 0：传入 0 不会启用默认值。
```

## 3 五种形参

### 3.1 仅限位置形参

形参的种类决定允许怎样调用函数。args 和 kwargs 是常见命名习惯，分别用于收集额外的位置实参和关键字实参；名称可以替换，星号语法才决定作用。

| 英文名称 | 中文名称 | 定义中的写法 |
| --- | --- | --- |
| positional-only | 仅限位置形参 | 位于 / 之前 |
| positional-or-keyword | 位置或关键字形参 | 普通形参，位于 / 之后、星号分隔之前 |
| var-positional | 可变位置形参 | \*args，接收元组 |
| keyword-only | 仅限关键字形参 | 位于单独的 \* 或 \*args 之后 |
| var-keyword | 可变关键字形参 | \*\*kwargs，接收字典 |

这里“可变”指接收的实参数量可变，不是说接收它们的容器都可修改。/ 和单独的 \* 是分隔符，本身不接收实参；没有 / 时，普通形参允许按位置或关键字传入。

下面的 text 位于 / 之前，只能按位置传入；prefix 表示前缀，可以按位置或关键字传入。

```python
def add_prefix(text, /, prefix="提示："):
    """在文本前添加前缀。"""
    return prefix + text


print(add_prefix("已保存"))  # 提示：已保存
print(add_prefix("已保存", "结果："))  # 结果：已保存
print(add_prefix("已保存", prefix="结果："))  # 结果：已保存

# add_prefix(text="已保存") 会引发 TypeError：text 仅限位置。
```

### 3.2 仅限关键字形参

单独的 \* 后面是仅限关键字形参，调用时必须写出形参名。这适合让可选行为更明确，例如是否转换大小写。

下面的 text 是输入文本，lowercase 表示是否转为小写。

```python
def clean_label(text, *, lowercase=False):
    """去掉文本两端空白，并按选项转换为小写。"""
    cleaned = text.strip()
    if lowercase:
        cleaned = cleaned.lower()
    return cleaned


print(clean_label(" Python "))  # Python
print(clean_label(" Python ", lowercase=True))  # python

# clean_label(" Python ", True) 会引发 TypeError。
```

### 3.3 收集额外的位置实参

定义中的 \*parts 把尚未匹配普通形参的位置实参收集为元组，没有额外实参时得到空元组。parts 在下例表示待拼接的多个字符串。

\*parts 后面的 separator 是仅限关键字形参，表示分隔符；不需要再写一个单独的星号。

```python
def join_labels(*parts, separator=" / "):
    """使用分隔符连接任意数量的字符串。"""
    return separator.join(parts)


print(join_labels("容器", "循环", "函数"))  # 容器 / 循环 / 函数
print(join_labels("基础", "进阶", separator=" → "))  # 基础 → 进阶
print(repr(join_labels()))  # ''：parts 是空元组。
```

### 3.4 收集额外的关键字实参

定义中的 \*\*fields 把尚未匹配其他形参的关键字实参收集为字典；没有额外关键字时得到空字典。

下例用 fields 表示记录中的字段，title 表示标题。\*\*fields 不会替普通形参接收多余的位置实参。

```python
def make_record(title, **fields):
    """创建包含标题和额外字段的新字典。"""
    record = {"title": title}
    record.update(fields)
    return record


print(make_record("Python", level=1, hours=12))
# {'title': 'Python', 'level': 1, 'hours': 12}
print(make_record("Python"))  # {'title': 'Python'}：fields 为空。
```

### 3.5 在一个定义中组合使用

下面的 collect_arguments 只用于观察绑定结果：first 接收第一个位置实参，second 是普通形参，rest 收集剩余位置实参，enabled 是仅限关键字形参，options 收集剩余关键字实参。

\*\*options 放在最后。实际设计函数时，只加入调用方确实需要的参数；普通参数有明确名称时，不必全部改成可变参数。

```python
def collect_arguments(first, /, second=0, *rest, enabled=True, **options):
    """返回五种形参各自接收到的值。"""
    return first, second, rest, enabled, options


result = collect_arguments(1, 2, 3, 4, enabled=False, color="blue")
print(result)  # (1, 2, (3, 4), False, {'color': 'blue'})

print(collect_arguments(1, second=2))
# (1, 2, (), True, {})：没有额外实参时，收集容器为空。
```

## 4 调用时解包实参

在函数定义中，星号语法用于收集；在函数调用中，它用于展开已有数据。

| 写法 | 中文含义 |
| --- | --- |
| \*values | 将可迭代对象 values 中的元素展开为位置实参 |
| \*\*options | 将映射 options 中的条目展开为关键字实参 |

沿用第 1 节的 total_price 和第 3.3 节的 join_labels。下例的 values 保存单价与数量，options 保存同一组带名称的实参，labels 保存多个标签。

展开的关键字必须是字符串；它们仍要遵守形参匹配规则。显式传参或多份映射展开时，重复提供同一个关键字会引发 TypeError，不会像字典合并那样覆盖。

```python
values = [12, 3]
options = {"price": 12, "quantity": 3}
labels = ["容器", "循环", "函数"]

print(total_price(*values))  # 36：等价于 total_price(12, 3)。
print(total_price(**options))  # 36：按字典的键匹配形参。
print(total_price(*values[:1], quantity=3))  # 36：解包与显式传参。

print(join_labels(*labels, separator="、"))  # 容器、循环、函数

# total_price(12, quantity=3, **{"quantity": 4}) 会引发 TypeError。
```

## 5 参数绑定与可变对象

### 5.1 重新绑定与原地修改

调用函数时，形参绑定到传入的对象，不会自动复制对象。重新给形参赋值，只改变这个局部名称的绑定；修改它引用的可变对象，则会影响同样引用该对象的调用方。

下面对同一份列表分别执行两种操作。rebind_items 表示重新绑定形参，append_item 表示向传入列表追加元素。

```python
def rebind_items(items):
    """返回一个新列表，不修改传入列表。"""
    items = items + ["循环"]
    return items


def append_item(items):
    """向传入列表追加一个元素。"""
    items.append("函数")


lessons = ["容器"]
new_lessons = rebind_items(lessons)
print(lessons, new_lessons)  # ['容器'] ['容器', '循环']

append_item(lessons)
print(lessons)  # ['容器', '函数']：调用方看到原地修改。
print(new_lessons)  # ['容器', '循环']：此前创建的新列表不受影响。
```

### 5.2 可变默认值为什么会保留变化

默认值只在执行定义时求值。用列表作默认值并修改它，会使省略该实参的后续调用继续使用已经改变的列表。

下面故意展示这种容易出错的写法。append_tag_shared 的 tags 是默认标签列表；第一次调用返回的对象，也会反映第二次调用对它的修改。

```python
def append_tag_shared(tag, tags=[]):
    """演示默认列表被多次调用共享的行为。"""
    tags.append(tag)
    return tags


first_tags = append_tag_shared("基础")
print(first_tags)  # ['基础']

second_tags = append_tag_shared("进阶")
print(first_tags)  # ['基础', '进阶']：早先返回的对象也被修改。
print(second_tags is first_tags)  # True：两次返回同一个列表。
```

### 5.3 在调用时创建默认容器

如果希望每次省略实参时都使用新列表，可将默认值设为 None，再在函数体中创建列表。下例约定 None 表示“未提供列表”。

使用 is None 检查这个约定；不要改成 if not tags，否则调用方显式传入的空列表也会被换掉。

如果函数会修改调用方传入的对象，应在文档字符串中说明。需要保留输入时，返回新容器；遇到嵌套对象，再按第 4 章的拷贝规则选择处理方式。

```python
def append_tag(tag, tags=None):
    """追加标签；未传列表时新建，传入列表时直接修改它。"""
    if tags is None:
        tags = []
    tags.append(tag)
    return tags


first_tags = append_tag("基础")
second_tags = append_tag("进阶")
print(first_tags, second_tags)  # ['基础'] ['进阶']
print(first_tags is second_tags)  # False：两个独立列表。

provided_tags = []
result_tags = append_tag("练习", provided_tags)
print(provided_tags)  # ['练习']：显式传入的空列表被保留并修改。
print(result_tags is provided_tags)  # True
```

## 6 文档字符串与调用约定

文档字符串（docstring）是函数体的第一条字符串语句。第一行简述用途；说明较长时，空一行再写输入约定、返回值和必要的副作用。

函数对象的 \_\_doc\_\_ 属性可读取文档字符串，help 也会使用它。类型标注在后续章节系统学习，本章先用文字交代输入范围。

下面的 passed_scores 接受 0～100 的整数成绩和通过分数线，返回新列表，不修改输入。

```python
def passed_scores(scores, *, threshold=60):
    """返回达到分数线的成绩。

    scores 中的元素和 threshold 均约定为 0～100 的整数。
    返回新列表，不修改传入的 scores。
    """
    return [score for score in scores if score >= threshold]


scores = [58, 60, 90]
print(passed_scores(scores))  # [60, 90]
print(scores)  # [58, 60, 90]：输入保持不变。
print(passed_scores.__doc__)  # 可查看用途、输入约定和修改行为。
```

## 7 综合应用：生成成绩摘要

把筛选与汇总封装为 summarize_scores（生成成绩摘要），由调用方负责打印结果。

scores 是“姓名 → 整数成绩”的字典，threshold 是 0～100 的整数分数线。函数排除范围外的成绩，返回包含人数、通过名单和平均分的新字典；没有合法成绩时，平均分为 None。

```python
def summarize_scores(scores, *, threshold=60):
    """汇总整数成绩；分数线约定为 0～100，不修改输入。"""
    valid = {
        name: score
        for name, score in scores.items()
        if 0 <= score <= 100
    }
    passed = [name for name, score in valid.items() if score >= threshold]
    average = sum(valid.values()) / len(valid) if valid else None
    return {"count": len(valid), "passed": passed, "average": average}


scores = {"Ada": 90, "Bo": 58, "Chen": 120, "Dina": 80}
summary = summarize_scores(scores, threshold=80)

print(summary)  # 3 人；通过名单为 Ada、Dina；平均分为 76.0。
print(summarize_scores({}))
# {'count': 0, 'passed': [], 'average': None}
print(scores["Chen"])  # 120：原始记录保留。
```

## 本章小结

- 定义中的名称是形参，调用时传入的值是实参；return 与打印承担不同任务。
- 默认值在执行定义时求值；显式传入实参时不使用该默认值。
- / 限定位置传参，单独的 \* 限定后续形参按关键字传入。
- \*args、\*\*kwargs 在定义中收集实参；调用中的星号展开已有数据。
- 重新绑定形参不会重绑调用方的变量；原地修改会影响共享对象。
- 用文档字符串说明输入、返回值和修改行为；需要独立的默认容器时，在调用中创建。
