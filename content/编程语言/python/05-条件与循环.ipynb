## 1 条件如何判断

### 1.1 真值

条件不必直接是 True 或 False。False、None、数值零、空字符串和空容器按假值处理；本章使用的其他非空容器按真值处理。

字符串按是否为空判断，不按文字含义判断，所以 "False" 和 "0" 都是真值。需要区分“没有提供”和“提供了零”时，应明确检查 is None。

```python
print(bool([]), bool({}), bool(range(0)))  # 预期：False False False
print(bool([0]), bool("False"), bool("0"))  # 预期：True True True
zero = 0
print(bool(zero), zero is None)  # 预期：False False；零不是 None。
print(not [])  # 预期：True；not 把真值判断取反，结果是布尔值。
```

### 1.2 比较与比较链

| 写法 | 中文名称／含义 |
| --- | --- |
| == | 值相等 |
| != | 值不相等 |
| < | 小于 |
| <= | 小于或等于 |
| \> | 大于 |
| \>= | 大于或等于 |
| is | 是否为同一对象 |
| is not | 是否不是同一对象 |

比较链可以表达区间。0 <= score <= 100 中，score 是待检查的分数，两项条件都要成立，中间表达式只求值一次。

字符串按 Unicode 码位逐项比较；列表和元组在同类型内部按元素的字典序比较。不是所有对象都支持大小排序，例如字典、复数不支持，字符串也不能直接与整数比较大小。

```python
score = 85
print(0 <= score <= 100)  # 预期：True
print(3 == 3.0, 3 != 4)  # 预期：True True
print("10" < "2")  # 预期：True；比较文本顺序，不是数值 10 与 2。
print([1, 9] < [2, 0])  # 预期：True；第一项已决定大小。
```

### 1.3 逻辑运算与短路

下表中，a、b 表示两个待求值的表达式。

| 写法 | 中文名称／含义 |
| --- | --- |
| a and b | a 为假时返回 a，否则计算并返回 b |
| a or b | a 为真时返回 a，否则计算并返回 b |
| not a | 返回与 a 真值相反的布尔值 |

and 与 or 在结果已能确定时跳过右侧，称为短路求值。它们返回实际操作数，不保证返回 bool。

比较运算优先于 not，not 优先于 and，and 优先于 or；组合较复杂时加括号。用 or 提供默认值会替换所有假值，包括本来有效的 0。

```python
print("" or "未填写", "Python" and 3)  # 预期：未填写 3
print(0 or 10)  # 预期：10；不能用此写法区分零与缺失。

count = 0
print(count != 0 and 10 / count > 1)  # 预期：False；右侧没有执行，不发生除零。
print(not False and True or False)  # 预期：True
```

### 1.4 成员检测

in 判断是否包含，not in 判断是否不包含。字符串检查子串，列表、元组、集合检查元素，字典默认检查键。

若要查字典的值或完整键值对，明确使用 values 或 items。不要用数值大小比较或对象身份代替成员检测。

```python
profile = {"name": "Ada", "level": 2}
print("Py" in "Python", "Java" not in ["Python"])  # 预期：True True
print("name" in profile, "Ada" in profile)  # 预期：True False
print("Ada" in profile.values())  # 预期：True
print(("level", 2) in profile.items())  # 预期：True
```

## 2 选择执行路径

### 2.1 if、elif 与 else

if 开始判断，elif 添加后续条件，else 处理前面都不成立的情况；elif 和 else 均可省略。条件后的冒号与缩进标明对应代码块。

从上到下只执行第一个条件为真的分支，其余条件不再计算。若写成多个独立的 if，则每个判断都会单独进行。

```python
score = 85
if score >= 90:
    level = "优秀"
elif score >= 60:
    level = "通过"
else:
    level = "未通过"

print(level)  # 预期：通过；先检查 90 分，再检查 60 分。
```

### 2.2 条件表达式

“真时的值 if 条件 else 假时的值”可以在两种值之间选择。先判断条件，只计算被选中的那一侧。

它适合简单赋值；多步骤处理继续使用 if 语句，不把多个条件表达式嵌套在一起。

```python
provided_limit = 0
limit = 10 if provided_limit is None else provided_limit
print(limit)  # 预期：0；只替换 None，保留有效的零。

count = 0
average = 12 / count if count else None
print(average)  # 预期：None；除法分支未执行。
```

## 3 for 与 while

### 3.1 遍历元素与指定次数

for 从可迭代对象中逐项取值，每次把当前项绑定到循环变量，再执行循环体。遍历字典时默认取得键，用 items 可以同时取得键与值。

需要指定次数时，使用第 4 章介绍的 range。循环变量在迭代中会被不断赋值，修改它不会改变 range 下一次提供的数。

```python
lessons = ["容器", "循环"]
for lesson in lessons:
    print(lesson)  # 预期依次显示：容器、循环。

scores = {"Ada": 80, "Bo": 90}
for name, score in scores.items():
    print(name, score)  # 预期依次显示：Ada 80、Bo 90。

total = 0
for number in range(1, 4):
    total += number
print(total)  # 预期：6；累加 1、2、3。
```

### 3.2 按条件重复

while 每轮开始前检查条件，条件为假就停止；首次判断为假时，循环体一次也不执行。

需要让状态逐步靠近停止条件。下面 remaining 表示剩余次数，每轮减少一次；如果忘记更新，它会一直循环。

```python
remaining = 3
while remaining > 0:
    print(remaining)  # 预期依次显示：3、2、1。
    remaining -= 1

print("结束", remaining)  # 预期：结束 0
```

### 3.3 break 与 continue

| 语句 | 中文名称／含义 |
| --- | --- |
| break | 立即退出最近一层循环 |
| continue | 跳过本轮剩余部分，进入下一次迭代 |

在 while 中，continue 会返回条件检查；必要的状态更新不能总被它跳过。嵌套循环中的 break 不会一次退出所有层。

```python
numbers = [-2, 3, 0, 8]
accepted = []
for number in numbers:
    if number < 0:
        continue
    if number == 0:
        break
    accepted.append(number)

print(accepted)  # 预期：[3]；负数跳过，遇到零停止，8 没有被处理。
```

### 3.4 循环的 else

for 正常取完全部元素、while 因条件为假结束时，执行其 else；即使循环体一次也未执行，也属于这种正常结束。

break 会跳过所属循环的 else；异常等提前退出也不属于正常结束。这里 else 与 for 对齐，表达“遍历完成仍没找到”，不是某一次 if 的补充分支。

```python
numbers = [1, 3, 5]
for target in [3, 7]:
    for number in numbers:
        if number == target:
            print("找到", target)
            break  # 只退出搜索 number 的内层循环。
    else:
        print("未找到", target)

# 预期依次显示：找到 3、未找到 7。
# 把 numbers 改为空列表时，两次搜索都应执行 else。
```

### 3.5 pass 不改变执行流程

pass 是不执行任何动作的语句，用在语法要求代码块、但暂时没有操作的位置。它不是 continue，也不是 break。

下面只是观察 pass 的行为；占位语句本身不代表业务处理已经完成。

```python
for state in ["skip", "run"]:
    if state == "skip":
        pass
    print(state)

# 预期依次显示 skip、run；pass 后面的 print 仍然执行。
```

### 3.6 遍历时避免增删原容器

遍历列表时增删元素可能跳过项目；遍历字典时增删键可能报错或漏项。筛选内容时优先新建容器；确需删除原字典中的键时，可以遍历键的快照。

```python
scores = [80, -1, 90]
valid_scores = []
for score in scores:
    if score >= 0:
        valid_scores.append(score)
print(valid_scores, scores)  # 预期：[80, 90] [80, -1, 90]

states = {"Ada": "active", "Bo": "inactive"}
for name in list(states):  # list 收集键的快照，删除不影响当前遍历。
    if states[name] == "inactive":
        del states[name]
print(states)  # 预期：{'Ada': 'active'}
```

## 4 遍历与汇总工具

### 4.1 编号与反向遍历

enumerate（枚举）在遍历时提供“编号、元素”；编号默认从 0 开始，也可通过 start 指定起点。

reversed（反向迭代）按相反顺序提供元素，不修改原列表。它适用于列表、元组、range 等支持反向遍历的对象，不适用于集合。

```python
lessons = ["容器", "循环", "函数"]

for number, lesson in enumerate(lessons, start=1):
    print(number, lesson)  # 编号依次为 1、2、3。

print(list(reversed(lessons)))  # ['函数', '循环', '容器']
print(lessons)  # 原列表仍按“容器、循环、函数”排列。
```

### 4.2 并行遍历

zip 将多个可迭代对象中相同位置的元素组成元组。默认在最短的输入耗尽时结束；如果业务要求输入等长，使用 strict=True，长度不一致会在遍历时引发 ValueError。

zip、enumerate 和 reversed 返回迭代器，元素按需提供。迭代器耗尽后，再次遍历不会重新开始；迭代器的完整机制在后续章节学习。

```python
names = ["Ada", "Bo", "Chen"]
scores = [80, 90]

print(list(zip(names, scores)))  # [('Ada', 80), ('Bo', 90)]

for name, score in zip(names[:2], scores, strict=True):
    print(name, score)  # 本例两份输入等长，逐项配对。

pairs = zip(names, scores)
print(list(pairs))  # 第一次遍历得到两个元组。
print(list(pairs))  # []：迭代器已经耗尽。
```

### 4.3 判断与数值汇总

| 工具 | 中文含义 | 空输入的结果 |
| --- | --- | --- |
| all | 所有元素的真值都为真 | True |
| any | 至少一个元素的真值为真 | False |
| sum | 数值求和，默认从 0 开始累加 | 0 |
| min | 最小值 | 未提供 default 时引发 ValueError |
| max | 最大值 | 未提供 default 时引发 ValueError |

all 和 any 一旦能确定结果，就停止继续读取输入。这里的短路发生在遍历阶段；调用前已构造好的列表不会因此少计算其中的元素。

min 和 max 接收一个可迭代对象时，可用 default 指定空输入的结果。

```python
flags = [True, False, True]
scores = [80, 90]

print(all(flags), any(flags))  # False True
print(sum(scores), min(scores), max(scores))  # 170 80 90
print(all([]), any([]), sum([]))  # True False 0
print(min([], default=None), max([], default=None))  # None None
```

## 5 推导式与赋值表达式

### 5.1 列表推导式

列表推导式（list comprehension）用于根据已有数据生成新列表。先取出元素，检查可选的 if 条件，再计算要加入结果的表达式。

下面的 number 是每次取出的数值，numbers 是输入列表。先看普通循环，再对照等价的推导式；当条件或处理步骤较多时，保留普通循环更容易阅读。

```python
numbers = [-2, 0, 3, 4]
squares = []

for number in numbers:
    if number > 0:
        squares.append(number ** 2)

compact_squares = [number ** 2 for number in numbers if number > 0]
print(squares)  # [9, 16]
print(compact_squares == squares)  # True：两种写法得到相同结果。
print(numbers)  # [-2, 0, 3, 4]：输入没有改变。
```

### 5.2 集合推导式与字典推导式

集合推导式（set comprehension）使用花括号生成元素，自动去重；字典推导式（dict comprehension）使用“键: 值”生成条目，相同键后生成的值会覆盖先前的值。

下面的 name 表示姓名，number 表示从 1 开始的编号。

```python
names = ["Ada", "Bo", "Ada"]

name_lengths = {len(name) for name in names}
positions = {
    name: number
    for number, name in enumerate(names, start=1)
}

print(sorted(name_lengths))  # [2, 3]：集合去重，排序只用于稳定展示。
print(positions)  # {'Ada': 3, 'Bo': 2}：Ada 对应最后一次出现的编号。
```

### 5.3 赋值表达式

赋值表达式（assignment expression）使用 :=，把表达式的结果绑定到一个名称，同时产生这个值。它也常被称为海象运算符（walrus operator）。

下面将列表长度保存为 count，再比较 count 是否大于 2。括号明确了先保存长度、后比较的顺序。

普通赋值使用 =，相等比较使用 ==。只有确实需要在表达式中保留中间结果时，才考虑 :=。

```python
items = ["容器", "循环", "函数"]

if (count := len(items)) > 2:
    print(f"共有 {count} 项")  # 共有 3 项

print(count)  # 3：保存的是长度，不是比较结果 True。
```

## 6 基本模式匹配

### 6.1 字面值、备选模式与通配符

match / case 执行结构模式匹配（structural pattern matching）：按顺序尝试各个 case，执行首个匹配成功且守卫条件通过的分支，不会继续执行后面的分支。

| 写法 | 含义 |
| --- | --- |
| case 200 | 匹配给定字面值 |
| case 401 \| 403 | 匹配任一备选模式 |
| case _ | 通配符，匹配任何值且不绑定名称 |

没有分支匹配时，整个 match 结束。case 后的普通裸名称是捕获模式，会绑定名称；不要用它来表示“与已有变量的值相等”。

```python
status = 403

match status:
    case 200:
        message = "成功"
    case 401 | 403:
        message = "无权访问"
    case 404:
        message = "未找到"
    case _:
        message = "其他状态"

print(message)  # 无权访问
```

### 6.2 序列模式与守卫条件

序列模式可以匹配列表、元组等序列，并将其中的元素绑定到名称；str、bytes 和 bytearray 不按序列模式匹配。

守卫条件（guard）写在 case 模式后的 if 中，仅在模式匹配后检查；条件不通过时继续尝试下一分支。

下面约定移动命令由名称和两个数值组成。dx 表示水平移动量，dy 表示垂直移动量；捕获名称本身不检查数值类型。

```python
command = ["move", 2, -1]

match command:
    case ["move", dx, dy] if dx != 0 or dy != 0:
        print(f"移动：水平 {dx}，垂直 {dy}")  # 水平 2，垂直 -1
    case ["move", 0, 0]:
        print("保持原位")  # 本例不输出；把 command 改为 ["move", 0, 0] 才进入此分支。
    case _:
        print("命令格式不匹配")  # 本例不输出；前面的移动模式已经匹配。
```

### 6.3 映射模式

映射模式按指定的键匹配值，并允许输入包含额外的键。下面的 width 和 height 捕获宽度、高度；输入约定这两项为数值，守卫再检查它们是否为正数。

本章掌握字面值、序列和映射的基本用法即可。类模式需要先学习类与对象。

```python
event = {
    "kind": "resize",
    "width": 800,
    "height": 600,
    "source": "toolbar",
}

match event:
    case {"kind": "resize", "width": width, "height": height} if (
        width > 0 and height > 0
    ):
        print(f"调整为 {width} × {height}")  # 调整为 800 × 600
    case _:
        print("事件格式不匹配或尺寸无效")  # 本例不输出；800 × 600 满足前一分支。

# source 是额外的键，不妨碍上述映射模式匹配。
```

## 7 综合应用：成绩汇总

约定输入是“姓名 → 整数成绩”的字典，合法成绩范围为 0～100。先筛选合法记录，再统计通过名单和平均分。

筛选过程建立新容器；平均分在没有合法成绩时使用 None，避免除以零。

```python
scores = {"Ada": 90, "Bo": 58, "Chen": 120, "Dina": 80}

valid_scores = {
    name: score
    for name, score in scores.items()
    if 0 <= score <= 100
}
passed_names = [
    name
    for name, score in valid_scores.items()
    if score >= 60
]
average = (
    sum(valid_scores.values()) / len(valid_scores)
    if valid_scores else None
)

print(valid_scores)  # {'Ada': 90, 'Bo': 58, 'Dina': 80}
print(passed_names)  # ['Ada', 'Dina']
print(average)  # 76.0
print(scores["Chen"])  # 120：筛选没有删除输入中的条目。
```

## 本章小结

- 真值测试服务于条件判断；and、or 会短路，并可能返回非布尔值。
- for 遍历元素，while 重复检查条件；break、continue 和 pass 的作用不同。
- 循环的 else 在正常结束时执行；zip 默认按最短输入结束，等长要求使用 strict=True。
- 简单的数据转换可使用推导式；需要保存表达式中间值时可使用 :=。
- match 按模式和守卫选择分支；捕获名称不等同于比较已有变量。
