## 1 运行代码

选中下面的代码单元，按 Shift+Enter 执行，结果会显示在单元下方。Python 也可以在终端中运行：

| 方式 | 怎么运行 |
| --- | --- |
| Notebook | 在代码单元中写代码并执行 |
| 终端交互 | 在已激活环境的终端输入 python，看到 >>> 后输入 Python 代码；用 quit() 退出 |
| 脚本 | 将代码保存为 .py 文件，例如 hello.py；在文件所在目录的终端执行 python hello.py |

python 和 python hello.py 是终端命令。Notebook 代码单元中只写 Python 代码，不复制 >>> 提示符。

print 是用于输出的内置函数。函数名后加圆括号表示调用；引号中的文字称为字符串。

```python
# 预期在单元下方看到：你好，Python！
print("你好，Python！")
```

## 2 输出与输入

### 2.1 输出：print

print 可以接收多个值，值之间用逗号分开。默认用一个空格分隔，输出结束后换行。

sep 指定值之间的分隔文字，end 指定本次输出的结尾；两者都写在调用的括号内。

```python
print("课程", "Python")  # 预期：课程 Python
print("Python", "入门", sep=" / ")  # 预期：Python / 入门

# 第一次输出以空格结尾，两次输出连成一行：学习进度： 开始
print("学习进度：", end=" ")
print("开始")
```

### 2.2 输入：input

input 显示提示，等你输入一行并按 Enter，再返回字符串。即使输入 12，得到的也是文字，不会自动变成数字。

用 = 将输入结果赋给 learner_name，后面通过这个名称使用它。变量与类型转换会在第 2 章“内置类型与运算”展开。

```python
# 示例输入：小林（直接输入文字，不加引号）；预期：你好，小林
learner_name = input("请输入姓名：")
print("你好，", learner_name, sep="")
```

## 3 名称与关键字

变量名属于标识符，也就是代码中使用的名称。命名时注意：

（1）使用英文字母、数字和下划线组合时，第一位不能是数字；例如 course_name、lesson2、_name 合法，2lesson 不合法。

（2）名称区分大小写，course_name 与 Course_name 是不同名称。

（3）if、for、class 等关键字是语法保留名称，不能用作变量名。match、case、type 和 _ 是 Python 3.12 的软关键字，仅在特定语法位置保留。

Python 也允许符合规则的 Unicode 字符作为名称，例如中文。本课程采用小写英文单词与下划线命名。

print、input、help 是内置函数名，不是关键字；避免用它们命名变量，以免遮蔽原有功能。

```python
course_name = "Python"

print(course_name)  # 预期：Python；通过名称读取它对应的值。
print("course_name")  # 预期：course_name；引号内是文字。
```

## 4 代码书写

### 4.1 注释与空格

引号外的 # 开始注释，一直延伸到行末；Python 不把注释作为语句执行。引号里的 # 是文字。

注释用于说明用途或原因。书写时，赋值等号两侧各留一个空格，逗号后留一个空格；print(...) 这样的调用，名称与左括号之间不留空格。

```python
# 输出课程标签，方便辨认。
course_name = "Python"
print("课程：", course_name)  # 预期：课程： Python
print("# Python")  # 预期保留 #，输出：# Python
```

### 4.2 缩进

Python 用行首缩进划分语句块。本课程按 PEP 8 的建议，每层使用 4 个空格，同层对齐，不混用 Tab 和空格。

下面暂用 if True 表示条件成立，执行缩进的语句；回到行首的 print 已经在 if 之外。条件语句会在第 5 章“条件与循环”展开。

```python
# 预期按顺序输出三行；注意前两次 print 的缩进。
if True:
    print("进入代码块")
    print("仍在代码块")
print("离开代码块")
```

### 4.3 括号内换行

长语句可以在圆括号、方括号或花括号内部换行，无需在行末添加反斜杠。下面虽然写成多行，仍是一次 print 调用。

续行是在排版同一条语句，与 if 后表示语句块的缩进不同。

```python
# 预期只输出一行：Python / 基础 / 练习
print(
    "Python",
    "基础",
    "练习",
    sep=" / ",
)
```

忘记函数用法时，可以用 help 查询。help(print) 查看 print 的帮助，里面的 print 后面不加调用括号。

先读功能说明，再找需要的参数。下面重点查看 sep 和 end；完整的参数语法会在第 6 章“函数与参数”介绍。

```python
# 在帮助中找到 sep、end 的默认值，并与第 2.1 节的输出对照。
help(print)
```

## 6 阅读报错

先看报错末尾的错误类型和说明，再看上面的行号及位置标记。语法错误的标记指向发现问题的位置，实际原因也可能在它前面。

| 错误类型 | 本章常见原因 |
| --- | --- |
| SyntaxError | 语法写法不完整，例如 print("Python" 缺少右括号 |
| IndentationError | 缩进不正确，例如 if 后的语句没有缩进 |
| NameError | 找不到名称，例如拼写错误，或还没有执行给它赋值的语句 |

例如把 course_name 错写成 course_nmae，报错末尾会包含：

> NameError: name 'course_nmae' is not defined

下面保留正确写法；对照注释中的反例，检查名称是否一致。异常处理会在第 11 章“异常处理与调试”展开。

```python
course_name = "Python"
# 反例：print(course_nmae) 中的名称拼错，会触发 NameError。
print(course_name)  # 预期：Python
```
