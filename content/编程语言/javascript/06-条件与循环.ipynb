## 1 if 与多分支条件

if 将括号内的表达式转换为布尔值，条件成立时执行对应语句；else 提供未成立的分支。else if 按书写顺序检查条件，一旦命中便跳过后续分支，因此从更具体的条件写起。

用花括号明确各分支范围。比较用 ===，不要把赋值 = 写进本来要比较的位置。下面 score 是 0–100 范围内的已知分数；合法性检查和等级划分是不同的任务。

```javascript
const score = 86;
let grade;
if (score >= 90) {
  grade = "优秀";
} else if (score >= 60) {
  grade = "通过";
} else {
  grade = "重试";
}
console.log(grade);
const input = "";
if (input === "") {
  console.log("需要输入");
}
// 输出依次为：
// 通过
// 需要输入
```

## 2 switch 与贯穿

switch 先计算一次判别值，再按严格相等规则寻找 case。命中后从对应位置往下执行，遇到 break 才退出整个 switch；若没有 break，会继续执行后续 case 的语句，称为贯穿（fall-through）。default 处理没有匹配项的情况。

把多个 case 放在同一组语句之前，可让多个输入共享处理。case 本身不会创建块作用域；各分支需要声明同名 let 或 const 时，用额外花括号隔开。

```javascript
// 第一组让 read 与 preview 共用动作；write 用独立块限定局部变量。
const command = "preview";
let action;
switch (command) {
  case "read":
  case "preview":
    action = "只读";
    break;
  case "write": {
    const permission = "已授权";
    action = permission;
    break;
  }
  default:
    action = "未知命令";
}
console.log(action);
// 第二组故意贯穿；第三组对比字符串与数值是否能匹配。
let trail = "";
switch (1) {
  case 1:
    trail += "A"; // 这里有意不写 break
  case 2:
    trail += "B";
    break;
}
console.log(trail);
switch ("1") {
  case 1:
    console.log("数值");
    break;
  default:
    console.log("字符串没有匹配数值");
}
// 输出依次为：
// 只读
// AB
// 字符串没有匹配数值
```

## 3 for 的三个阶段

把圆括号里的三个部分按运行顺序展开，可以看出更新发生在循环体之后。

for 的圆括号依次是初始化、继续条件、每轮更新，以分号分隔。初始化只做一次；先检查条件，再执行循环体，然后更新，再次判断。条件省略时视为持续成立，必须有明确的退出方式。

本例 i 是从 0 开始的索引，text.length 是上界，使用 i < text.length 避免多读一次。let 声明只在循环范围可见；经典递增计数不能用 const，因为更新会重新赋值。

![for：先判断，再执行，再更新。以 text = "JS" 为例；箭头表示执行先后。](image/illustration/06-01-for-control-flow.svg)

图示说明：依据本章末所列语言规则自行绘制。图展示正常迭代；break、continue 的改变在本章后文分别说明。

用图跟踪下面第一个循环的 i，再观察第三个循环如何直接走 false 分支，解释为什么它执行零次。

```javascript
const text = "JS";
let joined = "";
for (let i = 0; i < text.length; i += 1) {
  joined += text[i];
}
console.log(joined);
let sum = 0;
for (let value = 1; value <= 4; value += 1) {
  sum += value;
}
console.log(sum);
let emptyRuns = 0;
for (let i = 0; i < 0; i += 1) {
  emptyRuns += 1;
}
console.log(emptyRuns);
// 输出依次为：
// JS
// 10
// 0
```

## 4 while 与 do...while

while 在每轮开始前判断，条件最初为假时执行零次；do...while 先执行循环体，再判断，因此至少执行一次。两种方式都要让状态朝结束条件推进，否则可能成为无限循环。

下面 remaining 是剩余处理次数。循环体修改它，从 3 逐步减到 0。do...while 示例故意用 false 说明“先做一次”的语义，并不代表需要任意等待或重试。

```javascript
let remaining = 3;
let processed = 0;
while (remaining > 0) {
  remaining -= 1;
  processed += 1;
}
console.log(remaining, processed);
let attempts = 0;
do {
  attempts += 1;
} while (false);
console.log(attempts);
// 输出依次为：
// 0 3
// 1
```

## 5 break 与 continue

break 结束最近的循环或 switch；continue 跳过本轮后续语句，开始下一轮。经典 for 中 continue 后仍会执行更新表达式；while 中则重新检查条件，所以把唯一的更新写在 continue 后面会阻止循环推进。

下面跳过偶数，并在到达 7 时结束。把停止条件放在跳过条件之前，可以明确它不受筛选影响。

```javascript
let selected = "";
for (let value = 1; value <= 9; value += 1) {
  if (value === 7) break;
  if (value % 2 === 0) continue;
  selected += value;
}
console.log(selected);
let cursor = 0;
let kept = 0;
while (cursor < 4) {
  cursor += 1; // 更新先于可能发生的 continue
  if (cursor === 2) continue;
  kept += 1;
}
console.log(cursor, kept);
// 输出依次为：
// 135
// 4 3
```

## 6 标签语句控制外层

标签以“名称:”给语句命名。break label 可以离开包含它的带标签语句；continue label 只能指向包含它的循环。label 在这里代表实际标签名称，例如 search 和 rows，不是变量，也不是任意跳转的 goto。

标签不能跨越函数边界。嵌套循环中确需退出外层时可以使用；更复杂的流程通常适合提取为函数并用 return 返回结果。

```javascript
// break search 同时离开两层循环，保存找到的第一个行列位置。
let found = "无";
search: for (let row = 0; row < 3; row += 1) {
  for (let col = 0; col < 3; col += 1) {
    if (row + col === 3) {
      found = `${row},${col}`;
      break search;
    }
  }
}
console.log(found);
// continue rows 跳过本行剩余列，直接开始下一行。
let cells = "";
rows: for (let row = 0; row < 2; row += 1) {
  for (let col = 0; col < 3; col += 1) {
    if (col === 1) continue rows;
    cells += `${row}${col} `;
  }
}
console.log(cells.trim());
// 标签也能命名普通块；break finish 离开该块。
let marker = "开始";
finish: {
  marker = "结束";
  break finish;
  marker = "不会到达";
}
console.log(marker);
// 输出依次为：
// 1,2
// 00 10
// 结束
```

## 7 for...in 遍历键，for...of 遍历值

for...in 枚举对象的可枚举字符串属性键，包括继承属性，不包含 Symbol 键。需要自有属性时用 Object.hasOwn() 筛选。下面 Object.create() 仅用于建立一个带继承属性的对象，原型机制在对象及原型章节展开。

for...of 从可迭代对象取得值，例如数组元素和字符串码点；普通对象默认不可迭代。数组也是对象，但把它当值序列时优先用 for...of，避免读到额外属性。这里 push() 向数组收集数据，join() 把数组连接为便于显示的字符串，数组方法在数组章节系统讲解。

for...of 每轮的 const 是新绑定，并非不断重新赋值给同一个 const。遍历期间修改集合会影响某些访问行为；除非明确了解相应规则，本章保持遍历对象不变。

```javascript
const settings = Object.create({ inherited: "来自原型" });
settings.own = "自身";
const allKeys = [];
const ownKeys = [];
for (const key in settings) {
  allKeys.push(key);
  if (Object.hasOwn(settings, key)) ownKeys.push(key);
}
console.log(allKeys.join(","), ownKeys.join(","));
const lessons = ["JS", "TS"];
lessons.note = "额外属性";
const keys = [];
const values = [];
for (const key in lessons) keys.push(key);
for (const value of lessons) values.push(value);
console.log(keys.join(","), values.join(","));
let symbols = 0;
for (const value of "A🚀") symbols += 1;
console.log(symbols);
// 输出依次为：
// own,inherited own
// 0,1,note JS,TS
// 2
```

## 练习

1. 将等级分界改为 85、60，分别输入 59、60、84、85；核对只在边界处分档，且每次只选择一个等级。
2. 遍历 1–10，只累加奇数；核对总和为 25，再把上界改为 0，核对总和为 0。
3. 将二维搜索条件改为 row + col === 4；核对返回 2,2，并确认找到后没有再处理其他位置。
4. 对含继承属性的对象分别收集所有可枚举键与自有可枚举键；核对后者不包含 inherited。修复对象不可迭代反例时可用 for...in 配合 Object.hasOwn()。

### 提示

区分“本轮不计入”与“整个搜索已完成”；先确定循环边界，再决定 continue 或带标签的 break。

### 参考解析

新的四个分数依次为重试、通过、通过、优秀。1–10 中奇数和为 25；上界为 0 时循环一次也不执行，所以和保持 0。二维搜索仍按行优先扫描，目标 4 首次出现于 2,2，break search 结束两层循环。for...in 包括继承键，Object.hasOwn 的筛选才得到自有键。
