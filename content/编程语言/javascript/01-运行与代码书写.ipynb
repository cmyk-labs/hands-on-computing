## 1 运行第一段代码

console.log() 把信息写到控制台；圆括号中是要输出的值，双引号包围的是文本。console.error() 用于输出错误信息，本身不等于抛出异常。

```javascript
console.log("你好，JavaScript"); // 你好，JavaScript
console.log(1 + 2); // 3
console.error("提示：请先保存文件"); // Node.js 中写入标准错误流；这里没有抛出异常
```

Step 1：执行整个 [basics.mjs](scripts/01-running-and-writing/basics.mjs)，会依次看到本章几个示例的输出：

```bash
node scripts/01-running-and-writing/basics.mjs
```

只试一行表达式时，在终端输入 node，进入交互式环境 REPL，再输入 1 + 2；输入 .exit 退出。浏览器开发者工具的 Console 也能输入 1 + 2，查看计算结果。

## 2 语言、引擎与宿主环境

- JavaScript 是编程语言；ECMAScript 标准规定其核心语法与行为。
- 引擎负责执行代码，例如 V8；宿主环境提供代码运行所需的其他能力，例如浏览器和 Node.js。
- console 由宿主提供。浏览器页面中的 document 表示文档，Node.js 默认没有这个浏览器对象。

globalThis 提供跨环境访问全局对象的统一方式，但不会让不同宿主拥有相同的 API。下面的 === 比较两边是否相等，typeof 查看值的类型标签。

```javascript
console.log(globalThis.console === console); // true：这里访问的是同一个 console
console.log(typeof document); // Node.js：undefined；本章浏览器页面：object
```

## 3 名称、注释与语句

名称区分大小写。标识符可使用符合规则的 Unicode 字符，以及 $ 和下划线；数字不能作为首字符，保留字不能随意用作变量名。普通变量常用 unitPrice 这样的 camelCase 写法，这属于命名约定。

表达式产生一个值，例如单价乘数量；语句执行一项操作，例如声明变量或输出结果。// 注释到行末，/* 与 */ 包围多行注释。

```javascript
// const 为值声明一个名称；变量的完整规则见“类型与变量”。
const unitPrice = 12;
const quantity = 2;
/* 两个名称分别表示单价和数量，* 表示乘法。 */
console.log(unitPrice * quantity); // 24
```

## 4 分号与换行

自动分号插入（automatic semicolon insertion，ASI）让某些位置可以省略分号，但换行不一定结束语句。入门时在声明和表达式语句末尾显式写分号，可以更容易看清边界。

```javascript
// 这里省略分号，三个语句仍可以分别结束。
const price = 12
const amount = 2
console.log(price * amount) // 24

// 换行后仍可接着做加法，因此这两行属于同一个初始化表达式。
const total = 1
+ 2;
console.log(total); // 3
```

## 5 普通脚本与模块

本章用明确的文件后缀或加载方式区分运行模式：

- Node.js 的 .mjs 文件按 ES 模块运行；.cjs 文件按 CommonJS 模块运行。
- Node.js 中 .js 的解释还受 package.json 等条件影响，不能只看后缀判断。
- 浏览器用 &lt;script&gt; 加载普通脚本，用带有 type="module" 的 &lt;script&gt; 加载 ES 模块。CommonJS 是 Node.js 的另一套模块系统，不是这里所说的浏览器普通脚本。

完整页面见 [index.html](scripts/01-running-and-writing/index.html)。以下 src 相对于这个 HTML 文件，而不是相对于 Notebook：

```html
<script src="classic.js"></script>
<script type="module" src="basics.mjs"></script>
```

## 6 严格模式

严格模式（strict mode）会把某些容易出错的写法变为错误，例如给未声明的名称赋值。

- ES 模块自动启用严格模式，不需要再加指令。
- 普通脚本或 CommonJS 文件可以在开头写 "use strict";。指令要位于其他普通语句之前。

配套 [mode.cjs](scripts/01-running-and-writing/mode.cjs)：

```javascript
"use strict";
console.log("CommonJS 文件已运行"); // CommonJS 文件已运行
```

Step 1：运行 CommonJS 示例。

```bash
node scripts/01-running-and-writing/mode.cjs
```

严格模式仍然运行 JavaScript，不会增加 TypeScript 那样的静态类型检查。

## 7 阅读错误

先找错误名称，再看首个指向自己文件的位置。行号和列号能定位问题附近，但语法错误的真正原因也可能在前一行。

- SyntaxError：代码不符合语法，文件不能正常解析。
- ReferenceError：引用出了问题，例如严格模式下使用未声明的名称。
- TypeError：操作不适用于当前值，例如把数值当函数调用。

三个文件需要分别执行；每次预期都以非零状态退出，完整错误措辞可能随环境变化。

## 本章小结

- 语言标准规定核心规则，宿主提供 console、document 等运行能力。
- 先确认运行模式，再理解严格模式；换行不能总当作语句结束。
