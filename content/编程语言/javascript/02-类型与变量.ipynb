## 1 类型总览

ECMAScript 有七种原始类型和 Object 类型。下表采用常见的 JavaScript 类型写法；数组和函数都属于对象，不是与这八类并列的新类型。

| 类型名称 | 中文名称／含义 | 值或创建示例 |
| --- | --- | --- |
| undefined | 未定义值类型 | undefined |
| null | 空值类型 | null |
| boolean | 布尔类型 | true、false |
| number | 数值类型 | 12、12.5 |
| bigint | 大整数类型 | 12n |
| string | 字符串类型 | "课程" |
| symbol | 符号类型 | Symbol("id") |
| object | 对象类型 | { title: "课程" } |

字面量是在代码中直接写出的值，例如 12、12n、"课程"、true、null，以及对象字面量。undefined 是全局属性名称；Symbol("id") 是函数调用，都不属于字面量。Symbol() 每次调用创建独一无二的符号，"id" 只是描述。

下面用 typeof 查看类型标签，它返回字符串：

```javascript
console.log(typeof undefined); // undefined
console.log(typeof null); // object：历史遗留结果，null 仍是原始值
console.log(typeof true); // boolean
console.log(typeof 12.5); // number
console.log(typeof 12n); // bigint
console.log(typeof "课程"); // string
console.log(typeof Symbol("id")); // symbol
console.log(typeof { title: "课程" }); // object
```

## 2 typeof 的边界

typeof 的结果不是完整的类型分类表：null 得到 "object"，函数得到 "function"，数组则得到 "object"。识别数组时可用 Array.isArray()。

```javascript
console.log(typeof []); // object：数组属于对象
console.log(Array.isArray([])); // true：专门判断是否为数组
console.log(typeof console.log); // function：可调用对象的特殊标签
```

## 3 用 let 与 const 声明变量

变量把名称与值联系起来。let 允许重新赋值；const 不允许重新赋值。普通 const 声明必须带初始值。= 是赋值，不是比较。

需要改变绑定时用 let，其余通常优先用 const。尚未初始化为具体值的 let 变量，其值为 undefined；null 则通常由代码显式赋给变量来表示“这里没有值”。

```javascript
let progress;
console.log(progress); // undefined：已声明，但没有给定初始值
progress = 1;
progress = 2;
const courseName = "JavaScript";
console.log(progress, courseName); // 2 JavaScript
```

## 4 var 与块作用域

这里的一对花括号包围多条语句，构成一个块。let 和 const 的作用域包括块；var 不受普通块限制，它通常归属所在函数或顶层环境。

同一作用域不能重复声明同一个 let 或 const 名称；内层块可以声明自己的同名变量，这称为遮蔽（shadowing）。var 可以重复声明已有的 var 名称，但不能因此与同一作用域的 let 或 const 冲突。

```javascript
let status = "外层";
{
  let status = "内层";
  var legacyCount = 1;
  console.log(status); // 内层：访问当前块中的名称
}
console.log(status); // 外层：是另一个变量
console.log(legacyCount); // 1：var 不以这个块作为作用域边界

var retryCount = 1;
var retryCount = 2; // 同一作用域中可以再次用 var 声明这个 var 名称
console.log(retryCount); // 2
```

当前文件是 ES 模块，顶层 var 也不会因此变成全局对象的属性。新代码通常使用 let 和 const；var 的更多作用域行为在后续章节展开。

## 5 动态类型

JavaScript 是动态类型语言：类型属于值，变量没有固定的值类型。let 可以先保存数值，再保存字符串；这不会改变原先那个数值的类型。

```javascript
let result = 20;
console.log(typeof result); // number
result = "完成";
console.log(typeof result); // string：同一个变量可以改为保存另一种类型的值
```

## 6 原始值不可变

原始值本身不可修改。字符串方法可以返回新字符串；把新字符串赋给变量，是改变变量保存的值，不是修改旧字符串。下面 toUpperCase() 返回大写字符串。

```javascript
let originalText = "hello";
let copiedText = originalText;
copiedText = copiedText.toUpperCase(); // 返回转换后的新字符串
console.log(originalText, copiedText); // hello HELLO
```

## 7 对象引用与 const

对象可以具有可修改的属性。把一个对象赋给另一个变量，不会自动复制对象；两个变量会引用同一个对象。下面 score 是对象的分数属性，点号用于访问该属性。

const 固定变量的绑定，不会自动冻结对象。=== 比较两个对象时，判断是否为同一个对象，而不是逐个比较属性。

```javascript
const report = { score: 80 };
const alias = report; // 两个变量引用同一个对象
alias.score = 90;
console.log(report.score); // 90：通过另一名称修改了同一个对象
console.log(report === alias); // true：比较的是对象身份
console.log(report === { score: 90 }); // false：这是另一个新对象
```

## 8 声明之前能否访问

let 和 const 从进入作用域到完成初始化之间存在暂时性死区（temporal dead zone，TDZ），这期间访问名称会抛出 ReferenceError，使用 typeof 也一样。

var 的声明会被提前处理，但初始化赋值仍发生在原来的位置。实际代码应先声明再使用，不依赖这种差异写程序。

```javascript
console.log(earlyCount); // undefined：var 声明已建立，赋值尚未执行
var earlyCount = 3;
console.log(earlyCount); // 3
console.log(typeof notDeclaredAnywhere); // undefined：这个名称没有声明
```

typeof 对完全未声明的名称有特殊处理；它不代表能绕过 let 和 const 的暂时性死区。

## 本章小结

- 七种原始类型的值不可变；数组、函数和普通对象都属于对象。
- let 允许重新赋值，const 固定绑定；它们都具有块作用域。
- 对象赋值不会自动复制对象；typeof 也不能独自区分所有常见值。
