## 1 相等、近似与定义

计算 $2+5$ 得到 $7$；把某个数保留两位小数，得到的通常是近似值；给一个表达式取名字，则是在作定义。阅读公式时先区分这三件事。

下表中，$a$、$b$ 是实数，$s$ 是新定义的名称。

| 写法 | 中文名称／含义 | 例子 |
| --- | --- | --- |
| $a=b$ | 相等：两边表示同一个值 | $2+5=7$ |
| $a\approx b$ | 近似相等：保留精度后接近 | $1/7\approx0.14$ |
| $s\overset{\mathrm{def}}{=}a+b$ | 定义：规定 $s$ 表示右边的量 | 把两数之和命名为 $s$ |

Python 的 = 用于赋值，== 用于相等比较。数学等式与浮点计算也要分开：有限精度的 float 可能不能精确保存十进制小数。math.isclose 按容差判断接近，不证明数学上相等；下面只允许 $10^{-12}$ 的绝对差值，适用于这个量级的小数核对。

```python
import math

a = 2
b = 5
s = a + b
print(s, s == 7)  # 预期：7 True；赋值后再比较。

computed = 0.1 + 0.2
print(computed, computed == 0.3)  # 预期：0.30000000000000004 False
print(math.isclose(computed, 0.3, rel_tol=0.0, abs_tol=1e-12))  # True
# abs_tol 是允许的绝对差值；rel_tol=0.0 关闭相对容差。
```

## 2 集合与区间

### 2.1 元素与集合运算

用 $A=\{1,3,5\}$ 表示由这三个数构成的集合。$3\in A$ 表示 3 属于 $A$；$2\notin A$ 表示 2 不属于 $A$。集合不按位置区分元素，也不因重复列出同一个元素而改变。

设 $B=\{3,4\}$。交集 $A\cap B$ 只保留共有元素；并集 $A\cup B$ 保留至少属于一个集合的元素；差集 $A\setminus B$ 保留属于 $A$ 但不属于 $B$ 的元素。$\varnothing$ 表示空集。

Python 的 set 可表示这里的有限集合，sorted 只用于让输出按数值排序。

```python
set_a = {1, 3, 5, 3}  # noqa: B033 — 此处有意演示集合去重。
set_b = {3, 4}
print(sorted(set_a))  # 预期：[1, 3, 5]；重复的 3 不增加元素。
print(3 in set_a, 2 not in set_a)  # 预期：True True
print(sorted(set_a & set_b))  # 交集：[3]
print(sorted(set_a | set_b))  # 并集：[1, 3, 4, 5]
print(sorted(set_a - set_b))  # 差集：[1, 5]
print(set_a & {2, 4})  # 空交集显示为 set()。
```

### 2.2 用区间表达取值范围

$\mathbb{R}$ 表示实数集合。区间描述连续的一段实数；例如 $[1,4)$ 包含 1，排除 4。下面 $x$ 是待判断的实数，方括号包含端点，圆括号排除端点。

| 写法 | 中文名称／含义 | 条件 |
| --- | --- | --- |
| $[1,4]$ | 闭区间 | $1\le x\le4$ |
| $(1,4)$ | 开区间 | $1<x<4$ |
| $[1,4)$ | 左闭右开区间 | $1\le x<4$ |
| $(1,4]$ | 左开右闭区间 | $1<x\le4$ |
| $(0,\infty)$ | 正实数范围 | $x>0$ |

$\infty$ 表示无上界，不是可以包含的实数端点。区间 $[1,4)$ 不只是整数 1、2、3，也包含 1.5 等实数；代码检查若干输入，不是在枚举整个区间。

```python
for x in [0, 1, 1.5, 3, 4]:
    print(x, 1 <= x < 4)  # 按 x=0、1、1.5、3、4 的顺序，判断依次为 False、True、True、True、False。
# 预期：1、1.5、3 对应 True；0 和 4 对应 False。
```

## 3 索引与有限求和

### 3.1 把下标对应到列表位置

三个数 $x_1=4$、$x_2=1$、$x_3=6$ 中，下标 $i$ 表示第几项。这里 $i$ 是取 1、2、3 的整数，$x_i$ 不是 $x$ 与 $i$ 相乘。

数学下标可自行约定起点；本例从 1 开始，而 Python 列表从 0 开始。因此 $x_i$ 对应 values[i - 1]。先写清这个对应，再把公式转为代码。

```python
values = [4, 1, 6]

for i in range(1, 4):
    print(i, values[i - 1])  # 数学下标 1、2、3 对应列表中的 4、1、6。
# 三行依次为 1 4、2 1、3 6；range(1, 4) 不包含右端的 4。
```

### 3.2 展开求和再核对

对 $n$ 个数 $x_1,\ldots,x_n$，$n$ 是正整数，有限求和记作

$$\sum_{i=1}^{n}x_i=x_1+x_2+\cdots+x_n.$$

$\sum$ 是求和符号；$i$ 是求和索引，依次取从 1 到 $n$ 的所有整数，两端都包含。对上面的三个数，有 $\sum_{i=1}^{3}i x_i=1\times4+2\times1+3\times6=24$。

把索引换成 $j$，同时替换式子中相应的下标和因子，和不变。Python 的 sum 累加传入的各项；先列出每项，可以检查是否遗漏端点或错用列表位置。

```python
values = [4, 1, 6]
n = len(values)
terms = [i * values[i - 1] for i in range(1, n + 1)]

print(terms)  # 预期：[4, 2, 18]；逐项对应 i * x_i。
print(sum(terms))  # 预期：24
print(sum(j * values[j - 1] for j in range(1, n + 1)))  # 仍为 24。
```

## 4 连乘与空输入

若 $a_1,\ldots,a_n$ 是 $n$ 个实数，$n$ 为正整数，则

$$\prod_{i=1}^{n}a_i=a_1a_2\cdots a_n.$$

$\prod$ 是连乘符号，索引 $i$ 的读法与求和相同。例如 $\prod_{i=1}^{3}(i+1)=2\times3\times4=24$，这里有三个因子。

当索引集合为空时，约定空和为 0，空积为 1，分别对应加法和乘法中不改变其他数的值。Python 的 sum([]) 与 math.prod([]) 在默认起始值下遵循这一约定。

```python
factors = [i + 1 for i in range(1, 4)]
print(factors, math.prod(factors))  # 预期：[2, 3, 4] 24
print(sum([]), math.prod([]))  # 预期：0 1；空输入没有提供加数或因子。
```

## 5 函数的输入与定义域

### 5.1 输入一个值，得到唯一输出

把输入 3 先平方再加 2，输出为 11。将这个规则写成 $f(x)=x^2+2$，其中 $x$ 是实数输入，$f$ 是函数名，$f(x)$ 是对应输出，括号不表示乘法。

函数给定义域中的每个输入指定唯一输出；定义域是允许输入的集合，值域是实际能够输出的集合。本例定义域为 $\mathbb{R}$，值域为 $[2,\infty)$。不同输入可以产生相同输出，例如 $f(-3)=f(3)$。

```python
def square_then_add_two(x):
    return x**2 + 2


print(square_then_add_two(3))  # 预期：11
print(square_then_add_two(-3))  # 预期：11；输出唯一不要求输入唯一。
print(square_then_add_two(0))  # 预期：2；这是本例的最小输出。
```

### 5.2 先检查公式允许哪些输入

定义域不能只看变量名。实数平方根要求根号内非负，除法要求分母不为零。例如

$$q(x)=\frac{\sqrt{x+1}}{x-2}$$

中，$x$ 是实数输入，$q(x)$ 是输出。两项条件为 $x\ge-1$ 和 $x\ne2$，所以定义域是 $[-1,2)\cup(2,\infty)$。下面直接选择定义域内的输入，用 math.sqrt 求平方根；定义域外的值没有本式的实数输出。

```python
for x in [-1, 0, 3]:
    print(x, math.sqrt(x + 1) / (x - 2))  # 预期依次约为 0、-0.5、2。
# 输入均满足定义域；-0.0 与 0 数值相等，这是浮点数的带符号零。
```

## 6 复合函数的顺序与条件

令 $g(x)=x+1$、$f(x)=x^2$，两者的输入都是实数。复合 $(f\circ g)(x)$ 表示 $f(g(x))$：先算里面的 $g$，再把输出交给 $f$。

$$f(g(x))=(x+1)^2,\qquad g(f(x))=x^2+1.$$

从括号读公式时先内后外，从图中追踪数值时顺着箭头前进。对同一个输入 2，先预测中间量分别是什么。

![输入 2 先加 1 后平方得到 9，先平方后加 1 得到 5；箭头标出中间量 3 和 4。](image/illustration/01-01-function-composition.svg)

图中只展开本例两条计算路径，两个函数都能接受每一步的实数结果。一般复合还需要内层输出落在外层定义域中；下面先用两个短函数核对顺序不同造成的结果差异。

```python
def add_one(x):
    return x + 1


def square(x):
    return x**2


print(square(add_one(2)))  # 预期：9；先加 1，再平方。
print(add_one(square(2)))  # 预期：5；先平方，再加 1。
```

复合还必须同时满足两层条件：原输入在内层函数的定义域中，内层输出在外层函数的定义域中。取 $r(x)=x-4$ 和 $h(u)=\sqrt{u}$，其中 $x$ 是实数，外层输入 $u\ge0$，则 $h(r(x))=\sqrt{x-4}$ 只允许 $x\ge4$。

```python
for x in [4, 8]:
    inner_value = x - 4
    print(x, math.sqrt(inner_value))  # 预期两行：4 0.0、8 2.0。
# 内层输出依次为 0、4，均属于外层平方根的定义域。
```

## 7 分段函数与端点

一个函数可以在定义域的不同部分采用不同公式。下面自定义一个规则：负输入平方，非负输入加 1。$x$ 是实数，$p(x)$ 是输出。

$$p(x)=\begin{cases}x^2,&x<0,\\x+1,&x\ge0.\end{cases}$$

计算前先选条件，再使用该段公式。这里两段覆盖全部实数，0 归入第二段；如果漏掉端点，就没有为声明的定义域提供完整规则。

```python
def piecewise_value(x):
    if x < 0:
        return x**2
    return x + 1


for x in [-2, -0.1, 0, 0.1, 2]:
    print(x, piecewise_value(x))  # 按给定输入顺序，函数值依次约为 4、0.01、1、1.1、3。
# 重点检查 0：输出 1，而不是把 x 代入第一段得到的 0。
# 两侧输入 -0.1、0.1 的输出分别约为 0.01、1.1。
```

## 8 指数与对数

### 8.1 指数表示乘法尺度

$2^3=2\times2\times2=8$。在指数函数 $b^x$ 中，$b>0$ 是固定底数，$x$ 是实数输入；通常取 $b\ne1$，因为底数 1 只得到常数 1。$b>1$ 时随 $x$ 增大而增大，$0<b<1$ 时随 $x$ 增大而减小，输出始终为正。

自然常数 $e\approx2.71828$ 常用作底数，$\exp(x)$ 表示 $e^x$。Python 用 ** 表示乘方，用 math.exp 计算自然指数。

```python
for x in [-2, 0, 2]:
    print(x, 2**x, 0.5**x)  # x=-2、0、2 时，两列依次为 (0.25, 4)、(1, 1)、(4, 0.25)。
# 第一组指数值为 0.25、1、4；第二组为 4.0、1.0、0.25。
print(math.exp(0), math.exp(1))  # 预期：1.0，以及约 2.71828。
```

### 8.2 对数反求指数

已知 $2^3=8$，反过来可写 $\log_2 8=3$。一般地，对于底数 $b>0$ 且 $b\ne1$、输入 $u>0$，

$$y=\log_b u\quad\Longleftrightarrow\quad b^y=u.$$

$y$ 是所求的实数指数，$\Longleftrightarrow$ 表示两种写法等价。对数的定义域为 $(0,\infty)$，不能在实数范围直接计算零或负数的对数。

自然对数 $\ln u$ 以 $e$ 为底。math.log(u) 对应自然对数，math.log2(u) 和 math.log10(u) 分别使用底数 2 和 10；看到省略底数的 log，应核对上下文或 API 约定。

```python
print(math.log2(8), math.log10(100))  # 预期：3.0 2.0

for u in [1, 4]:
    restored = math.exp(math.log(u))
    print(u, math.isclose(restored, u, rel_tol=0.0, abs_tol=1e-12))
    # 预期：1 True、4 True；正输入在 1e-12 绝对差值内还原。
```

### 8.3 对数把乘积转成和

对正实数 $u$、$v$，有 $\ln(uv)=\ln u+\ln v$。可令 $a=\ln u$、$b=\ln v$，得到 $u=e^a$、$v=e^b$，于是

$$uv=e^{a+b},\qquad \ln(uv)=a+b=\ln u+\ln v.$$

这里 $a$、$b$ 是实数指数。两因子都为正保证右边的两个对数有定义；不能把该规则误写为 $\ln(u+v)=\ln u+\ln v$。

```python
u = 2.0
v = 6.0
left = math.log(u * v)
right = math.log(u) + math.log(v)
print(math.isclose(left, right, rel_tol=0.0, abs_tol=1e-12))  # 预期：True
print(math.log(u + v), right)  # 约为 2.07944、2.48491，说明加法不能照搬。
# 这里核对两个具体输入；一般恒等关系来自上面的代数推导。
```

## 9 综合应用：按定义域筛选后求和

自制五个输入 $x_1=-2$、$x_2=-1$、$x_3=0$、$x_4=2$、$x_5=3$，继续使用第 5.2 节的 $q(x)=\sqrt{x+1}/(x-2)$。先排除定义域外的输入，再累加有效输出。

用 $I$ 表示有效输入的索引集合。本例 $I=\{2,3,5\}$，所以

$$S=\sum_{i\in I}q(x_i)=q(-1)+q(0)+q(3)=0-0.5+2=1.5.$$

这里 $S$ 是筛选后的输出总和；$i\in I$ 表示只对指定索引求和，不要求这些索引连续。这个任务自行规定了排除无效输入，不把它们的函数值定义为零。

```python
inputs = [-2, -1, 0, 2, 3]
valid_indices = []
outputs = []
# 数学下标从 1 开始，读取列表时用 i - 1；只累加定义域内的函数值。
for i in range(1, len(inputs) + 1):
    x = inputs[i - 1]
    if x >= -1 and x != 2:
        valid_indices.append(i)
        outputs.append(math.sqrt(x + 1) / (x - 2))

print(valid_indices)  # 预期：[2, 3, 5]；保留原输入的数学下标。
print(outputs)  # 预期：[-0.0, -0.5, 2.0]；逐项核对后再累加。
print(sum(outputs))  # 预期：1.5
```
